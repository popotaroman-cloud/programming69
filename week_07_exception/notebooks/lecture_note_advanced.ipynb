## 10: ข้อผิดพลาดที่ผู้ใช้กำหนดเอง (Custom Exception)

**แหล่งอ้างอิง**: TutorialsPoint — [Python User-Defined Exceptions](https://www.tutorialspoint.com/python/python_userdefined_exception.htm)

### แนวคิด

นอกจากข้อผิดพลาดขณะรัน (exception) ที่มากับภาษาโปรแกรมมิ่งโดยตรง ผู้เขียนโปรแกรมยังสามารถกำหนดชนิดข้อผิดพลาดของตนเองขึ้นมาใหม่ได้
เพื่อสื่อความหมายเฉพาะเจาะจงกับปัญหาในโดเมนของโปรแกรมนั้น ๆ เช่น "จำนวนเงินไม่พอถอน" หรือ "หนังสือไม่พร้อมให้ยืม" ซึ่งไม่มีชนิดข้อผิดพลาด
มาตรฐานใดสื่อความหมายได้ตรงเท่ากับชื่อที่ตั้งขึ้นเอง แนวคิดนี้มีอยู่ในภาษาโปรแกรมเชิงวัตถุ (object-oriented) หลายภาษา โดยทั่วไปทำได้ด้วยการ
สร้างคลาสใหม่ที่สืบทอด (inherit) คุณสมบัติมาจากคลาสข้อผิดพลาดพื้นฐานของภาษานั้น เพื่อให้ข้อผิดพลาดที่ผู้ใช้กำหนดเองยังคงทำงานร่วมกับกลไก
บล็อกทดลองรัน/บล็อกดักข้อผิดพลาด (try/except) ของภาษาได้ตามปกติทุกประการ

ในภาษา Python การสร้างข้อผิดพลาดที่ผู้ใช้กำหนดเอง (custom exception) ทำได้โดยสร้างคลาสใหม่ที่สืบทอดจากคลาส `Exception`
(หรือคลาสย่อยของ `Exception` — ดูหัวข้อ 11 เรื่องลำดับชั้นของข้อผิดพลาด) คลาสใหม่นี้จะได้รับพฤติกรรมพื้นฐานของข้อผิดพลาดมาโดยอัตโนมัติ
เช่น สามารถโยนข้อผิดพลาด (raise) และดักข้อผิดพลาด (except) ได้เหมือนข้อผิดพลาดมาตรฐาน อาจเพิ่ม `__init__` เพื่อเก็บข้อมูลเฉพาะของ
ข้อผิดพลาดนั้น (เช่น ชื่อหนังสือที่ยืมไม่ได้ หรือยอดเงินที่ขาด) ก็ได้ ต่อยอดจากหัวข้อ 9 (`raise`) และหัวข้อ 8 (`except X as e`)
ใน `lecture_note_intermediate.md`

### ตัวอย่างที่ 1: การสร้างข้อผิดพลาดที่ผู้ใช้กำหนดเองแบบพื้นฐาน (พื้นฐานของแนวคิด)

````python
%reset -f
class InvalidGradeError(Exception):
    pass

def check_grade(score):
    if score < 0 or score > 100:
        raise InvalidGradeError("คะแนนต้องอยู่ระหว่าง 0 ถึง 100")
    return score

try:
    check_grade(120)
except InvalidGradeError as e:
    print("พบข้อผิดพลาด:", e)

print(check_grade(85))
````

**Output:**
```
พบข้อผิดพลาด: คะแนนต้องอยู่ระหว่าง 0 ถึง 100
85
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | คำอธิบายเพิ่มเติม |
|---|---|---|---|
| 1–2 | นิยามคลาส `InvalidGradeError` | `InvalidGradeError` = คลาสใหม่ | สืบทอดจาก `Exception` โดยตรง ไม่เพิ่มพฤติกรรมใด ๆ (`pass`) |
| 4–7 | นิยามฟังก์ชัน `check_grade` | `check_grade` = function object | ตรวจเงื่อนไขคะแนนก่อนคืนค่า |
| 10 | `check_grade(120)` | เงื่อนไขบรรทัด 5 เป็นจริง | โยนข้อผิดพลาด `InvalidGradeError` ที่บรรทัด 6 |
| 11–12 | `except InvalidGradeError as e` | `e` = อ็อบเจกต์ข้อผิดพลาดที่โยนมา | บล็อกดักข้อผิดพลาดจับได้ตรงชนิด แสดงข้อความผ่าน `print("พบข้อผิดพลาด:", e)` |
| 14 | `check_grade(85)` | เงื่อนไขบรรทัด 5 เป็นเท็จ | คืนค่า 85 ตามปกติ ไม่มีข้อผิดพลาดเกิดขึ้น |

**คำอธิบายภาพรวม:** บรรทัด 1–2 สร้างชนิดข้อผิดพลาดใหม่ชื่อ `InvalidGradeError` โดยสืบทอดจาก `Exception` ซึ่งเป็นวิธีขั้นต่ำที่สุด
ในการสร้างข้อผิดพลาดที่ผู้ใช้กำหนดเอง บรรทัด 4–7 ใช้ `InvalidGradeError` นี้ในการโยนข้อผิดพลาดเมื่อคะแนนอยู่นอกช่วงที่กำหนด
บรรทัด 9–12 สาธิตการดักข้อผิดพลาดชนิดนี้ด้วย `except InvalidGradeError as e` ซึ่งทำงานเหมือนกับการดักข้อผิดพลาดมาตรฐานทุกประการ
บรรทัด 14 แสดงกรณีที่ไม่มีข้อผิดพลาดเกิดขึ้น ฟังก์ชันคืนค่าตามปกติ

**จุดสังเกต:** แม้ `InvalidGradeError` จะไม่มีการเขียน `__init__` ของตัวเองเลย (มีเพียง `pass`) แต่ยังสามารถส่งข้อความ
`"คะแนนต้องอยู่ระหว่าง 0 ถึง 100"` เข้าไปตอนสร้างอ็อบเจกต์และให้ `print(e)` แสดงข้อความนั้นออกมาได้ถูกต้อง เพราะคลาส `Exception`
ที่เป็นคลาสฐานมี `__init__` และ `__str__` มาตรฐานรองรับการรับข้อความไว้ให้แล้ว ผู้เขียนโปรแกรมจึงไม่จำเป็นต้องเขียน `__init__` เองเสมอไป
หากข้อผิดพลาดนั้นต้องการเพียงข้อความอธิบายเดียวโดยไม่มีข้อมูลเพิ่มเติมอื่น

### ตัวอย่างที่ 2: การใช้งานจริง — ข้อผิดพลาดที่มีข้อมูลประกอบเฉพาะของตนเอง (ระบบยืมหนังสือ)

````python
%reset -f
class BookNotAvailableError(Exception):
    def __init__(self, title, remaining):
        self.title = title
        self.remaining = remaining
        self.message = f"หนังสือ '{title}' ไม่พร้อมให้ยืม (คงเหลือ {remaining} เล่ม)"
        super().__init__(self.message)

    def __str__(self):
        return self.message

library_stock = {"Python เบื้องต้น": 0, "โครงสร้างข้อมูล": 3}

def borrow_book(title):
    remaining = library_stock.get(title, 0)
    if remaining <= 0:
        raise BookNotAvailableError(title, remaining)
    library_stock[title] = remaining - 1
    return f"ยืม '{title}' สำเร็จ คงเหลือ {library_stock[title]} เล่ม"

try:
    print(borrow_book("Python เบื้องต้น"))
except BookNotAvailableError as e:
    print("ยืมไม่สำเร็จ:", e)
    print("ชื่อหนังสือที่พบปัญหา:", e.title)

print(borrow_book("โครงสร้างข้อมูล"))
````

**Output:**
```
ยืมไม่สำเร็จ: หนังสือ 'Python เบื้องต้น' ไม่พร้อมให้ยืม (คงเหลือ 0 เล่ม)
ชื่อหนังสือที่พบปัญหา: Python เบื้องต้น
ยืม 'โครงสร้างข้อมูล' สำเร็จ คงเหลือ 2 เล่ม
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | คำอธิบายเพิ่มเติม |
|---|---|---|---|
| 1–9 | นิยามคลาส `BookNotAvailableError` | คลาสใหม่ที่มี `title`, `remaining`, `message` | เพิ่ม `__init__` และ `__str__` ของตัวเอง เก็บข้อมูลเฉพาะของข้อผิดพลาดนี้ |
| 6 | `super().__init__(self.message)` | ส่ง `message` ให้คลาสฐาน `Exception` | ทำให้ `str(e)` และกลไกมาตรฐานของ exception ทำงานถูกต้อง |
| 11 | `library_stock = {...}` | dict เก็บจำนวนหนังสือคงเหลือ | `"Python เบื้องต้น"` มีค่า 0 (ไม่พร้อมให้ยืม) |
| 21 | `borrow_book("Python เบื้องต้น")` | `remaining = 0` ที่บรรทัด 14 | เงื่อนไขบรรทัด 15 เป็นจริง จึงโยน `BookNotAvailableError` ที่บรรทัด 16 |
| 22–24 | `except BookNotAvailableError as e` | `e.title = "Python เบื้องต้น"` | เข้าถึงข้อมูลเฉพาะ (`e.title`) ที่แนบมากับอ็อบเจกต์ข้อผิดพลาดได้โดยตรง |
| 26 | `borrow_book("โครงสร้างข้อมูล")` | `remaining = 3` | เงื่อนไขบรรทัด 15 เป็นเท็จ ยืมสำเร็จ คงเหลือลดลงเป็น 2 |

**คำอธิบายภาพรวม:** บรรทัด 1–9 นิยามข้อผิดพลาดที่ผู้ใช้กำหนดเองที่มีข้อมูลประกอบ (`title`, `remaining`) นอกเหนือจากข้อความอธิบายทั่วไป
โดยเก็บค่าเหล่านี้ไว้เป็นแอตทริบิวต์ของอ็อบเจกต์ข้อผิดพลาดเองในบรรทัด 3–4 แล้วจึงส่งข้อความสรุปให้คลาสฐานผ่าน `super().__init__()`
ที่บรรทัด 6 บรรทัด 13–18 ใช้ข้อผิดพลาดนี้ในฟังก์ชัน `borrow_book` เพื่อโยนข้อผิดพลาดเมื่อหนังสือหมด บรรทัด 20–24 แสดงประโยชน์ของ
การแนบข้อมูลไว้กับอ็อบเจกต์ข้อผิดพลาด คือโค้ดฝั่งที่ดักข้อผิดพลาดสามารถเข้าถึง `e.title` เพื่อนำไปใช้ต่อได้โดยตรง ไม่ต้องแยกวิเคราะห์
จากข้อความ บรรทัด 26 แสดงกรณีที่ยืมสำเร็จตามปกติ

**จุดสังเกต:** ข้อผิดพลาดที่ผู้ใช้กำหนดเองสามารถพก "ข้อมูลโครงสร้าง" (structured data) ติดไปกับตัวมันเองได้ ไม่ใช่แค่ข้อความเดียว
ต่างจากการพิมพ์ข้อความ error ธรรมดาที่โค้ดฝั่งเรียกต้องแยกวิเคราะห์ข้อความเองหากต้องการนำข้อมูลไปใช้ต่อ ในตัวอย่างนี้ `e.title`
เข้าถึงได้ทันทีโดยไม่ต้องแกะจากข้อความ `str(e)` เลย ซึ่งเป็นประโยชน์หลักของการเขียน `__init__` ของตนเองเพิ่มเติมจากตัวอย่างที่ 1

### ตัวอย่างที่ 3: ผลของการไม่เรียก `super().__init__()` ในคลาสข้อผิดพลาดที่ผู้ใช้กำหนดเอง (ระบบถอนเงิน)

````python
%reset -f
class InsufficientFundsError(Exception):
    def __init__(self, balance, amount):
        self.balance = balance
        self.amount = amount
        # ไม่ได้เรียก super().__init__() ส่งข้อความใด ๆ ให้คลาสฐาน

def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientFundsError(balance, amount)
    return balance - amount

try:
    withdraw(500, 800)
except InsufficientFundsError as e:
    print("str(e) แสดงผลว่า:", repr(str(e)))
    print("แต่ยังเข้าถึง e.balance ได้:", e.balance)
    print("และเข้าถึง e.amount ได้:", e.amount)
    print("e.args ที่คลาสฐานเก็บไว้:", e.args)
````

**Output:**
```
str(e) แสดงผลว่า: '(500, 800)'
แต่ยังเข้าถึง e.balance ได้: 500
และเข้าถึง e.amount ได้: 800
e.args ที่คลาสฐานเก็บไว้: (500, 800)
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | คำอธิบายเพิ่มเติม |
|---|---|---|---|
| 1–5 | นิยามคลาส `InsufficientFundsError` | `__init__` เก็บ `balance`, `amount` เป็นแอตทริบิวต์ | ไม่มีบรรทัดเรียก `super().__init__(...)` เลย |
| 9 | `raise InsufficientFundsError(500, 800)` | สร้างอ็อบเจกต์ข้อผิดพลาดด้วยอาร์กิวเมนต์ `(500, 800)` | Python เรียก `__new__` ของ `BaseException` ก่อนเรียก `__init__` เสมอ |
| — | (กลไกภายในของ `BaseException.__new__`) | `e.args = (500, 800)` | ค่านี้ถูกกำหนดโดย `__new__` โดยอัตโนมัติ ไม่ได้ขึ้นกับว่า `__init__` จะเรียก `super().__init__()` หรือไม่ |
| 15 | `str(e)` | คืนค่า `"(500, 800)"` | `BaseException.__str__` แบบมาตรฐานแสดงผลจาก `self.args` เมื่อมีมากกว่า 1 ค่า จะแสดงเป็น tuple |
| 16–17 | `e.balance`, `e.amount` | 500, 800 | เข้าถึงได้ปกติเพราะถูกกำหนดโดยตรงในบรรทัด 3–4 ของ `__init__` |
| 18 | `e.args` | `(500, 800)` | ยืนยันว่า `args` มีค่าอยู่แม้ไม่ได้เรียก `super().__init__()` |

**คำอธิบายภาพรวม:** ตัวอย่างนี้สาธิตกรณีที่คลาสข้อผิดพลาดที่ผู้ใช้กำหนดเองเขียน `__init__` ของตัวเอง (บรรทัด 2–5) แต่ไม่ได้เรียก
`super().__init__()` เพื่อส่งข้อมูลให้คลาสฐาน `Exception` เมื่อโยนข้อผิดพลาดที่บรรทัด 9 และดักที่บรรทัด 14 พบว่า `str(e)`
ที่บรรทัด 15 ยังคงแสดงผลเป็น `"(500, 800)"` ได้ และ `e.args` ที่บรรทัด 18 ก็มีค่าอยู่ ทั้งที่โค้ดไม่ได้ตั้งค่าทั้งสองนี้เองอย่างชัดเจน
เลยในบรรทัด 2–5

**จุดสังเกต:** พฤติกรรมนี้ไม่ตรงกับสัญชาตญาณที่คาดไว้ล่วงหน้า เพราะโดยทั่วไปเมื่อคลาสลูกเขียน `__init__` ทับคลาสฐานโดยไม่เรียก
`super().__init__()` แอตทริบิวต์ที่คลาสฐานควรตั้งค่าให้มักจะไม่ถูกตั้งค่า แต่กรณีของ `BaseException` (คลาสต้นทางของ `Exception`)
แตกต่างออกไป เพราะ Python กำหนดค่า `args` ผ่านเมท็อด `__new__` ของ `BaseException` ซึ่งทำงานก่อนหน้า `__init__` เสมอ
และรับอาร์กิวเมนต์ชุดเดียวกับที่ส่งเข้าคลาส ไม่ได้ขึ้นกับสิ่งที่เกิดขึ้นภายใน `__init__` ของคลาสลูกเลย ผลคือ `str(e)` ยังคงทำงานได้
แม้ผู้เขียนคลาสจะลืมเรียก `super().__init__()` ก็ตาม

**เชื่อมโยงกับหัวข้ออื่น:** ต่อยอดจากหัวข้อ 9 (`raise`) และหัวข้อ 8 (`except X as e`) ใน `lecture_note_intermediate.md`
เกี่ยวข้องโดยตรงกับหัวข้อ 11 (ลำดับชั้นของข้อผิดพลาด) ในไฟล์นี้ เพราะข้อผิดพลาดที่ผู้ใช้กำหนดเองคือคลาสลูกที่สืบทอดจาก `Exception`
และเป็นพื้นฐานของหัวข้อ 15 (exception chaining) ที่มักใช้ข้อผิดพลาดที่ผู้ใช้กำหนดเองเป็นชนิดที่แปลงมาจากข้อผิดพลาดระดับล่าง

## 11: ลำดับชั้นของข้อผิดพลาด (Exception Hierarchy)

**แหล่งอ้างอิง**: GeeksforGeeks — [Python Exception Handling](https://www.geeksforgeeks.org/python/python-exception-handling/);
TutorialsPoint — [Python Built-in Exceptions](https://www.tutorialspoint.com/python/python_builtin_exceptions.htm)

### แนวคิด

ข้อผิดพลาดขณะรัน (exception) ที่ภาษาโปรแกรมมิ่งเตรียมไว้ให้ ไม่ได้เป็นรายการแบนราบที่แยกจากกันโดยสิ้นเชิง แต่จัดเรียงเป็น
"ลำดับชั้น" (hierarchy) แบบต้นไม้ กล่าวคือข้อผิดพลาดชนิดที่เจาะจงกว่าจะสืบทอดคุณสมบัติมาจากข้อผิดพลาดชนิดที่กว้างกว่า
เช่นเดียวกับการจัดหมวดหมู่สิ่งมีชีวิตที่ชนิดย่อยสืบทอดคุณสมบัติมาจากชนิดที่กว้างกว่า การจัดลำดับชั้นเช่นนี้ทำให้สามารถดักข้อผิดพลาด
"ทั้งกลุ่ม" ได้ด้วยการดักที่ชนิดบรรพบุรุษ (parent) เพียงชนิดเดียว โดยไม่ต้องระบุชนิดลูกหลาน (child) ทุกชนิดทีละรายการ

ในภาษา Python ข้อผิดพลาดทั้งหมดสืบทอดมาจากคลาส `BaseException` ซึ่งเป็นรากของลำดับชั้นทั้งหมด ถัดลงมาคือคลาส `Exception`
ซึ่งเป็นคลาสฐานของข้อผิดพลาดทั่วไปเกือบทั้งหมดที่เกิดจากการทำงานผิดพลาดของโปรแกรม (แยกออกจาก `SystemExit` และ
`KeyboardInterrupt` ซึ่งสืบทอดจาก `BaseException` โดยตรงแต่ไม่ใช่ `Exception` เพราะสื่อถึงการควบคุมการทำงานของโปรแกรม
มากกว่าข้อผิดพลาด) ภายใต้ `Exception` ยังมีคลาสกลาง (intermediate class) อีกหลายชั้น เช่น `ArithmeticError` เป็นคลาสแม่ของ
`ZeroDivisionError`, `OverflowError`, `FloatingPointError` และ `LookupError` เป็นคลาสแม่ของ `IndexError`, `KeyError`
กลไกของ `except` ในบล็อกดักข้อผิดพลาดจะจับข้อผิดพลาดที่เป็น "ชนิดเดียวกัน หรือเป็นคลาสลูกของชนิดที่ระบุ" เสมอ ด้วยเหตุนี้
`except Exception` จึงดักข้อผิดพลาดเกือบทุกชนิดที่เกิดขึ้นระหว่างการทำงานของโปรแกรมทั่วไปได้ เพราะข้อผิดพลาดเหล่านั้นล้วนเป็น
คลาสลูกของ `Exception` ทั้งสิ้น เชื่อมโยงโดยตรงกับหัวข้อ 3 (ชนิดข้อผิดพลาดที่พบบ่อย) ใน `lecture_note_basic.md`
และหัวข้อ 4 (จับหลาย exception) ในไฟล์เดียวกัน

### ตัวอย่างที่ 1: ตรวจสอบลำดับชั้นของ `ZeroDivisionError` (พื้นฐานของแนวคิด)

````python
%reset -f
try:
    result = 10 / 0
except ZeroDivisionError as e:
    print("จับได้ด้วย ZeroDivisionError")

print(ZeroDivisionError.__mro__)
print(issubclass(ZeroDivisionError, ArithmeticError))
print(issubclass(ArithmeticError, Exception))
print(issubclass(Exception, BaseException))
````

**Output:**
```
จับได้ด้วย ZeroDivisionError
(<class 'ZeroDivisionError'>, <class 'ArithmeticError'>, <class 'Exception'>, <class 'BaseException'>, <class 'object'>)
True
True
True
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | คำอธิบายเพิ่มเติม |
|---|---|---|---|
| 2 | `10 / 0` | โยนข้อผิดพลาด `ZeroDivisionError` | ตัวหารเป็นศูนย์ |
| 3–4 | `except ZeroDivisionError as e` | จับได้ตรงชนิด | พิมพ์ข้อความยืนยัน |
| 6 | `ZeroDivisionError.__mro__` | tuple ของคลาสตามลำดับการสืบทอด | แสดงลำดับชั้นทั้งหมดตั้งแต่ตัวเองถึง `object` |
| 7 | `issubclass(ZeroDivisionError, ArithmeticError)` | `True` | ยืนยันว่า `ZeroDivisionError` เป็นคลาสลูกของ `ArithmeticError` |
| 8 | `issubclass(ArithmeticError, Exception)` | `True` | ยืนยันว่า `ArithmeticError` เป็นคลาสลูกของ `Exception` |
| 9 | `issubclass(Exception, BaseException)` | `True` | ยืนยันว่า `Exception` เป็นคลาสลูกของ `BaseException` |

**คำอธิบายภาพรวม:** บรรทัด 1–4 ทำให้เกิดและดัก `ZeroDivisionError` ตามปกติ บรรทัด 6 ใช้แอตทริบิวต์ `__mro__`
(Method Resolution Order) ของคลาสเพื่อแสดงลำดับการสืบทอดทั้งหมดของ `ZeroDivisionError` อย่างชัดเจนเป็น tuple ของคลาส
ตั้งแต่ตัวเองไปจนถึง `object` ซึ่งเป็นคลาสฐานสูงสุดของทุกสิ่งในภาษา Python บรรทัด 7–9 ยืนยันความสัมพันธ์แบบคลาสแม่-คลาสลูก
ของแต่ละชั้นในลำดับชั้นนั้นทีละคู่ด้วย `issubclass()`

**จุดสังเกต:** ผลลัพธ์จาก `__mro__` แสดงให้เห็นว่า `ZeroDivisionError` ไม่ได้สืบทอดจาก `Exception` โดยตรง แต่สืบทอดผ่าน
คลาสกลาง `ArithmeticError` ก่อน การเข้าใจว่ามีคลาสกลางเหล่านี้อยู่ช่วยอธิบายว่าทำไม `except ArithmeticError` เพียงบล็อกเดียว
จึงดักได้ทั้ง `ZeroDivisionError`, `OverflowError` และ `FloatingPointError` พร้อมกัน (ดูตัวอย่างที่ 2) โดยไม่ต้องเขียน
`except` แยกทีละชนิด

### ตัวอย่างที่ 2: การใช้งานจริง — ดักข้อผิดพลาดกลุ่มเดียวกันด้วยคลาสแม่ (เครื่องคำนวณ)

````python
%reset -f
import math

def calculate(op, a, b):
    try:
        if op == "divide":
            return a / b
        elif op == "power":
            return math.exp(a ** b)
    except ArithmeticError as e:
        return f"คำนวณไม่ได้ ({type(e).__name__}): {e}"

print(calculate("divide", 10, 0))
print(calculate("power", 10, 5000))
print(calculate("divide", 10, 2))
````

**Output:**
```
คำนวณไม่ได้ (ZeroDivisionError): division by zero
คำนวณไม่ได้ (OverflowError): int too large to convert to float
5.0
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | คำอธิบายเพิ่มเติม |
|---|---|---|---|
| 12 | `calculate("divide", 10, 0)` | `10 / 0` โยน `ZeroDivisionError` | บรรทัด 9 ดักได้เพราะ `ZeroDivisionError` เป็นคลาสลูกของ `ArithmeticError` |
| 13 | `calculate("power", 10, 5000)` | `10 ** 5000` เป็นจำนวนเต็มขนาดใหญ่มาก แปลงเป็น float ไม่ได้ | `math.exp(...)` โยน `OverflowError` ซึ่งเป็นคลาสลูกของ `ArithmeticError` เช่นกัน |
| 9–10 | `except ArithmeticError as e` | `type(e).__name__` ต่างกันในแต่ละครั้ง | บล็อกเดียวจับได้ทั้งสองชนิดเพราะทั้งคู่เป็นคลาสลูกของ `ArithmeticError` |
| 14 | `calculate("divide", 10, 2)` | คืนค่า 5.0 | ไม่มีข้อผิดพลาดเกิดขึ้น จึงคืนผลลัพธ์ตามปกติ |

**คำอธิบายภาพรวม:** ฟังก์ชัน `calculate` มีบล็อกดักข้อผิดพลาดเพียงบล็อกเดียวคือ `except ArithmeticError as e` (บรรทัด 9)
แต่สามารถรองรับได้ทั้งกรณีหารด้วยศูนย์ (บรรทัด 12 → `ZeroDivisionError`) และกรณีตัวเลขมีขนาดใหญ่เกินกว่าจะแปลงเป็น float ได้
(บรรทัด 13 → `OverflowError`) เพราะข้อผิดพลาดทั้งสองชนิดนี้เป็นคลาสลูกของ `ArithmeticError` ทั้งคู่ (ตามที่แสดงในตัวอย่างที่ 1)
บรรทัด 10 ใช้ `type(e).__name__` เพื่อแสดงชนิดที่แท้จริงของข้อผิดพลาดที่จับได้ในแต่ละครั้ง ทำให้ทราบว่าเกิดข้อผิดพลาดชนิดใดแม้จะดัก
ด้วยคลาสแม่ก็ตาม

**จุดสังเกต:** การดักด้วยคลาสแม่ (`ArithmeticError`) ไม่ได้ทำให้สูญเสียข้อมูลว่าข้อผิดพลาดจริง ๆ เป็นชนิดใด ตราบใดที่ยังใช้
`type(e).__name__` หรือ `e` ตรวจสอบต่อ ข้อดีคือลดจำนวนบล็อก `except` ที่ต้องเขียนซ้ำเมื่อต้องจัดการข้อผิดพลาดหลายชนิด
ในกลุ่มเดียวกันด้วยวิธีการเดียวกัน แต่หากต้องจัดการแต่ละชนิดด้วยวิธีที่ต่างกันโดยสิ้นเชิง การเขียน `except` แยกทีละชนิดตามหัวข้อ 4
ใน `lecture_note_basic.md` ยังคงเหมาะสมกว่า

### ตัวอย่างที่ 3: ผลของการดักด้วยคลาสกว้างเกินไปต่อข้อผิดพลาดที่ไม่เกี่ยวข้องกัน (แบบฟอร์มลงทะเบียน)

````python
%reset -f
def register(form):
    try:
        age = int(form["age"])
        naem = form["name"]  # พิมพ์ผิด ตั้งใจให้เกิด NameError เมื่อใช้ตัวแปรนี้ผิดชื่อ
        print(f"ลงทะเบียน {name} อายุ {age} สำเร็จ")
    except Exception as e:
        print(f"ลงทะเบียนไม่สำเร็จ ({type(e).__name__}): {e}")

register({"age": "abc", "name": "มานะ"})
register({"age": "20", "name": "ปิติ"})
````

**Output:**
```
ลงทะเบียนไม่สำเร็จ (ValueError): invalid literal for int() with base 10: 'abc'
ลงทะเบียนไม่สำเร็จ (NameError): name 'name' is not defined
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | คำอธิบายเพิ่มเติม |
|---|---|---|---|
| 9 | `register({"age": "abc", ...})` | `int("abc")` โยน `ValueError` ที่บรรทัด 3 | เกิดจากข้อมูลนำเข้าไม่ถูกต้อง (คาดการณ์ไว้ล่วงหน้าได้) |
| 6–7 | `except Exception as e` | จับได้ ชนิด `ValueError` | พิมพ์ข้อความทั่วไปที่ระบุชนิดจริงผ่าน `type(e).__name__` |
| 10 | `register({"age": "20", ...})` | `int("20")` สำเร็จ บรรทัด 4 กำหนดตัวแปรชื่อ `naem` (พิมพ์ผิด) | บรรทัด 5 อ้างถึงตัวแปรชื่อ `name` ซึ่งไม่เคยถูกกำหนดค่า จึงโยน `NameError` |
| 6–7 | `except Exception as e` | จับได้ ชนิด `NameError` | ข้อผิดพลาดนี้เกิดจากข้อบกพร่องในโค้ดเอง (พิมพ์ชื่อตัวแปรผิด) ไม่ใช่ปัญหาจากข้อมูลนำเข้า |

**คำอธิบายภาพรวม:** ฟังก์ชัน `register` ดักข้อผิดพลาดด้วย `except Exception as e` เพียงบล็อกเดียว (บรรทัด 6) ซึ่งครอบคลุม
ทั้งข้อผิดพลาดที่เกิดจากข้อมูลนำเข้าไม่ถูกต้อง (`ValueError` จากบรรทัด 9) และข้อผิดพลาดที่เกิดจากข้อบกพร่องในตัวโค้ดเอง
คือการพิมพ์ชื่อตัวแปรผิด (`naem` แทน `name` ที่บรรทัด 4 ทำให้บรรทัด 5 โยน `NameError` เมื่อเรียกจากบรรทัด 10) เนื่องจากทั้ง
`ValueError` และ `NameError` เป็นคลาสลูกของ `Exception` ทั้งคู่ (ตามลำดับชั้นที่อธิบายไว้ในหัวข้อนี้) บล็อก `except Exception`
จึงจับได้ทั้งสองกรณีด้วยข้อความรูปแบบเดียวกันทุกประการ ต่างกันเพียงค่าที่ `type(e).__name__` แสดงออกมา

**จุดสังเกต:** ข้อความผลลัพธ์ทั้งสองบรรทัดมีรูปแบบเดียวกันทุกประการ ("ลงทะเบียนไม่สำเร็จ (...)：...") ทั้งที่สาเหตุที่แท้จริง
แตกต่างกันโดยสิ้นเชิง กรณีแรกเป็นปัญหาจากข้อมูลที่ผู้ใช้ป้อน (แก้ได้ด้วยการขอข้อมูลใหม่) ส่วนกรณีที่สองเป็นข้อบกพร่องที่อยู่ในโค้ด
เอง (การพิมพ์ชื่อตัวแปรผิดที่บรรทัด 4) ซึ่งจะเกิดขึ้นซ้ำทุกครั้งไม่ว่าใครจะป้อนข้อมูลอย่างไรก็ตาม การดักด้วยคลาสที่กว้างอย่าง
`Exception` ทำให้ข้อผิดพลาดทั้งสองประเภทนี้ถูกปฏิบัติเหมือนกันในทางกลไก แม้ว่าโดยธรรมชาติของปัญหาแล้วจะต้องการการจัดการที่
ต่างกันโดยสิ้นเชิงก็ตาม ผู้เขียนโปรแกรมจึงต้องอาศัย `type(e).__name__` หรือการตรวจสอบ `e` เพิ่มเติมเพื่อแยกแยะสองกรณีนี้ออกจากกัน

**เชื่อมโยงกับหัวข้ออื่น:** ต่อยอดจากหัวข้อ 3 (ชนิดข้อผิดพลาดที่พบบ่อย) และหัวข้อ 4 (จับหลาย exception) ใน `lecture_note_basic.md`
เป็นพื้นฐานสำคัญของหัวข้อ 10 (custom exception ในไฟล์นี้ ซึ่งต้องสืบทอดจาก `Exception`) และหัวข้อ 14 (bare except ในไฟล์นี้
ซึ่งดักกว้างกว่า `except Exception` อีกขั้นหนึ่ง)

## 12: การยืนยันเงื่อนไข (`assert` / `AssertionError`)

**แหล่งอ้างอิง**: TutorialsPoint — [Python Assertions](https://www.tutorialspoint.com/python/python_assertions.htm)

### แนวคิด

การยืนยันเงื่อนไข (assertion) คือกลไกที่ให้ผู้เขียนโปรแกรมระบุ "สิ่งที่ควรเป็นจริงเสมอ" ไว้ในโค้ดโดยตรง หากเงื่อนไขนั้นกลับกลายเป็นเท็จ
ระหว่างการทำงาน แสดงว่ามีบางอย่างผิดพลาดไปจากสมมติฐานที่ตั้งไว้ตั้งแต่แรก จึงหยุดการทำงานทันทีเพื่อให้ผู้เขียนโปรแกรมทราบและแก้ไข
แนวคิดนี้แตกต่างจากการตรวจสอบข้อผิดพลาดทั่วไป (เช่นการโยนข้อผิดพลาดด้วย `raise` ในหัวข้อ 9) ตรงที่ assertion มีไว้สำหรับตรวจสอบ
สมมติฐานภายในของโปรแกรมเองระหว่างการพัฒนาและทดสอบเป็นหลัก มากกว่าจะใช้ตรวจสอบข้อมูลนำเข้าจากผู้ใช้หรือแหล่งภายนอกที่คาดเดา
ไม่ได้ตั้งแต่ต้น

ในภาษา Python คำสั่ง `assert` มีรูปแบบ `assert เงื่อนไข, ข้อความ` หากเงื่อนไขเป็นเท็จ Python จะโยนข้อผิดพลาดชนิด
`AssertionError` พร้อมข้อความที่ระบุไว้โดยอัตโนมัติทันที เชื่อมโยงกับหัวข้อ 5 (defensive function) ใน `lecture_note_basic.md`
และหัวข้อ 9 (`raise`) ใน `lecture_note_intermediate.md` — ทั้งสามกลไก (assert, raise, try/except) ทำงานร่วมกันได้
ในโปรแกรมเดียวกัน โดยแต่ละกลไกมีจุดประสงค์ต่างกัน: `raise` มักใช้ตรวจสอบเงื่อนไขที่มาจากภายนอก (เช่นข้อมูลผู้ใช้)
ส่วน `assert` มักใช้ตรวจสอบสมมติฐานภายในระหว่างพัฒนา

### ตัวอย่างที่ 1: ตรวจสอบขอบเขตค่าอุณหภูมิ (พื้นฐานของแนวคิด)

````python
%reset -f
def celsius_to_fahrenheit(celsius):
    assert celsius >= -273.15, "อุณหภูมิต่ำกว่าศูนย์องศาสัมบูรณ์ (absolute zero) ไม่ได้"
    return celsius * 9 / 5 + 32

print(celsius_to_fahrenheit(25))
print(celsius_to_fahrenheit(-300))
````

**Output:**
```

Traceback (most recent call last):

    assert celsius >= -273.15, "อุณหภูมิต่ำกว่าศูนย์องศาสัมบูรณ์ (absolute zero) ไม่ได้"
           ^^^^^^^^^^^^^^^^^^

```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | คำอธิบายเพิ่มเติม |
|---|---|---|---|
| 5 | `celsius_to_fahrenheit(25)` | เงื่อนไขบรรทัด 2 เป็นจริง (25 ≥ -273.15) | ทำงานต่อจนคืนค่า 77.0 |
| 6 | `celsius_to_fahrenheit(-300)` | เงื่อนไขบรรทัด 2 เป็นเท็จ (-300 < -273.15) | โยน `AssertionError` ทันที บรรทัด 3 ไม่ถูกทำงาน |
| — | (ไม่มีบล็อก try/except ครอบ) | โปรแกรมหยุดทำงานทันที | แสดงข้อความแสดงลำดับข้อผิดพลาด (traceback) เต็มรูปแบบออกทางหน้าจอ |

**คำอธิบายภาพรวม:** บรรทัด 2 ตั้งสมมติฐานว่าอุณหภูมิเซลเซียสที่ส่งเข้ามาต้องไม่ต่ำกว่าศูนย์องศาสัมบูรณ์ ซึ่งเป็นขีดจำกัดทางฟิสิกส์
ที่ไม่ควรมีค่าใดต่ำกว่านี้ได้ในความเป็นจริง บรรทัด 5 ส่งค่าที่สมเหตุสมผล (25) จึงผ่านการยืนยันเงื่อนไขและคำนวณต่อไปได้ตามปกติ
บรรทัด 6 ส่งค่าที่ผิดสมมติฐาน (-300) เงื่อนไขที่บรรทัด 2 จึงเป็นเท็จ ทำให้ Python โยน `AssertionError` พร้อมข้อความที่ระบุไว้ทันที
เนื่องจากไม่มีบล็อกทดลองรัน/บล็อกดักข้อผิดพลาดใดครอบฟังก์ชันนี้ไว้ โปรแกรมจึงหยุดทำงานและแสดงข้อความแสดงลำดับข้อผิดพลาดเต็มรูปแบบ

**จุดสังเกต:** ข้อความแสดงลำดับข้อผิดพลาดแสดงตำแหน่งของนิพจน์เงื่อนไขที่ล้มเหลว (`celsius >= -273.15`) พร้อมเครื่องหมาย `^`
กำกับตำแหน่งที่ตรวจสอบไว้อย่างชัดเจน ต่างจากการโยนข้อผิดพลาดด้วย `raise` ทั่วไปตรงที่ `assert` ผูกเงื่อนไขและข้อความไว้ในบรรทัด
เดียวกันอย่างกระชับ และหากไม่มีการดักข้อผิดพลาดไว้ โปรแกรมจะหยุดทำงานทันทีเช่นเดียวกับข้อผิดพลาดชนิดอื่นที่ไม่ถูกดัก

### ตัวอย่างที่ 2: การใช้งานจริง — ตรวจสอบสมมติฐานก่อนคำนวณค่าเฉลี่ย (สถิติคะแนน)

````python
%reset -f
def average(scores):
    assert len(scores) > 0, "รายการคะแนนต้องไม่ว่างเปล่า"
    return sum(scores) / len(scores)

print(average([80, 90, 70]))
print(average([]))
````

**Output:**
```
80.0
Traceback (most recent call last):
  File "t12_ex2.py", line 6, in <module>
    print(average([]))
          ^^^^^^^^^^^
  File "t12_ex2.py", line 2, in average
    assert len(scores) > 0, "รายการคะแนนต้องไม่ว่างเปล่า"
           ^^^^^^^^^^^^^^^
AssertionError: รายการคะแนนต้องไม่ว่างเปล่า
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | คำอธิบายเพิ่มเติม |
|---|---|---|---|
| 5 | `average([80, 90, 70])` | `len(scores) = 3 > 0` เป็นจริง | คำนวณ `240 / 3 = 80.0` และคืนค่า |
| 6 | `average([])` | `len(scores) = 0` เงื่อนไขบรรทัด 2 เป็นเท็จ | โยน `AssertionError` ก่อนถึงบรรทัด 3 |
| — | (สมมติฐานที่ป้องกันไว้) | หาก `assert` ไม่มีอยู่ | บรรทัด 3 จะคำนวณ `sum([]) / len([])` ซึ่งคือ `0 / 0` และโยน `ZeroDivisionError` แทน |

**คำอธิบายภาพรวม:** ฟังก์ชัน `average` ตั้งสมมติฐานไว้ที่บรรทัด 2 ว่ารายการคะแนนที่ส่งเข้ามาต้องมีอย่างน้อยหนึ่งค่าเสมอ ก่อนที่จะ
คำนวณค่าเฉลี่ยที่บรรทัด 3 บรรทัด 5 ส่งรายการที่มีข้อมูล จึงผ่านการตรวจสอบและคำนวณได้ตามปกติ บรรทัด 6 ส่งรายการว่างเปล่า
ซึ่งขัดกับสมมติฐานที่ตั้งไว้ `assert` จึงหยุดการทำงานก่อนที่จะไปถึงคำสั่งคำนวณจริงที่บรรทัด 3

**จุดสังเกต:** หากไม่มีบรรทัด 2 อยู่เลย การส่งรายการว่างเข้าไปจะทำให้บรรทัด 3 คำนวณ `sum([]) / len([])` ซึ่งเท่ากับ `0 / 0`
และโปรแกรมจะหยุดทำงานด้วย `ZeroDivisionError` แทน ซึ่งเป็นข้อผิดพลาดที่ชี้ไปยังอาการปลายทาง (การหารด้วยศูนย์)
ไม่ใช่สาเหตุต้นทาง (รายการว่างเปล่า) การวาง `assert` ไว้ก่อนทำให้ข้อความข้อผิดพลาดที่ได้ชี้ตรงไปยังสาเหตุที่แท้จริงของปัญหาแทน

### ตัวอย่างที่ 3: ผลของการเขียน `assert` เป็น tuple โดยไม่ตั้งใจ (ตรวจจำนวนสินค้า)

````python
%reset -f
def check_quantity(qty):
    assert (qty > 0, "จำนวนสินค้าต้องมากกว่า 0")
    return qty

print(check_quantity(5))
print(check_quantity(-3))
print(check_quantity(0))
````

**Output:**
```

  assert (qty > 0, "จำนวนสินค้าต้องมากกว่า 0")

0
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | คำอธิบายเพิ่มเติม |
|---|---|---|---|
| 2 | `assert (qty > 0, "...")` | นิพจน์ที่ประเมินจริง ๆ คือ tuple `(qty > 0, "...")` | เครื่องหมายจุลภาคภายในวงเล็บทำให้ทั้งก้อนกลายเป็น tuple 2 สมาชิก ไม่ใช่เงื่อนไขเดียวพร้อมข้อความ |
| — | ตัวแปลภาษา (interpreter) ตรวจพบรูปแบบนี้ | แสดง `SyntaxWarning: assertion is always true, perhaps remove parentheses?` | ข้อความนี้ปรากฏตอนแปลไฟล์ ก่อนโปรแกรมจะรันจริงด้วยซ้ำ |
| 5 | `check_quantity(5)` | tuple `(True, "...")` ไม่ว่างเปล่า จึงเป็นค่าจริงเสมอ | `assert` ผ่านเสมอไม่ว่า `qty` จะเป็นเท่าใด คืนค่า 5 |
| 6 | `check_quantity(-3)` | tuple `(False, "...")` ก็ยังไม่ว่างเปล่า จึงยังเป็นค่าจริง | `assert` ผ่านทั้งที่ `qty > 0` เป็นเท็จ คืนค่า -3 |
| 7 | `check_quantity(0)` | tuple `(False, "...")` ไม่ว่างเปล่าเช่นกัน | `assert` ผ่านเช่นเดียวกัน คืนค่า 0 |

**คำอธิบายภาพรวม:** บรรทัด 2 ตั้งใจเขียนเงื่อนไขพร้อมข้อความในรูปแบบ `assert เงื่อนไข, ข้อความ` แต่พิมพ์วงเล็บครอบทั้งสองส่วนไว้
ด้วยกันโดยไม่ตั้งใจ ทำให้ไวยากรณ์ที่แท้จริงกลายเป็น `assert (เงื่อนไข, ข้อความ)` ซึ่งภาษา Python ตีความว่าเป็นการยืนยันค่าของ
"tuple ที่มีสองสมาชิก" ไม่ใช่การยืนยัน "เงื่อนไขพร้อมข้อความประกอบ" ตามที่ตั้งใจ เนื่องจาก tuple ที่มีสมาชิกอย่างน้อยหนึ่งตัว
ถือเป็นค่าจริง (truthy) เสมอไม่ว่าสมาชิกภายในจะเป็นค่าใดก็ตาม บรรทัด 5–7 จึงผ่านการยืนยันเงื่อนไขทุกครั้งไม่ว่า `qty` จะเป็นค่าบวก
ลบ หรือศูนย์ก็ตาม

**จุดสังเกต:** ตัวแปลภาษา Python ตรวจพบรูปแบบนี้และแสดง `SyntaxWarning` เตือนไว้ตั้งแต่ตอนแปลไฟล์ (ก่อนรันบรรทัดคำสั่งใด ๆ
ด้วยซ้ำ) อย่างไรก็ตาม `SyntaxWarning` ไม่ได้หยุดการทำงานของโปรแกรมเหมือนข้อผิดพลาด (exception) โปรแกรมจึงยังคงทำงานต่อไปได้
ตามปกติทั้งสามบรรทัด (5, 6, 7) โดยไม่มีบรรทัดใดหยุดทำงานเลย แม้ค่า `qty` ที่ส่งเข้ามาในบรรทัด 6 และ 7 จะขัดกับเงื่อนไข
`qty > 0` ที่ตั้งใจไว้ตั้งแต่แรกก็ตาม จุดสำคัญคือช่องว่างระหว่างสิ่งที่ตัวอักษรบนหน้าจอ "ดูเหมือน" จะตรวจสอบ กับสิ่งที่ไวยากรณ์ของ
ภาษาจริง ๆ ตีความ

**เชื่อมโยงกับหัวข้ออื่น:** ต่อยอดจากหัวข้อ 5 (defensive function) ใน `lecture_note_basic.md` และหัวข้อ 9 (`raise`)
ใน `lecture_note_intermediate.md` เกี่ยวข้องกับหัวข้อ 3 (ชนิดข้อผิดพลาดที่พบบ่อย) ในแง่ที่ `AssertionError` เป็นข้อผิดพลาด
อีกชนิดหนึ่งที่ดักได้ด้วย `except` ตามปกติ (แม้ตัวอย่างในหัวข้อนี้จะไม่ได้ดักไว้เพื่อแสดงข้อความแสดงลำดับข้อผิดพลาดเต็มรูปแบบก็ตาม)

## 13: การซ้อนบล็อก try/except/finally (Nested try/except/finally)

**แหล่งอ้างอิง**: TutorialsPoint — [Python Nested try Block](https://www.tutorialspoint.com/python/python_nested_try_block.htm)

### แนวคิด

โครงสร้างการทดลองรันและดักข้อผิดพลาด (try/except) หนึ่งชุดสามารถวางซ้อนอยู่ภายในโครงสร้างเดียวกันอีกชุดหนึ่งได้ ไม่ว่าจะซ้อนอยู่ใน
ส่วนทดลองรัน (try) ส่วนดักข้อผิดพลาด (except) หรือส่วนที่รันเสมอ (finally) ของชุดภายนอกก็ตาม แนวคิดนี้มีประโยชน์เมื่อขั้นตอนการทำงาน
แบ่งเป็นหลายระดับ และแต่ละระดับอาจเกิดข้อผิดพลาดคนละชนิดที่ต้องจัดการด้วยวิธีต่างกัน การซ้อนกันทำให้สามารถจัดการข้อผิดพลาดที่จุดที่
ใกล้กับต้นเหตุที่สุดก่อน แล้วปล่อยให้ข้อผิดพลาดที่ไม่ตรงกับเงื่อนไขของชุดภายในไหลต่อออกไปยังชุดภายนอกโดยอัตโนมัติ

หลักการสำคัญคือ: หากบล็อกดักข้อผิดพลาดของ try ชุดใน (inner) ไม่มีชนิดที่ตรงกับข้อผิดพลาดที่เกิดขึ้น ข้อผิดพลาดนั้นจะไม่หายไปไหน
แต่จะไหลออกไปยัง try ชุดนอก (outer) ที่ครอบอยู่ต่อไป จนกว่าจะพบบล็อกดักข้อผิดพลาดที่ตรงชนิด หรือไม่พบเลยจนโปรแกรมหยุดทำงาน
ส่วน finally ของทุกชุดที่ข้อผิดพลาดไหลผ่าน จะรันเสมอไม่ว่าข้อผิดพลาดจะถูกจัดการที่ชั้นใดก็ตาม เชื่อมโยงโดยตรงกับหัวข้อ 2
(try/except พื้นฐาน) ใน `lecture_note_basic.md` และหัวข้อ 6 (else), หัวข้อ 7 (finally) ใน `lecture_note_intermediate.md`

### ตัวอย่างที่ 1: การไหลของข้อผิดพลาดจาก try ชั้นในสู่ except ชั้นนอก (แปลงรหัสข้อมูลเป็นตัวเลข)

````python
%reset -f
records = {"1": "10", "2": "abc", "3": "30"}

def parse_record(key):
    try:
        try:
            text = records[key]
            number = int(text)
        except KeyError:
            print(f"ไม่พบรหัส {key}")
            return None
    except ValueError:
        print(f"รหัส {key} มีค่าที่แปลงเป็นตัวเลขไม่ได้: {text}")
        return None
    return number

print(parse_record("1"))
print(parse_record("2"))
print(parse_record("9"))
````

None

None

**คำอธิบายภาพรวม:** ฟังก์ชัน `parse_record` มีโครงสร้าง try ซ้อนกันสองชั้น ชั้นในดักเฉพาะ `KeyError` (บรรทัด 5–10)
ส่วนชั้นนอกดัก `ValueError` (บรรทัด 4, 11–13) เมื่อเรียกด้วย `"1"` (บรรทัด 16) ทุกอย่างสำเร็จโดยไม่มีข้อผิดพลาด
เมื่อเรียกด้วย `"2"` (บรรทัด 17) ค่าที่ดึงมาได้แปลงเป็นตัวเลขไม่ได้ เกิด `ValueError` ซึ่งไม่ตรงกับ `except KeyError` ของ
ชั้นใน ข้อผิดพลาดจึงไหลออกไปยังชั้นนอกซึ่งดักตรงชนิดพอดี เมื่อเรียกด้วย `"9"` (บรรทัด 18) รหัสไม่มีอยู่ใน dictionary
เกิด `KeyError` ซึ่งตรงกับ `except` ของชั้นในพอดี จึงถูกดักและจบการทำงานตั้งแต่ชั้นในโดยไม่ต้องไหลออกไปถึงชั้นนอกเลย

**จุดสังเกต:** ข้อผิดพลาดชนิดเดียวกัน (`ValueError` หรือ `KeyError`) ถูกดักที่ "ชั้น" ที่ต่างกันโดยขึ้นอยู่กับว่าชนิดของมันตรงกับ
บล็อกดักข้อผิดพลาดของชั้นใดเป็นชั้นแรกที่พบระหว่างการไหลออกจากจุดเกิดเหตุ ผู้เขียนโปรแกรมต้องออกแบบว่าจะดักข้อผิดพลาดแต่ละชนิด
ที่ "ชั้น" ใดอย่างตั้งใจ ไม่ใช่ปล่อยให้เป็นไปโดยบังเอิญ เพราะ except ของชั้นในที่ระบุชนิดไม่ตรงจะไม่ดักอะไรเลย ปล่อยให้ข้อผิดพลาด
ไหลผ่านไปเหมือนกับไม่มี try ชั้นในอยู่เลยสำหรับชนิดนั้นโดยเฉพาะ

### ตัวอย่างที่ 2: การใช้งานจริง — แยกจัดการข้อผิดพลาดคนละขั้นตอนพร้อม finally หลายชั้น (คำสั่งซื้อสินค้า)

````python
%reset -f
stock = {"เสื้อ": 2, "กางเกง": 0}

def process_order(item, payment_ok):
    try:
        try:
            if not payment_ok:
                raise ValueError("การชำระเงินไม่สำเร็จ")
            remaining = stock[item]
        except KeyError:
            print(f"ไม่พบสินค้า '{item}' ในคลัง")
            return
        finally:
            print(f"บันทึก log การตรวจสอบสินค้า '{item}'")

        if remaining <= 0:
            raise LookupError(f"สินค้า '{item}' หมดสต็อก")
        stock[item] -= 1
        print(f"สั่งซื้อ '{item}' สำเร็จ คงเหลือ {stock[item]}")
    except ValueError as e:
        print("ยกเลิกคำสั่งซื้อ:", e)
    except LookupError as e:
        print("ยกเลิกคำสั่งซื้อ:", e)
    finally:
        print(f"ปิดการทำรายการของ '{item}'")

process_order("เสื้อ", True)
process_order("กางเกง", True)
process_order("หมวก", True)
process_order("เสื้อ", False)
````

บันทึก log การตรวจสอบสินค้า 'กางเกง'

และ 23–24) การเรียกด้วย `"เสื้อ"` และชำระเงินสำเร็จ (บรรทัด 26) ทำให้ทุกขั้นตอนผ่านโดยไม่มีข้อผิดพลาด การเรียกด้วย `"กางเกง"`
(บรรทัด 27) สินค้าหมดสต็อกทำให้เกิด `LookupError` ที่ดักโดย except ชั้นนอก การเรียกด้วย `"หมวก"` (บรรทัด 28) ซึ่งไม่มีอยู่ใน
`stock` เลย ทำให้เกิด `KeyError` ที่ถูกดักโดย except ชั้นในทันที และการเรียกด้วย `"เสื้อ"` อีกครั้งแต่ชำระเงินไม่สำเร็จ
(บรรทัด 29) ทำให้เกิด `ValueError` ที่ดักโดย except ชั้นนอก ทุกกรณีแสดงให้เห็นว่า `finally` ทั้งสองชั้นรันครบเสมอไม่ว่าผลลัพธ์
ของการประมวลผลจะเป็นอย่างไร

**จุดสังเกต:** ในกรณีของ `"หมวก"` (บรรทัด 28) แม้ว่าโค้ดจะออกจากฟังก์ชันด้วยคำสั่ง `return` ที่อยู่ **ภายใน except ชั้นใน**
(บรรทัด 11) ไม่ใช่ด้วยข้อผิดพลาดที่ไหลออกไปถึงชั้นนอกเลย แต่ผลลัพธ์ยังคงแสดง `"ปิดการทำรายการของ 'หมวก'"` (finally ของชั้นนอก
บรรทัด 24) ตามมาด้วย นั่นเป็นเพราะ `return` ต้องไหลออกจากทุกชั้นของ try/finally ที่ครอบตัวมันอยู่ตามลำดับก่อนจะออกจากฟังก์ชันจริง
แม้ว่า `return` นั้นจะไม่ได้เกี่ยวข้องกับข้อผิดพลาดของ try ชั้นนอกเลยก็ตาม กล่าวคือ `finally` รันเมื่อมีการออกจากขอบเขตของ
try ไม่ว่าจะออกด้วยข้อผิดพลาดหรือด้วย `return` ก็ตาม

### ตัวอย่างที่ 3: ผลของ `return` ใน `finally` ต่อข้อผิดพลาดของ try ชั้นใน (เซนเซอร์วัดค่า)

````python
%reset -f
def read_sensor(value):
    try:
        try:
            if value < 0:
                raise ValueError("ค่าที่อ่านได้ติดลบ ผิดปกติ")
            return 100 / value
        finally:
            print("กำลังปิดการเชื่อมต่อเซนเซอร์")
            return -1
    except ValueError as e:
        print("จับข้อผิดพลาดที่ except ชั้นนอก:", e)
        return None

print(read_sensor(20))
print(read_sensor(-5))
print(read_sensor(0))
````

```

-1
กำลังปิดการเชื่อมต่อเซนเซอร์
-1
กำลังปิดการเชื่อมต่อเซนเซอร์
-1
```

|---|---|---|---|
| 14 | `read_sensor(20)` | บรรทัด 6 คำนวณ `100 / 20 = 5.0` และเตรียม `return 5.0` | ก่อนค่านี้จะถูกส่งออกจริง การควบคุมไหลเข้า `finally` บรรทัด 7–9 ก่อนเสมอ |
| 9 | `return -1` ใน finally | ค่าที่เตรียมไว้ (5.0) ถูกแทนที่ด้วย -1 | ผลลัพธ์ที่ผู้เรียกได้รับคือ -1 ไม่ใช่ 5.0 |
| 15 | `read_sensor(-5)` | บรรทัด 5 โยน `ValueError` | ก่อนข้อผิดพลาดนี้จะไหลออกจาก try ชั้นใน การควบคุมไหลเข้า `finally` บรรทัด 7–9 ก่อนเสมอเช่นกัน |
| 9 | `return -1` ใน finally | ข้อผิดพลาด `ValueError` ที่กำลังจะไหลออกถูกยกเลิกไปโดยสิ้นเชิง | ฟังก์ชันคืนค่า -1 โดย except ชั้นนอก (บรรทัด 10) ไม่ได้ทำงานเลย |
| 16 | `read_sensor(0)` | บรรทัด 6 คำนวณ `100 / 0` โยน `ZeroDivisionError` | เช่นเดียวกับกรณี -5 ข้อผิดพลาดนี้ถูกยกเลิกไปโดย `return` ใน finally เช่นกัน |

**คำอธิบายภาพรวม:** ทั้งสามการเรียก (บรรทัด 14–16) ให้ผลลัพธ์เป็น `-1` เหมือนกันหมด ทั้งที่ค่าที่ควรจะได้แตกต่างกันโดยสิ้นเชิง
ตามตรรกะของ try ชั้นใน (5.0 สำหรับกรณีแรก ควรจะเป็นข้อผิดพลาดที่ถูกดักสำหรับอีกสองกรณี) สาเหตุคือ `finally` ที่บรรทัด 7–9
มีคำสั่ง `return -1` อยู่ภายใน ซึ่งบังคับให้ฟังก์ชันคืนค่า -1 เสมอไม่ว่าจะเกิดอะไรขึ้นใน try ชั้นในก่อนหน้านั้นก็ตาม

**จุดสังเกต:** ผลลัพธ์ที่ได้ไม่ตรงกับสัญชาตญาณอย่างมาก โดยเฉพาะกรณี `read_sensor(-5)` และ `read_sensor(0)` ที่ควรจะเห็น
ข้อความ `"จับข้อผิดพลาดที่ except ชั้นนอก: ..."` จากบรรทัด 11 แต่กลับไม่ปรากฏเลยในผลลัพธ์จริง เพราะ `return` ที่อยู่ใน
`finally` ไม่เพียงแทนที่ค่าที่ `return` จาก try เท่านั้น แต่ยัง **ยกเลิกข้อผิดพลาดที่กำลังไหลออกจาก try ชั้นในโดยสิ้นเชิง**
ไม่ว่าข้อผิดพลาดนั้นจะเป็นข้อผิดพลาดที่โยนเองด้วย `raise` (กรณี -5) หรือข้อผิดพลาดที่เกิดขึ้นเองจากการทำงานของภาษา
(กรณี 0 ที่คำนวณ `100 / 0`) ก็ตาม กล่าวคือ `except ValueError` ที่บรรทัด 10 ไม่มีโอกาสได้ทำงานเลยในทั้งสองกรณีนี้
เพราะ `finally` "แย่งควบคุม" การไหลออกจากฟังก์ชันไปก่อนที่ข้อผิดพลาดจะไปถึงบล็อกดักข้อผิดพลาดของ try ชั้นนอก

บล็อกดักข้อผิดพลาด (except) ในบางภาษาโปรแกรมมิ่งสามารถเขียนได้โดยไม่ระบุชนิดของข้อผิดพลาดที่ต้องการดักเลย เรียกว่าการดักข้อผิดพลาด
แบบไม่ระบุชนิด (bare except หรือ catch-all handler) รูปแบบนี้จะดักข้อผิดพลาด "ทุกชนิด" ที่เกิดขึ้นภายในส่วนทดลองรัน (try)
โดยไม่คัดกรองเลยว่าข้อผิดพลาดนั้นตรงกับสิ่งที่ผู้เขียนโปรแกรมตั้งใจจะจัดการหรือไม่

ในภาษา Python การเขียน `except:` โดยไม่ระบุชนิดใด ๆ ต่อท้าย จะดักข้อผิดพลาดได้กว้างกว่าแม้แต่ `except Exception`
(ตามลำดับชั้นในหัวข้อ 11) เพราะ `except:` ดักได้ถึงระดับ `BaseException` ซึ่งรวมถึงข้อผิดพลาดที่ปกติไม่ถือว่าเป็น
"ข้อผิดพลาดจากการทำงานผิดพลาดของโปรแกรม" เช่น `SystemExit` หรือ `KeyboardInterrupt` ด้วย กลไกนี้ทำงานโดยไม่ได้แยกแยะว่า
ข้อผิดพลาดที่เกิดขึ้นจริงคือชนิดที่คาดการณ์ไว้ล่วงหน้า (เช่นข้อมูลผู้ใช้ไม่ถูกต้อง) หรือเป็นข้อผิดพลาดที่เกิดจากข้อบกพร่องอื่นในโค้ดเอง
ที่ไม่เกี่ยวข้องกับเงื่อนไขที่ตั้งใจจะตรวจสอบเลยแม้แต่น้อย ผลที่ตามมาทางกลไกคือชนิดที่แท้จริงของข้อผิดพลาดจะไม่ปรากฏให้เห็นอีกต่อไป
หากโค้ดในบล็อก except ไม่ได้พิมพ์ `type(e).__name__` หรือข้อความของ `e` ออกมาโดยเฉพาะ

### ตัวอย่างที่ 1: การดักข้อผิดพลาดทุกชนิดด้วย `except:` เดียว (ตะกร้าสินค้า)

````python
%reset -f
cart = ["เสื้อ", "กางเกง"]

def show_item(index):
    try:
        print(f"สินค้าลำดับที่ {index}: {cart[index]}")
    except:
        print("เกิดข้อผิดพลาด")

show_item(0)
show_item(5)
show_item("1")
````

**Output:**
```
สินค้าลำดับที่ 0: เสื้อ
เกิดข้อผิดพลาด
เกิดข้อผิดพลาด
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | คำอธิบายเพิ่มเติม |
|---|---|---|---|
| 9 | `show_item(0)` | `cart[0] = "เสื้อ"` | ไม่มีข้อผิดพลาด แสดงผลตามปกติ |
| 10 | `show_item(5)` | `cart[5]` อยู่นอกขอบเขตของ list | โยน `IndexError` ดักได้ด้วย `except:` (บรรทัด 6) แสดง `"เกิดข้อผิดพลาด"` |
| 11 | `show_item("1")` | `cart["1"]` ใช้ index เป็น string | โยน `TypeError` ดักได้ด้วย `except:` เช่นกัน แสดงข้อความเดียวกันทุกประการ |

**คำอธิบายภาพรวม:** บรรทัด 6 เขียน `except:` โดยไม่ระบุชนิดข้อผิดพลาดใดเลย บรรทัด 10 ทำให้เกิด `IndexError`
(เพราะ index 5 อยู่นอกขอบเขตของ `cart` ที่มีเพียง 2 สมาชิก) และบรรทัด 11 ทำให้เกิด `TypeError` (เพราะ index ที่ส่งเข้ามา
เป็นสตริงแทนที่จะเป็นจำนวนเต็ม) ทั้งสองกรณีเป็นข้อผิดพลาดคนละชนิดกันโดยสิ้นเชิง แต่ `except:` ดักได้ทั้งคู่และแสดงข้อความ
`"เกิดข้อผิดพลาด"` เหมือนกันทุกตัวอักษร

**จุดสังเกต:** ผลลัพธ์ที่ได้จากบรรทัด 10 และ 11 เหมือนกันทุกประการ ทั้งที่สาเหตุที่แท้จริงต่างกันคนละชนิด (index เกินขอบเขต
กับ ชนิดข้อมูลของ index ผิด) หากไม่เปิดโค้ดต้นทางมาดู ผู้ใช้โปรแกรมหรือแม้แต่ผู้เขียนโปรแกรมเองในภายหลังจะไม่สามารถทราบได้เลยว่า
เกิดอะไรขึ้นจริง ๆ จากข้อความที่แสดงเพียงอย่างเดียว

### ตัวอย่างที่ 2: การใช้งานจริง — ประมวลผลชุดข้อมูลโดยไม่ให้รายการที่มีปัญหาทำให้ทั้งชุดหยุดทำงาน (นับคะแนนจากไฟล์บันทึก)

````python
%reset -f
records = ["25", "thirty", "40", ""]

def process_batch(records):
    total = 0
    error_count = 0
    for text in records:
        try:
            total += int(text)
        except:
            error_count += 1
    return total, error_count

total, error_count = process_batch(records)
print("ผลรวม:", total)
print("จำนวนรายการที่มีปัญหา:", error_count)
````

**Output:**
```
ผลรวม: 65
จำนวนรายการที่มีปัญหา: 2
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | คำอธิบายเพิ่มเติม |
|---|---|---|---|
| 6–8 | วนซ้ำ `"25"` | `int("25") = 25` | บวกเข้า `total` สำเร็จ `total = 25` |
| 6–8 | วนซ้ำ `"thirty"` | `int("thirty")` โยน `ValueError` | ดักด้วย `except:` (บรรทัด 9) เพิ่ม `error_count` เป็น 1 |
| 6–8 | วนซ้ำ `"40"` | `int("40") = 40` | บวกเข้า `total` สำเร็จ `total = 65` |
| 6–8 | วนซ้ำ `""` | `int("")` โยน `ValueError` | ดักด้วย `except:` เช่นกัน เพิ่ม `error_count` เป็น 2 |
| 11 | `return total, error_count` | คืนค่าสองค่าพร้อมกันเป็น tuple `(65, 2)` | ผู้เรียกใช้ tuple unpacking ที่บรรทัด 13 เพื่อแยกรับสองค่า |

**คำอธิบายภาพรวม:** ฟังก์ชัน `process_batch` วนอ่านรายการข้อความทีละตัว พยายามแปลงเป็นจำนวนเต็มและบวกสะสมไว้ที่ `total`
หากรายการใดแปลงไม่ได้ (เช่น `"thirty"` หรือสตริงว่าง) `except:` ที่บรรทัด 9 จะดักไว้และนับเป็นรายการที่มีปัญหาแทนที่จะปล่อยให้
โปรแกรมทั้งหมดหยุดทำงาน ทำให้การประมวลผลรายการที่เหลือดำเนินต่อไปได้จนครบ บรรทัด 13–15 แสดงผลรวมและจำนวนรายการที่มีปัญหา

**จุดสังเกต:** ในบริบทนี้ `except:` แบบไม่ระบุชนิดถูกใช้เพื่อจุดประสงค์เฉพาะคือ "ไม่ให้รายการใดรายการหนึ่งที่มีปัญหาหยุดการ
ประมวลผลทั้งชุด" ซึ่งเป็นจุดประสงค์ที่ชัดเจนและจำกัดขอบเขต แต่ผลข้างเคียงทางกลไกยังคงเหมือนตัวอย่างที่ 1 คือไม่มีทางทราบจากผลลัพธ์
ว่ารายการที่นับเป็น "มีปัญหา" ทั้งสองรายการนั้นล้มเหลวด้วยสาเหตุเดียวกันหรือคนละสาเหตุกัน (ในที่นี้ทั้งสองบังเอิญเป็น `ValueError`
เหมือนกัน แต่โครงสร้างโค้ดไม่ได้ยืนยันสิ่งนี้เลย หากมีสาเหตุอื่นปะปนมาก็จะถูกนับรวมด้วยเช่นกัน)

### ตัวอย่างที่ 3: ผลของ `except:` ต่อข้อผิดพลาดที่เกิดจากข้อบกพร่องในโค้ดเอง (บันทึกค่าจากเซนเซอร์)

````python
%reset -f
def log_reading(sensor_value):
    try:
        celsius = float(sensor_value)
        avarage = celsius / 2  # พิมพ์ผิดตั้งใจ (ประกาศตัวแปรผิดชื่อ)
        print(f"ค่าเฉลี่ยที่คำนวณได้: {average}")
    except:
        print(f"อ่านค่าจากเซนเซอร์ไม่สำเร็จ: {sensor_value}")

log_reading("abc")
log_reading("40")
````

**Output:**
```
อ่านค่าจากเซนเซอร์ไม่สำเร็จ: abc
อ่านค่าจากเซนเซอร์ไม่สำเร็จ: 40
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | คำอธิบายเพิ่มเติม |
|---|---|---|---|
| 9 | `log_reading("abc")` | `float("abc")` โยน `ValueError` ที่บรรทัด 3 | ดักด้วย `except:` แสดงข้อความว่าอ่านค่าจากเซนเซอร์ไม่สำเร็จ |
| 10 | `log_reading("40")` | `float("40") = 40.0` สำเร็จ, บรรทัด 4 กำหนดตัวแปรชื่อ `avarage` (พิมพ์ผิด) | บรรทัด 5 อ้างถึง `average` ซึ่งไม่เคยถูกกำหนดค่า จึงโยน `NameError` |
| 6–7 | `except:` ดักได้ทั้งสองกรณี | ข้อความที่แสดงเหมือนกันทุกประการ | ทั้งที่กรณีที่สองไม่เกี่ยวข้องกับ "การอ่านค่าจากเซนเซอร์" เลย แต่เป็นข้อบกพร่องในโค้ด |

**คำอธิบายภาพรวม:** บรรทัด 9 ส่งค่าที่แปลงเป็นจำนวนจริงไม่ได้ ("abc") ทำให้เกิด `ValueError` ซึ่งเป็นกรณีที่ข้อความ
`"อ่านค่าจากเซนเซอร์ไม่สำเร็จ"` สื่อความหมายได้ตรงประเด็น แต่บรรทัด 10 ส่งค่าที่ถูกต้องสมบูรณ์ ("40") ปัญหากลับเกิดจาก
การพิมพ์ชื่อตัวแปรผิดที่บรรทัด 4 (`avarage` แทน `average`) ทำให้บรรทัด 5 อ้างถึงตัวแปรที่ไม่เคยมีอยู่จริงและโยน `NameError`
ซึ่ง `except:` ที่บรรทัด 6 ก็ดักไว้และแสดงข้อความเดียวกันกับกรณีแรกทุกประการ

**จุดสังเกต:** ข้อความ `"อ่านค่าจากเซนเซอร์ไม่สำเร็จ: 40"` ที่ปรากฏจริงในผลลัพธ์บรรทัดที่สองนั้นทำให้เข้าใจผิดได้ว่าเซนเซอร์
หรือข้อมูลนำเข้ามีปัญหา ทั้งที่ค่า `"40"` ที่ส่งเข้ามาถูกต้องสมบูรณ์และแปลงเป็นจำนวนจริงสำเร็จตั้งแต่บรรทัด 3 สาเหตุที่แท้จริงคือ
ข้อบกพร่องในโค้ดเอง (การพิมพ์ชื่อตัวแปรผิดที่บรรทัด 4) ซึ่งจะเกิดขึ้นซ้ำทุกครั้งที่ฟังก์ชันนี้ถูกเรียกไม่ว่าจะส่งค่าใดเข้ามาก็ตาม
`except:` ทำให้ข้อบกพร่องนี้ถูกซ่อนอยู่เบื้องหลังข้อความที่ฟังดูเหมือนเป็นปัญหาจากข้อมูลนำเข้าแทน หากต้องการทราบชนิดที่แท้จริง
จำเป็นต้องแก้ไขโค้ดชั่วคราวให้พิมพ์ `type(e).__name__` หรือใช้ `except Exception as e` แล้วพิมพ์ `e` ออกมาโดยเฉพาะ
(ดูหัวข้อ 8 ใน `lecture_note_intermediate.md`) จึงจะเห็นความแตกต่างระหว่างสองกรณีนี้ได้

**เชื่อมโยงกับหัวข้ออื่น:** ต่อยอดจากหัวข้อ 4 (จับหลาย exception) ใน `lecture_note_basic.md` และหัวข้อ 8
(`except X as e` พร้อมข้อความจริง) ใน `lecture_note_intermediate.md` เกี่ยวข้องโดยตรงกับหัวข้อ 11 (ลำดับชั้นของข้อผิดพลาด)
ในไฟล์นี้ เพราะ `except:` ดักกว้างกว่า `except Exception` อีกขั้นหนึ่งตามลำดับชั้นที่อธิบายไว้ในหัวข้อนั้น

## 15: การโยนข้อผิดพลาดซ้ำและการเชื่อมโยงข้อผิดพลาด (Re-raise / Exception Chaining)

**แหล่งอ้างอิง**: ไม่มีหน้าที่ตรงประเด็นในสามแหล่งอ้างอิงหลักของเอกสารชุดนี้ (W3Schools, GeeksforGeeks, TutorialsPoint)
หัวข้อนี้เรียบเรียงขึ้นเองจากความรู้ทั่วไปด้านวิศวกรรมซอฟต์แวร์และเอกสารภาษาโปรแกรมมิ่งเรื่องกลไกการโยนข้อผิดพลาดซ้ำ ระบุไว้ชัดเจน
ตามข้อเท็จจริงนี้แทนการอ้างอิงแหล่งใดแหล่งหนึ่งที่ไม่มีอยู่จริง

### แนวคิด

เมื่อบล็อกดักข้อผิดพลาด (except) ทำงานบางอย่างกับข้อผิดพลาดที่ดักได้ (เช่นบันทึก log) แล้วต้องการให้ข้อผิดพลาดนั้นยังคงไหลต่อไปยัง
ผู้เรียกระดับที่สูงกว่าเหมือนกับไม่ได้ถูกดักไว้เลย เรียกว่าการโยนข้อผิดพลาดซ้ำ (re-raise) แนวคิดนี้ต่างจากการดักแล้วจบเรื่องไปเลย
ตรงที่จุดประสงค์คือ "แทรกการทำงานบางอย่างระหว่างทาง" โดยไม่เปลี่ยนพฤติกรรมปลายทางของโปรแกรมว่าข้อผิดพลาดนี้ยังคงต้องถูกจัดการ
โดยผู้เรียกระดับบนอยู่ดี

อีกแนวคิดที่เกี่ยวข้องกันคือการเชื่อมโยงข้อผิดพลาด (exception chaining) ซึ่งเกิดขึ้นเมื่อขั้นตอนการจัดการข้อผิดพลาดหนึ่งต้องการ
"แปลง" ข้อผิดพลาดระดับล่าง (เช่นข้อผิดพลาดจากการแปลงชนิดข้อมูล) ให้กลายเป็นข้อผิดพลาดที่มีความหมายเฉพาะกับโดเมนของโปรแกรมมากขึ้น
(เช่นข้อผิดพลาดที่ผู้ใช้กำหนดเองตามหัวข้อ 10) โดยยังคงรักษาข้อมูลของข้อผิดพลาดต้นทางไว้ ไม่ให้สูญหายไปเมื่อโยนข้อผิดพลาดใหม่ขึ้นมา
แทนที่

ในภาษา Python มีรูปแบบการโยนข้อผิดพลาดซ้ำสองแบบหลัก: การเขียน `raise` เพียงคำเดียวโดยไม่ระบุสิ่งใดต่อท้ายภายในบล็อก except
จะโยนข้อผิดพลาดตัวเดิมที่กำลังถูกจัดการอยู่ซ้ำออกไปทั้งหมด ส่วนการเขียน `raise ExceptionใหมX(...) from เดิม` จะโยนข้อผิดพลาด
ชนิดใหม่ขึ้นมาแทนที่ พร้อมแนบข้อผิดพลาดเดิมไว้ในแอตทริบิวต์ `__cause__` ของข้อผิดพลาดใหม่ ทำให้ข้อความแสดงลำดับข้อผิดพลาด
(traceback) แสดงทั้งข้อผิดพลาดเดิมและข้อผิดพลาดใหม่ต่อเนื่องกัน และหากเขียน `raise Exceptionใหม่(...) from None` แทน
จะระงับการแสดงข้อผิดพลาดเดิมในข้อความแสดงลำดับข้อผิดพลาดโดยเจตนา เชื่อมโยงกับหัวข้อ 9 (`raise`) ใน `lecture_note_intermediate.md`
และหัวข้อ 10 (custom exception) ในไฟล์นี้

### ตัวอย่างที่ 1: การโยนข้อผิดพลาดซ้ำแบบพื้นฐานหลังบันทึก log (ฟังก์ชันหารปลอดภัยพร้อมบันทึกผล)

````python
%reset -f
def safe_divide_logged(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print(f"บันทึก log: พบการหารด้วยศูนย์ (a={a}, b={b})")
        raise

try:
    safe_divide_logged(10, 0)
except ZeroDivisionError:
    print("โปรแกรมหลักจับข้อผิดพลาดได้อีกครั้งหลังถูกโยนซ้ำ")

print(safe_divide_logged(10, 2))
````

**Output:**
```
บันทึก log: พบการหารด้วยศูนย์ (a=10, b=0)
โปรแกรมหลักจับข้อผิดพลาดได้อีกครั้งหลังถูกโยนซ้ำ
5.0
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | คำอธิบายเพิ่มเติม |
|---|---|---|---|
| 9 | `safe_divide_logged(10, 0)` | บรรทัด 3 โยน `ZeroDivisionError` | บล็อก except ภายในฟังก์ชัน (บรรทัด 4) ดักไว้ |
| 5 | `print("บันทึก log: ...")` | พิมพ์ข้อความบันทึก | ทำงานก่อนที่จะโยนข้อผิดพลาดซ้ำ |
| 6 | `raise` (ไม่มีสิ่งใดต่อท้าย) | โยน `ZeroDivisionError` ตัวเดิมซ้ำออกไป | ข้อผิดพลาดยังคงเป็นชนิดและข้อความเดิมทุกประการ ไม่ได้สร้างข้อผิดพลาดใหม่ |
| 10–11 | `except ZeroDivisionError` ที่ระดับโปรแกรมหลัก | ดักข้อผิดพลาดที่ถูกโยนซ้ำได้ | ยืนยันว่าข้อผิดพลาดไหลออกจากฟังก์ชันไปถึงผู้เรียกได้จริงแม้จะถูกดักและประมวลผลบางส่วนไปแล้วในฟังก์ชัน |
| 13 | `safe_divide_logged(10, 2)` | ไม่มีข้อผิดพลาด | คืนค่า 5.0 ตามปกติ |

**คำอธิบายภาพรวม:** ฟังก์ชัน `safe_divide_logged` ดักข้อผิดพลาด `ZeroDivisionError` ไว้ที่บรรทัด 4 เพื่อบันทึก log
ก่อนที่บรรทัด 6 จะใช้ `raise` เพียงคำเดียวโยนข้อผิดพลาดตัวเดิมออกไปซ้ำอีกครั้ง ทำให้ผู้เรียกฟังก์ชันนี้ (บรรทัด 8–11)
ยังคงต้องดักข้อผิดพลาดเองอยู่ดี เสมือนว่าฟังก์ชันไม่ได้ดักข้อผิดพลาดไว้เลยในแง่ของ "ผลลัพธ์สุดท้ายที่ผู้เรียกเห็น" ต่างกันเพียง
มีการบันทึก log แทรกอยู่ระหว่างทางเท่านั้น บรรทัด 13 แสดงกรณีปกติที่ไม่มีข้อผิดพลาดเกิดขึ้นเลย

**จุดสังเกต:** คำสั่ง `raise` ที่ไม่มีสิ่งใดต่อท้ายนั้นใช้ได้เฉพาะ **ภายในบล็อก except ที่กำลังจัดการข้อผิดพลาดอยู่เท่านั้น**
เพราะ Python ต้องอาศัยบริบทของข้อผิดพลาดที่ "กำลังถูกจัดการอยู่ในขณะนั้น" เป็นตัวอ้างอิงว่าจะโยนอะไรซ้ำออกไป ข้อผิดพลาดที่ถูก
โยนซ้ำด้วยวิธีนี้ยังคงเป็นอ็อบเจกต์เดิมทุกประการ (ชนิดเดิม ข้อความเดิม) ต่างจากการสร้างข้อผิดพลาดใหม่ขึ้นมาโยนแทนตามตัวอย่างที่ 2

### ตัวอย่างที่ 2: การใช้งานจริง — แปลงข้อผิดพลาดระดับล่างเป็นข้อผิดพลาดเฉพาะโดเมนด้วย `from` (ตัวโหลดค่าตั้งค่า)

````python
%reset -f
class ConfigError(Exception):
    pass

def load_timeout(config):
    try:
        return int(config["timeout"])
    except ValueError as e:
        raise ConfigError(f"ค่า timeout ในไฟล์ตั้งค่าไม่ถูกต้อง: {config['timeout']!r}") from e

try:
    load_timeout({"timeout": "สิบวินาที"})
except ConfigError as e:
    print("โหลดค่าตั้งค่าไม่สำเร็จ:", e)
    print("สาเหตุเดิม (__cause__):", repr(e.__cause__))
````

**Output:**
```
โหลดค่าตั้งค่าไม่สำเร็จ: ค่า timeout ในไฟล์ตั้งค่าไม่ถูกต้อง: 'สิบวินาที'
สาเหตุเดิม (__cause__): ValueError("invalid literal for int() with base 10: 'สิบวินาที'")
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | คำอธิบายเพิ่มเติม |
|---|---|---|---|
| 6 | `int(config["timeout"])` | `int("สิบวินาที")` โยน `ValueError` | ค่าที่เก็บในไฟล์ตั้งค่าไม่ใช่ตัวเลข |
| 7 | `except ValueError as e` | `e` = อ็อบเจกต์ `ValueError` เดิม | ดักไว้เพื่อแปลงเป็นข้อผิดพลาดเฉพาะโดเมน |
| 8 | `raise ConfigError(...) from e` | สร้างอ็อบเจกต์ `ConfigError` ใหม่ | แนบ `e` (ValueError เดิม) ไว้ที่แอตทริบิวต์ `__cause__` ของ `ConfigError` ใหม่นี้ |
| 12–13 | `except ConfigError as e` | `e` ตัวใหม่ = อ็อบเจกต์ `ConfigError` | ผู้เรียกเห็นเฉพาะ `ConfigError` ที่มีความหมายตรงกับโดเมน (การตั้งค่า) ไม่ใช่ `ValueError` ดิบ |
| 14 | `e.__cause__` | อ็อบเจกต์ `ValueError` เดิมที่แนบไว้ | ยังสามารถย้อนไปดูสาเหตุดั้งเดิมได้ครบถ้วนแม้ผู้เรียกจะดัก `ConfigError` เท่านั้น |

**คำอธิบายภาพรวม:** ฟังก์ชัน `load_timeout` ดัก `ValueError` ที่เกิดจากการแปลงค่าที่อ่านจากไฟล์ตั้งค่าเป็นจำนวนเต็มไม่สำเร็จ
(บรรทัด 6–7) แล้วโยนข้อผิดพลาดชนิดใหม่ `ConfigError` ขึ้นมาแทนที่ด้วย `raise ... from e` (บรรทัด 8) ซึ่งมีความหมาย
ตรงกับโดเมนของฟังก์ชันนี้มากกว่า (ปัญหาการตั้งค่า ไม่ใช่ปัญหาการแปลงชนิดข้อมูลทั่วไป) ผู้เรียกที่บรรทัด 10–14 จึงดักเพียง
`ConfigError` เท่านั้นโดยไม่จำเป็นต้องรู้ว่าภายในฟังก์ชันใช้ `int()` แปลงค่าอยู่เลย แต่ยังคงเข้าถึงสาเหตุดั้งเดิม (`ValueError`)
ได้ผ่าน `e.__cause__` หากต้องการตรวจสอบคำอธิบายเพิ่มเติม

**จุดสังเกต:** การใช้ `from e` ไม่ได้ทำให้ข้อมูลของข้อผิดพลาดเดิม (`ValueError`) หายไปเลย เพียงแต่ "ซ่อน" มันไว้เบื้องหลัง
ข้อผิดพลาดใหม่ที่มีความหมายตรงประเด็นกว่าสำหรับผู้เรียกทั่วไป ในขณะที่ยังเปิดทางให้ผู้ที่ต้องการรายละเอียดเชิงลึก (เช่นระหว่างการ
แก้ไขข้อบกพร่อง) เข้าถึงผ่าน `__cause__` ได้เสมอ ต่างจากการจับ `ValueError` แล้วโยน `ConfigError(...)` เฉย ๆ โดยไม่มี
`from e` ซึ่งจะทำให้ `__cause__` เป็น `None` และเชื่อมโยงกันเฉพาะทางอ้อมผ่าน `__context__` เท่านั้น (ดูตัวอย่างที่ 3
ซึ่งใช้ `from None` เพื่อระงับการเชื่อมโยงนี้ไปอีกขั้นหนึ่ง)

### ตัวอย่างที่ 3: ผลของ `from None` ต่อข้อความแสดงลำดับข้อผิดพลาด (ตรวจสอบชื่อผู้ใช้ตอนสมัครสมาชิก)

````python
%reset -f
class SignupError(Exception):
    pass

def validate_username(username):
    try:
        if len(username) < 3:
            raise ValueError("ชื่อผู้ใช้สั้นเกินไป")
    except ValueError as e:
        raise SignupError("ไม่สามารถลงทะเบียนได้") from None

validate_username("ab")
````

**Output:**
```
Traceback (most recent call last):
  File "t15_ex3.py", line 11, in <module>
    validate_username("ab")
  File "t15_ex3.py", line 9, in validate_username
    raise SignupError("ไม่สามารถลงทะเบียนได้") from None
SignupError: ไม่สามารถลงทะเบียนได้
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | คำอธิบายเพิ่มเติม |
|---|---|---|---|
| 11 | `validate_username("ab")` | `len("ab") = 2 < 3` | เงื่อนไขบรรทัด 6 เป็นจริง โยน `ValueError` ที่บรรทัด 7 |
| 8 | `except ValueError as e` | ดักได้ | เตรียมโยนข้อผิดพลาดใหม่แทนที่ |
| 9 | `raise SignupError(...) from None` | สร้าง `SignupError` ใหม่ พร้อมระบุ `from None` | `__cause__` ของ `SignupError` ถูกกำหนดเป็น `None` อย่างชัดเจน ระงับการเชื่อมโยงกับ `ValueError` เดิม |
| — | ไม่มีบล็อก try/except ครอบการเรียกที่บรรทัด 11 | โปรแกรมหยุดทำงาน | ข้อความแสดงลำดับข้อผิดพลาดแสดงเฉพาะ `SignupError` เพียงชนิดเดียว ไม่มี `ValueError` ปรากฏเลย |

**คำอธิบายภาพรวม:** โครงสร้างของตัวอย่างนี้เหมือนกับตัวอย่างที่ 2 (ดัก `ValueError` แล้วโยนข้อผิดพลาดชนิดใหม่แทนที่) แต่เปลี่ยน
จาก `from e` เป็น `from None` ที่บรรทัด 9 เมื่อเรียกด้วยชื่อผู้ใช้ที่สั้นเกินไป ("ab" มีความยาว 2 ตัวอักษร) โดยไม่มีบล็อก
ดักข้อผิดพลาดใดครอบการเรียกไว้เลย โปรแกรมหยุดทำงานและแสดงข้อความแสดงลำดับข้อผิดพลาดที่มีเพียง `SignupError` ชนิดเดียว
โดยไม่มีร่องรอยของ `ValueError` เดิมปรากฏอยู่เลยแม้แต่น้อย

**จุดสังเกต:** หากเปลี่ยนบรรทัด 9 จาก `raise SignupError("ไม่สามารถลงทะเบียนได้") from None` เป็น
`raise SignupError("ไม่สามารถลงทะเบียนได้")` เฉย ๆ (ไม่มี `from` ใด ๆ ต่อท้ายเลย) ข้อความแสดงลำดับข้อผิดพลาดที่ได้จริงจะ
เปลี่ยนไปทันที คือจะแสดง**ทั้งสอง** traceback ต่อเนื่องกัน คั่นด้วยข้อความ `"During handling of the above exception,
another exception occurred:"` ทำให้เห็นทั้ง `ValueError` เดิมและ `SignupError` ใหม่พร้อมกัน (ยืนยันด้วยการรันไฟล์เปรียบเทียบ
จริงที่ตัดเฉพาะ `from None` ออก) ความแตกต่างนี้เกิดจาก Python บันทึกข้อผิดพลาดที่ "กำลังถูกจัดการอยู่" ไว้ในแอตทริบิวต์
`__context__` ของข้อผิดพลาดใหม่โดยอัตโนมัติทุกครั้งที่มีการโยนข้อผิดพลาดใหม่ขึ้นระหว่างอยู่ในบล็อก except (ไม่ว่าจะเขียน `from`
หรือไม่ก็ตาม) การระบุ `from None` เป็นการสั่งให้ Python ไม่แสดง `__context__` นั้นในข้อความแสดงลำดับข้อผิดพลาด แม้ว่า
ข้อผิดพลาดเดิมจะยังคงถูกบันทึกไว้ใน `__context__` ของอ็อบเจกต์อยู่ดีก็ตาม เพียงแต่ไม่ถูกนำมาแสดงเมื่อพิมพ์ traceback ออกมา

**เชื่อมโยงกับหัวข้ออื่น:** ต่อยอดจากหัวข้อ 9 (`raise`) ใน `lecture_note_intermediate.md` และหัวข้อ 10
(custom exception) ในไฟล์นี้ ซึ่งมักถูกใช้ร่วมกับการเชื่อมโยงข้อผิดพลาดเพื่อแปลงข้อผิดพลาดระดับล่างเป็นข้อผิดพลาดเฉพาะโดเมน
เกี่ยวข้องกับหัวข้อ 13 (nested try/except/finally) ในไฟล์นี้ ในแง่ที่ทั้งคู่กล่าวถึงกลไกที่กำหนดว่าข้อผิดพลาดใดจะไปถึงผู้เรียก
ในท้ายที่สุด

### หัวข้อที่ 10: ข้อผิดพลาดที่ผู้ใช้กำหนดเอง (Custom Exception) — ข้อ 1–4

**ข้อ 1 (อธิบายแนวคิด)** จงอธิบายเหตุผลที่การสร้างข้อผิดพลาดที่ผู้ใช้กำหนดเอง (custom exception) ควรสืบทอด
จากคลาส `Exception` แทนที่จะสืบทอดจาก `BaseException` โดยตรง โดยอ้างอิงลำดับชั้นของข้อผิดพลาดที่กล่าวถึง
ในหัวข้อ 11

**ข้อ 2 (ทายผลลัพธ์)** กำหนดโค้ดต่อไปนี้สำหรับระบบจองห้องประชุม จงทายผลลัพธ์ที่จะปรากฏบนหน้าจอ
```python
class RoomBookingError(Exception):
    def __init__(self, room, day):
        self.room = room
        self.day = day
        super().__init__(f"ห้อง {room} ถูกจองแล้วในวัน {day}")

booked = {("A101", "จันทร์"), ("A101", "อังคาร")}

def book_room(room, day):
    if (room, day) in booked:
        raise RoomBookingError(room, day)
    booked.add((room, day))
    return f"จองห้อง {room} วัน {day} สำเร็จ"

try:
    print(book_room("A101", "จันทร์"))
except RoomBookingError as e:
    print("จองไม่สำเร็จ:", e)
    print("ห้องที่ขัดข้อง:", e.room)

print(book_room("A101", "พุธ"))
```

**ข้อ 3 (หาข้อผิดพลาด)** โค้ดต่อไปนี้ตั้งใจจะดักข้อผิดพลาดที่ผู้ใช้กำหนดเองไว้สำหรับระบบลงทะเบียนสอบ แต่กลับ
ทำให้โปรแกรมหยุดทำงานด้วยข้อผิดพลาดอีกชนิดหนึ่งแทน จงระบุว่าปัญหาอยู่ที่บรรทัดใด และเพราะเหตุใด
```python
class RegistrationError(Exception):
    pass

def register_exam(seat_left):
    if seat_left <= 0:
        raise RegistrationError("ที่นั่งสอบเต็มแล้ว")
    return "ลงทะเบียนสำเร็จ"

try:
    print(register_exam(0))
except RegistrationErorr:
    print("ลงทะเบียนไม่สำเร็จ")
```

**ข้อ 4 (เขียนโค้ด)** จงสร้างข้อผิดพลาดที่ผู้ใช้กำหนดเองชื่อ `TicketSoldOutError` สำหรับระบบจองตั๋วภาพยนตร์
ให้เก็บชื่อภาพยนตร์และจำนวนที่นั่งคงเหลือไว้เป็นแอตทริบิวต์ของตนเอง แล้วเขียนฟังก์ชัน `book_ticket(movie)`
ที่โยนข้อผิดพลาดนี้เมื่อที่นั่งคงเหลือเป็นศูนย์ จากนั้นทดสอบเรียกใช้ด้วยภาพยนตร์ที่ที่นั่งเต็มแล้วและภาพยนตร์ที่ยัง
มีที่นั่งว่างอยู่

### หัวข้อที่ 11: ลำดับชั้นของข้อผิดพลาด (Exception Hierarchy) — ข้อ 5–7

**ข้อ 5 (อธิบายแนวคิด)** จงอธิบายว่าเหตุใดการเขียน `except LookupError` เพียงบล็อกเดียวจึงสามารถดักข้อผิดพลาด
ได้ทั้ง `IndexError` และ `KeyError` พร้อมระบุความสัมพันธ์เชิงลำดับชั้นระหว่างคลาสทั้งสามนี้

**ข้อ 6 (ทายผลลัพธ์)** กำหนดโค้ดต่อไปนี้สำหรับระบบเช็คสต็อกสินค้าตามสาขา จงทายผลลัพธ์ที่จะปรากฏบนหน้าจอ
```python
warehouse = {"เสื้อ": [10, 20], "กางเกง": [5]}

def get_branch_stock(item, branch_index):
    try:
        return warehouse[item][branch_index]
    except LookupError as e:
        return f"ไม่พบข้อมูล ({type(e).__name__}): {e}"

print(get_branch_stock("เสื้อ", 1))
print(get_branch_stock("กางเกง", 3))
print(get_branch_stock("หมวก", 0))
```

**ข้อ 7 (หาข้อผิดพลาด)** โค้ดต่อไปนี้สำหรับระบบแปลงหน่วยน้ำหนักตั้งใจแยกดักข้อผิดพลาดเป็นสองกรณี (กรณีทั่วไป
และกรณี `ValueError` โดยเฉพาะ) แต่กรณีเฉพาะกลับไม่เคยทำงานเลยไม่ว่าจะป้อนค่าใดก็ตาม จงระบุว่าปัญหาอยู่ตรงไหน
พร้อมอธิบายเหตุผลโดยอ้างอิงลำดับชั้นของข้อผิดพลาด
```python
def convert_weight(text):
    try:
        return float(text)
    except Exception as e:
        return f"ข้อผิดพลาดทั่วไป: {e}"
    except ValueError:
        return "ข้อความไม่ใช่ตัวเลข"

print(convert_weight("หนักมาก"))
```

### หัวข้อที่ 12: การยืนยันเงื่อนไข (`assert` / `AssertionError`) — ข้อ 8–10

**ข้อ 8 (อธิบายแนวคิด)** จงอธิบายว่าเกิดอะไรขึ้นกับคำสั่ง `assert` ทั้งหมดในโปรแกรม เมื่อรันไพทอนด้วยตัวเลือก
โหมดที่ปรับให้กระชับขึ้น (`python -O`) และความเป็นไปนี้มีนัยอย่างไรต่อการเลือกใช้ `assert` เทียบกับการใช้ `raise`
ในการตรวจสอบข้อมูลนำเข้าจากภายนอกโปรแกรม

**ข้อ 9 (ทายผลลัพธ์)** กำหนดโค้ดต่อไปนี้สำหรับระบบสมัครสมาชิกฟิตเนส จงทายผลลัพธ์ของการเรียกใช้ทั้ง 2 ครั้ง
```python
def register_member(age):
    assert age >= 18, "ต้องมีอายุอย่างน้อย 18 ปีจึงจะสมัครสมาชิกได้"
    return "สมัครสมาชิกสำเร็จ"

print(register_member(25))
print(register_member(15))
```

**ข้อ 10 (เขียนโค้ด)** จงเขียนฟังก์ชันชื่อ `request_supply(requested, available)` สำหรับระบบเบิกวัสดุสำนักงาน
ที่ใช้ `assert` ยืนยันว่าจำนวนที่ขอเบิก (`requested`) ต้องไม่มากกว่าจำนวนที่มีอยู่ในคลัง (`available`) พร้อมข้อความ
อธิบายเงื่อนไข แล้วทดสอบเรียกใช้ด้วยค่า `(5, 10)` และ `(20, 10)`

### หัวข้อที่ 13: การซ้อนบล็อก try/except/finally (Nested try/except/finally) — ข้อ 11–13

**ข้อ 11 (อธิบายแนวคิด)** กำหนดโครงสร้าง try ซ้อนกันสองชั้น โดยชั้นในมี `finally` กำกับอยู่แต่ไม่มีบล็อก `except`
ที่ตรงกับข้อผิดพลาดที่เกิดขึ้น ส่วนชั้นนอกมีบล็อก `except` ที่ดักข้อผิดพลาดชนิดนั้นได้พอดี จงอธิบายลำดับการทำงาน
ตั้งแต่จุดเกิดข้อผิดพลาดจนถึงจุดที่โปรแกรมดำเนินต่อได้ตามปกติ โดยระบุว่า `finally` ของชั้นในทำงานเมื่อใดเทียบกับ
`except` ของชั้นนอก

**ข้อ 12 (หาข้อผิดพลาด)** โค้ดต่อไปนี้สำหรับระบบเช็คอินโรงแรมทำงานถูกต้องเมื่อเรียกด้วยหมายเลขห้องที่ไม่มีอยู่จริง
แต่กลับทำให้โปรแกรมหยุดทำงานเมื่อเรียกด้วยหมายเลขห้องที่มีการจองซ้ำ จงระบุว่าบรรทัดใดเป็นสาเหตุ และเพราะเหตุใด
บล็อก `except` ทั้งสองชั้นจึงไม่สามารถดักข้อผิดพลาดของกรณีหลังได้
```python
reservations = {"101": "จองแล้ว", "102": "ว่าง"}

def check_in(room):
    try:
        try:
            status = reservations[room]
            if status == "จองแล้ว":
                raise ValueError(f"ห้อง {room} มีการจองซ้ำ")
        except KeyError:
            return "ไม่พบหมายเลขห้องนี้"
    except KeyError:
        return "เกิดปัญหาการจองซ้ำ"
    return f"เช็คอินห้อง {room} สำเร็จ"

print(check_in("999"))
print(check_in("101"))
```

**ข้อ 13 (เขียนโค้ด)** จงเขียนฟังก์ชัน `withdraw(card, pin, amount)` สำหรับระบบถอนเงินตู้ ATM ที่มีโครงสร้าง
try ซ้อนกันสองชั้น: ชั้นในตรวจสอบว่าเลขบัตร (`card`) มีอยู่ในระบบหรือไม่ (พร้อม `finally` พิมพ์ข้อความยืนยันว่า
ตรวจสอบบัตรเสร็จสิ้นแล้ว) ส่วนชั้นนอกตรวจสอบรหัส PIN และยอดเงินคงเหลือด้วย `raise ValueError` แล้วดักด้วย
`except ValueError` และ `except KeyError` ตามลำดับ พร้อม `finally` ของชั้นนอกพิมพ์ข้อความปิดรายการเสมอ
ทดสอบเรียกใช้อย่างน้อย 3 กรณี (ถอนสำเร็จ, เลขบัตรไม่มีในระบบ, รหัส PIN ผิด)

### หัวข้อที่ 14: การดักข้อผิดพลาดแบบไม่ระบุชนิด (Bare `except`) — ข้อ 14–16

**ข้อ 14 (ทายผลลัพธ์)** กำหนดโค้ดต่อไปนี้สำหรับระบบเช็คสถานะพัสดุ จงทายผลลัพธ์ที่จะปรากฏบนหน้าจอ
```python
tracking = {"TH001": "จัดส่งแล้ว", "TH002": "กำลังจัดส่ง"}

def check_status(code):
    try:
        return tracking[code]
    except:
        return "ไม่พบข้อมูลพัสดุ"

print(check_status("TH001"))
print(check_status("TH999"))
print(check_status(12345))
```

**ข้อ 15 (หาข้อผิดพลาด)** โค้ดต่อไปนี้สำหรับระบบคำนวณดอกเบี้ยเงินฝากให้ผลลัพธ์เป็นข้อความ "คำนวณดอกเบี้ย
ไม่สำเร็จ" ทุกครั้งไม่ว่าจะป้อนค่าใดก็ตาม จงระบุว่าสาเหตุที่แท้จริงคืออะไร และเพราะเหตุใดข้อความที่ได้จึงไม่ได้
สื่อถึงสาเหตุนั้นเลย
```python
def calculate_interest(principal, months):
    try:
        montly_rate = 0.01
        interest = principal * monthly_rate * months
        return interest
    except:
        return "คำนวณดอกเบี้ยไม่สำเร็จ"

print(calculate_interest(1000, 6))
print(calculate_interest("1000", 6))
```

**ข้อ 16 (เขียนโค้ด)** จงเขียนฟังก์ชัน `convert_currency(amount, rate)` สำหรับระบบแปลงสกุลเงิน โดยใช้บล็อก
`except` ที่ระบุชนิดข้อผิดพลาดเจาะจงอย่างน้อยสองชนิด (`TypeError` และ `ValueError`) แทนการดักข้อผิดพลาดทุก
ชนิดด้วย `except:` เพียงบล็อกเดียว แล้วทดสอบเรียกใช้ด้วยค่าที่ทำให้เกิด `TypeError` จริงอย่างน้อย 1 กรณี

### หัวข้อที่ 15: การโยนข้อผิดพลาดซ้ำและการเชื่อมโยงข้อผิดพลาด (Re-raise / Exception Chaining) — ข้อ 17–20

**ข้อ 17 (อธิบายแนวคิด)** จงอธิบายความแตกต่างของค่าที่ปรากฏในแอตทริบิวต์ `__cause__` ของข้อผิดพลาดใหม่
เมื่อเขียนโค้ดสามรูปแบบต่อไปนี้ภายในบล็อก except ที่กำลังดักข้อผิดพลาดเดิมอยู่: (1) `raise ข้อผิดพลาดใหม่(...)`
เฉย ๆ โดยไม่มี `from` ต่อท้าย (2) `raise ข้อผิดพลาดใหม่(...) from e` (3) `raise ข้อผิดพลาดใหม่(...) from None`

**ข้อ 18 (ทายผลลัพธ์)** กำหนดโค้ดต่อไปนี้สำหรับระบบยืนยันรหัส OTP จงทายผลลัพธ์ที่จะปรากฏบนหน้าจอ
```python
def verify_otp(code, expected):
    try:
        if code != expected:
            raise ValueError("รหัส OTP ไม่ถูกต้อง")
    except ValueError:
        print("บันทึก log ความพยายามยืนยัน OTP ที่ล้มเหลว")
        raise
    return "ยืนยันสำเร็จ"

try:
    verify_otp("1234", "5678")
except ValueError as e:
    print("ระบบหลักจับข้อผิดพลาดได้:", e)

print(verify_otp("1111", "1111"))
```

**ข้อ 19 (หาข้อผิดพลาด)** โค้ดต่อไปนี้สำหรับระบบตรวจสอบใบสมัครสินเชื่อตั้งใจจะเชื่อมโยงข้อผิดพลาดเดิมไว้กับ
ข้อผิดพลาดใหม่ แต่เมื่อตรวจสอบ `e.__cause__` ภายหลังกลับได้ค่า `None` ทั้งที่มีข้อผิดพลาดเดิมเกิดขึ้นจริงก่อนหน้า
จงระบุว่าโค้ดส่วนใดที่ทำให้เป็นเช่นนี้ และรูปแบบใดของคำสั่ง `raise` ที่จะเชื่อมโยงข้อผิดพลาดเดิมไว้ได้จริง
```python
class LoanValidationError(Exception):
    pass

def validate_income(income_text):
    try:
        income = float(income_text)
    except ValueError as e:
        raise LoanValidationError("ข้อมูลรายได้ไม่ถูกต้อง")
    return income

try:
    validate_income("สามหมื่น")
except LoanValidationError as e:
    print("ตรวจสอบไม่ผ่าน:", e)
    print("สาเหตุเดิม:", e.__cause__)
```

**ข้อ 20 (เขียนโค้ด)** จงเขียนข้อผิดพลาดที่ผู้ใช้กำหนดเองชื่อ `MissingFieldError` และฟังก์ชัน
`process_application(form)` สำหรับระบบตรวจสอบข้อมูลแบบฟอร์มสมัครงาน ที่ดัก `KeyError` จากการเข้าถึงข้อมูล
ในแบบฟอร์มที่ขาดหายไป แล้วโยน `MissingFieldError` แทนที่ด้วย `from e` เพื่อรักษาความเชื่อมโยงกับข้อผิดพลาด
เดิมไว้ จากนั้นทดสอบเรียกใช้ด้วยแบบฟอร์มที่ขาดข้อมูลตำแหน่งงาน (`position`) ไปพร้อมตรวจสอบค่า `e.__cause__`
ของข้อผิดพลาดที่ดักได้
