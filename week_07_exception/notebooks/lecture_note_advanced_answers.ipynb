### หัวข้อที่ 10: ข้อผิดพลาดที่ผู้ใช้กำหนดเอง (Custom Exception)

**เฉลยข้อ 1** (อธิบายแนวคิด)

ตามลำดับชั้นของข้อผิดพลาดที่อธิบายไว้ในหัวข้อ 11 คลาส `BaseException` เป็นรากของลำดับชั้นทั้งหมด และมีคลาสลูก
โดยตรงอยู่หลายคลาสที่ไม่ได้สื่อถึง "ข้อผิดพลาดจากการทำงานผิดพลาดของโปรแกรม" เช่น `SystemExit` (การสั่งจบ
โปรแกรมโดยเจตนา) และ `KeyboardInterrupt` (ผู้ใช้กดยกเลิกโปรแกรมกลางคัน) หากข้อผิดพลาดที่ผู้ใช้กำหนดเอง
สืบทอดจาก `BaseException` โดยตรง คำสั่ง `except Exception` ซึ่งเป็นรูปแบบการดักข้อผิดพลาดทั่วไปที่พบบ่อยที่สุด
จะไม่สามารถดักข้อผิดพลาดที่ผู้ใช้กำหนดเองนั้นได้เลย เพราะ `except Exception` ดักได้เฉพาะคลาสลูกของ
`Exception` เท่านั้น การสืบทอดจาก `Exception` จึงทำให้ข้อผิดพลาดที่ผู้ใช้กำหนดเองเข้ากันได้กับรูปแบบการดัก
ข้อผิดพลาดทั่วไปที่ใช้กันเป็นมาตรฐานในโปรแกรมทั่วไป

**เฉลยข้อ 2** (ทายผลลัพธ์)

คำอธิบาย: บรรทัดที่ 16 พยายามจองห้อง A101 ในวันจันทร์ ซึ่งอยู่ใน `booked` แล้ว จึงโยน `RoomBookingError`
ที่ดักได้ที่บรรทัด 17–19 และเข้าถึง `e.room` ได้โดยตรง บรรทัดที่ 21 จองห้อง A101 ในวันพุธซึ่งยังไม่เคยถูกจอง
จึงสำเร็จตามปกติ

```python
%reset -f
class RoomBookingError(Exception):
    def __init__(self, room, day):
        self.room = room
        self.day = day
        super().__init__(f"ห้อง {room} ถูกจองแล้วในวัน {day}")

booked = {("A101", "จันทร์"), ("A101", "อังคาร")}

def book_room(room, day):
    if (room, day) in booked:
        raise RoomBookingError(room, day)
    booked.add((room, day))
    return f"จองห้อง {room} วัน {day} สำเร็จ"

try:
    print(book_room("A101", "จันทร์"))
except RoomBookingError as e:
    print("จองไม่สำเร็จ:", e)
    print("ห้องที่ขัดข้อง:", e.room)

print(book_room("A101", "พุธ"))
```

**เฉลยข้อ 4** (เขียนโค้ด)



```python
%reset -f
class TicketSoldOutError(Exception):
    def __init__(self, movie, remaining):
        self.movie = movie
        self.remaining = remaining
        super().__init__(f"ตั๋วหนัง '{movie}' หมดแล้ว (คงเหลือ {remaining})")

seats = {"Avengers": 0, "Frozen": 5}

def book_ticket(movie):
    remaining = seats.get(movie, 0)
    if remaining <= 0:
        raise TicketSoldOutError(movie, remaining)
    seats[movie] -= 1
    return f"จองตั๋ว '{movie}' สำเร็จ คงเหลือ {seats[movie]}"

try:
    print(book_ticket("Avengers"))
except TicketSoldOutError as e:
    print("จองไม่สำเร็จ:", e)

print(book_ticket("Frozen"))
```

### หัวข้อที่ 11: ลำดับชั้นของข้อผิดพลาด (Exception Hierarchy)

**เฉลยข้อ 5** (อธิบายแนวคิด)

`IndexError` และ `KeyError` ต่างสืบทอดจาก `LookupError` ซึ่งเป็นคลาสกลางที่รวมข้อผิดพลาดทุกชนิดที่เกิดจาก
การค้นหาข้อมูลด้วยคีย์หรือดัชนีที่ไม่มีอยู่จริง (`IndexError` สำหรับรายการ, `KeyError` สำหรับ dictionary) ตามหลัก
ของ `except` ที่จับข้อผิดพลาด "ชนิดเดียวกัน หรือเป็นคลาสลูกของชนิดที่ระบุ" การเขียน `except LookupError`
เพียงบล็อกเดียวจึงดักได้ทั้งสองชนิด เพราะทั้งคู่เป็นคลาสลูกของ `LookupError` เช่นเดียวกับที่ `except ArithmeticError`
ดักได้ทั้ง `ZeroDivisionError` และ `OverflowError` ตามที่แสดงไว้ในตัวอย่างที่ 2 ของหัวข้อนี้

**เฉลยข้อ 6** (ทายผลลัพธ์)

คำอธิบาย: `get_branch_stock("เสื้อ", 1)` เข้าถึง `warehouse["เสื้อ"][1]` ได้สำเร็จ (ค่า 20) `get_branch_stock
("กางเกง", 3)` เข้าถึงดัชนี 3 ของรายการที่มีเพียง 1 สมาชิก เกิด `IndexError` และ `get_branch_stock("หมวก", 0)`
ไม่มีคีย์ `"หมวก"` ใน `warehouse` เลย เกิด `KeyError` ทั้งสองกรณีถูกดักด้วย `except LookupError` เดียวกัน แต่
`type(e).__name__` ยังคงแสดงชนิดที่แท้จริงของแต่ละกรณีไว้ให้เห็นความแตกต่าง

```python
%reset -f
warehouse = {"เสื้อ": [10, 20], "กางเกง": [5]}

def get_branch_stock(item, branch_index):
    try:
        return warehouse[item][branch_index]
    except LookupError as e:
        return f"ไม่พบข้อมูล ({type(e).__name__}): {e}"

print(get_branch_stock("เสื้อ", 1))
print(get_branch_stock("กางเกง", 3))
print(get_branch_stock("หมวก", 0))
```

**เฉลยข้อ 7** (หาข้อผิดพลาด)

ปัญหาอยู่ที่ลำดับของบล็อก `except` บรรทัดที่ 4 ดักด้วย `except Exception` ก่อนบรรทัดที่ 6 ที่ดักด้วย
`except ValueError` เนื่องจาก `ValueError` เป็นคลาสลูกของ `Exception` ข้อผิดพลาดชนิด `ValueError` ใด ๆ
ที่เกิดขึ้นในบล็อก `try` จึงตรงกับเงื่อนไขของบรรทัดที่ 4 เสมอ (เพราะเป็นบล็อก `except` แรกที่ไล่ตรวจพบว่าตรงชนิด)
การควบคุมจึงกระโดดเข้าบรรทัด 4–5 ทุกครั้ง และไม่มีทางไปถึงบรรทัด 6–7 ได้เลยไม่ว่าจะป้อนค่าใดก็ตาม บล็อก
`except ValueError` ที่บรรทัด 6 จึงเป็นส่วนที่ไม่มีโอกาสถูกเรียกใช้งาน (unreachable) ผลลัพธ์ที่ยืนยันจากการรันจริง
ยืนยันว่าฟังก์ชันคืนค่าจากบล็อก `except Exception` เสมอ ดังนี้

```python
%reset -f
def convert_weight(text):
    try:
        return float(text)
    except Exception as e:
        return f"ข้อผิดพลาดทั่วไป: {e}"
    except ValueError:
        return "ข้อความไม่ใช่ตัวเลข"

print(convert_weight("หนักมาก"))
```

ไม่ว่าจะรันด้วยตัวเลือกใดก็ตาม `assert` จึงเหมาะกับการตรวจสอบสมมติฐานภายในระหว่างพัฒนาและทดสอบมากกว่า

**เฉลยข้อ 9** (ทายผลลัพธ์)

คำอธิบาย: การเรียกครั้งแรก (`age = 25`) เงื่อนไขของ `assert` เป็นจริง จึงผ่านและคืนค่าปกติ การเรียกครั้งที่สอง
(`age = 15`) เงื่อนไขเป็นเท็จ จึงโยน `AssertionError` พร้อมข้อความที่ระบุไว้ทันที และไม่มีบล็อกดักข้อผิดพลาดใด
ครอบการเรียกนี้ไว้ โปรแกรมจึงหยุดทำงาน

```python
%reset -f
def register_member(age):
    assert age >= 18, "ต้องมีอายุอย่างน้อย 18 ปีจึงจะสมัครสมาชิกได้"
    return "สมัครสมาชิกสำเร็จ"

print(register_member(25))
print(register_member(15))
```

### หัวข้อที่ 13: การซ้อนบล็อก try/except/finally (Nested try/except/finally)

**เฉลยข้อ 11** (อธิบายแนวคิด)

ลำดับการทำงานคือ เมื่อข้อผิดพลาดเกิดขึ้นในบล็อก try ชั้นใน และไม่มีบล็อก `except` ของชั้นในที่ตรงชนิด ข้อผิดพลาด
จะเริ่มไหลออกจาก try ชั้นใน แต่ก่อนที่จะไหลออกไปถึงชั้นนอกได้จริง ตัวแปลภาษาต้องรัน `finally` ของชั้นในให้ครบ
ก่อนเสมอ (ไม่ว่า `finally` นั้นจะเกี่ยวข้องกับข้อผิดพลาดที่กำลังไหลออกหรือไม่ก็ตาม) หลังจาก `finally` ของชั้นในทำงาน
เสร็จแล้ว ข้อผิดพลาดจึงไหลต่อไปถึงบล็อก `except` ของชั้นนอกซึ่งดักได้พอดี การควบคุมจึงเข้าสู่บล็อก `except`
ของชั้นนอก และเมื่อบล็อกนั้นทำงานเสร็จ โปรแกรมจึงดำเนินต่อไปยังคำสั่งถัดจากทั้งโครงสร้างได้ตามปกติ กล่าวโดยสรุป
ลำดับคือ: จุดเกิดข้อผิดพลาด → `finally` ของชั้นใน → `except` ของชั้นนอก → คำสั่งถัดไป

**เฉลยข้อ 12** (หาข้อผิดพลาด)

สาเหตุอยู่ที่บรรทัดที่ 11 ซึ่งเขียน `except KeyError` ซ้ำอีกครั้งที่ชั้นนอก ทั้งที่ข้อผิดพลาดซึ่งเกิดจากการจองซ้ำ
(บรรทัด 8) คือ `ValueError` ไม่ใช่ `KeyError` เมื่อเรียก `check_in("101")` เงื่อนไขบรรทัด 7 เป็นจริง จึงโยน
`ValueError` ที่บรรทัด 8 ข้อผิดพลาดนี้ไม่ตรงกับ `except KeyError` ของชั้นใน (บรรทัด 9) จึงไหลออกไปยังชั้นนอก
แต่ชั้นนอกก็ดักด้วย `except KeyError` เช่นกัน (บรรทัด 11) ซึ่งไม่ตรงชนิดอีกเช่นกัน ข้อผิดพลาดจึงไม่ถูกดักที่ชั้นใด
เลยและไหลออกจากฟังก์ชันไปจนทำให้โปรแกรมหยุดทำงาน ในขณะที่ `check_in("999")` ซึ่งไม่มีหมายเลขห้องนี้อยู่จริง
ทำให้เกิด `KeyError` ที่บรรทัด 6 ซึ่งตรงกับ `except KeyError` ของชั้นในพอดี จึงดักได้และคืนค่าตามปกติ ผลลัพธ์ที่
ยืนยันจากการรันจริงคือ

```python
%reset -f
reservations = {"101": "จองแล้ว", "102": "ว่าง"}

def check_in(room):
    try:
        try:
            status = reservations[room]
            if status == "จองแล้ว":
                raise ValueError(f"ห้อง {room} มีการจองซ้ำ")
        except KeyError:
            return "ไม่พบหมายเลขห้องนี้"
    except KeyError:
        return "เกิดปัญหาการจองซ้ำ"
    return f"เช็คอินห้อง {room} สำเร็จ"

print(check_in("999"))
print(check_in("101"))
```

**เฉลยข้อ 13** (เขียนโค้ด)



```python
%reset -f
accounts = {"1234": {"pin": "0000", "balance": 500}}

def withdraw(card, pin, amount):
    try:
        try:
            account = accounts[card]
        finally:
            print(f"ตรวจสอบบัตร {card} เสร็จสิ้น")
        if pin != account["pin"]:
            raise ValueError("รหัส PIN ไม่ถูกต้อง")
        if amount > account["balance"]:
            raise ValueError("ยอดเงินคงเหลือไม่พอ")
        account["balance"] -= amount
        return f"ถอนเงินสำเร็จ คงเหลือ {account['balance']}"
    except KeyError:
        return "ไม่พบบัญชีที่ผูกกับบัตรนี้"
    except ValueError as e:
        return str(e)
    finally:
        print(f"ปิดรายการของบัตร {card}")

print(withdraw("1234", "0000", 100))
print(withdraw("9999", "0000", 100))
print(withdraw("1234", "9999", 100))
```

### หัวข้อที่ 14: การดักข้อผิดพลาดแบบไม่ระบุชนิด (Bare `except`)

**เฉลยข้อ 14** (ทายผลลัพธ์)

คำอธิบาย: `check_status("TH001")` พบคีย์ตรงกันจึงคืนค่าปกติ `check_status("TH999")` ไม่พบคีย์นี้ใน
`tracking` เกิด `KeyError` ที่ถูกดักด้วย `except:` ส่วน `check_status(12345)` ใช้ตัวเลขเป็นคีย์ค้นหาซึ่งก็ไม่พบ
เช่นกัน เกิด `KeyError` อีกเช่นเดียวกัน ทั้งสองกรณีหลังจึงแสดงข้อความเดียวกันทุกประการ ทั้งที่ชนิดของคีย์ที่ป้อน
เข้ามาต่างกัน (สายอักขระกับจำนวนเต็ม) เพราะ `except:` ไม่แยกแยะชนิดของข้อผิดพลาดเลย

```python
%reset -f
tracking = {"TH001": "จัดส่งแล้ว", "TH002": "กำลังจัดส่ง"}

def check_status(code):
    try:
        return tracking[code]
    except:
        return "ไม่พบข้อมูลพัสดุ"

print(check_status("TH001"))
print(check_status("TH999"))
print(check_status(12345))
```

**เฉลยข้อ 15** (หาข้อผิดพลาด)

สาเหตุที่แท้จริงคือบรรทัดที่ 3 พิมพ์ชื่อตัวแปรผิดเป็น `montly_rate` แทนที่จะเป็น `monthly_rate` ทำให้บรรทัดที่ 4
อ้างถึงตัวแปรชื่อ `monthly_rate` ซึ่งไม่เคยถูกกำหนดค่าไว้เลย เกิด `NameError` ขึ้นทันทีในทุกการเรียกใช้ ไม่ว่า
`principal` ที่ส่งเข้ามาจะเป็นชนิดใดก็ตาม (ทั้งจำนวนเต็มในบรรทัดที่ 9 และสายอักขระในบรรทัดที่ 10) เพราะ
`NameError` เกิดขึ้นก่อนที่โค้ดจะไปถึงจุดที่ต้องพิจารณาชนิดของ `principal` เลยด้วยซ้ำ ข้อความ "คำนวณดอกเบี้ย
ไม่สำเร็จ" ที่ปรากฏจาก `except:` (บรรทัด 6–7) จึงทำให้เข้าใจผิดได้ว่าปัญหาเกิดจากข้อมูลนำเข้า ทั้งที่สาเหตุจริง
คือข้อบกพร่องในการพิมพ์ชื่อตัวแปรของโค้ดเอง ซึ่งจะเกิดซ้ำทุกครั้งไม่ว่าจะป้อนค่าใดก็ตาม ผลลัพธ์ที่ยืนยันจากการ
รันจริงยืนยันว่าทั้งสองบรรทัดแสดงข้อความเดียวกันทุกประการ

```python
%reset -f
def calculate_interest(principal, months):
    try:
        montly_rate = 0.01
        interest = principal * monthly_rate * months
        return interest
    except:
        return "คำนวณดอกเบี้ยไม่สำเร็จ"

print(calculate_interest(1000, 6))
print(calculate_interest("1000", 6))
```

**เฉลยข้อ 16** (เขียนโค้ด)

คำอธิบาย: การเรียกครั้งแรก `100 * 33.5` คำนวณได้ตามปกติ การเรียกครั้งที่สองนำสายอักขระ `"100"` มาคูณกับ
`float` โดยตรง ซึ่งไพทอนไม่รองรับ เกิด `TypeError` ที่ถูกดักจับด้วยบล็อกเฉพาะของชนิดนี้ แสดงข้อความที่สื่อ
ความหมายตรงประเด็นกว่าการดักข้อผิดพลาดทุกชนิดด้วยข้อความเดียว

```python
%reset -f
def convert_currency(amount, rate):
    try:
        return amount * rate
    except TypeError:
        return "ชนิดข้อมูลไม่ถูกต้องสำหรับการแปลงสกุลเงิน"
    except ValueError:
        return "ค่าที่ป้อนไม่ถูกต้อง"

print(convert_currency(100, 33.5))
print(convert_currency("100", 33.5))
```

### หัวข้อที่ 15: การโยนข้อผิดพลาดซ้ำและการเชื่อมโยงข้อผิดพลาด (Re-raise / Exception Chaining)

**เฉลยข้อ 17** (อธิบายแนวคิด)

รูปแบบที่ (1) `raise ข้อผิดพลาดใหม่(...)` โดยไม่มี `from` ต่อท้าย ทำให้ `__cause__` ของข้อผิดพลาดใหม่มีค่าเป็น
`None` โดยปริยาย แต่ข้อผิดพลาดเดิมยังคงถูกบันทึกไว้ใน `__context__` โดยอัตโนมัติ (ปรากฏใน traceback ในรูป
"During handling of the above exception, another exception occurred") รูปแบบที่ (2) `raise ข้อผิดพลาดใหม่
(...) from e` กำหนด `__cause__` ให้เป็นอ็อบเจกต์ `e` (ข้อผิดพลาดเดิม) อย่างชัดเจน ทำให้เข้าถึงสาเหตุดั้งเดิมผ่าน
`e.__cause__` ได้โดยตรง และ traceback แสดงทั้งสองข้อผิดพลาดต่อเนื่องกันในรูป "The above exception was the
direct cause of the following exception" ส่วนรูปแบบที่ (3) `raise ข้อผิดพลาดใหม่(...) from None` กำหนด
`__cause__` เป็น `None` อย่างจงใจ และสั่งให้ตัวแปลภาษาระงับการแสดง `__context__` ใน traceback ด้วย ทำให้
เห็นเฉพาะข้อผิดพลาดใหม่เพียงชนิดเดียว แม้ข้อผิดพลาดเดิมจะยังถูกบันทึกไว้ใน `__context__` ของอ็อบเจกต์อยู่ก็ตาม

**เฉลยข้อ 18** (ทายผลลัพธ์)

คำอธิบาย: การเรียก `verify_otp("1234", "5678")` ที่บรรทัด 11 ทำให้เงื่อนไขบรรทัด 3 เป็นจริง โยน `ValueError`
ที่บรรทัด 4 ซึ่งดักได้ที่บรรทัด 5 พิมพ์ข้อความบันทึก log แล้วโยนซ้ำด้วย `raise` เพียงคำเดียวที่บรรทัด 7 ข้อผิดพลาด
ตัวเดิมจึงไหลออกไปถึงบล็อก `except ValueError` ของโปรแกรมหลักที่บรรทัด 12–13 การเรียก `verify_otp("1111",
"1111")` ที่บรรทัด 15 ไม่มีข้อผิดพลาดเกิดขึ้นเลย จึงคืนค่า "ยืนยันสำเร็จ" ตามปกติ

```python
%reset -f
def verify_otp(code, expected):
    try:
        if code != expected:
            raise ValueError("รหัส OTP ไม่ถูกต้อง")
    except ValueError:
        print("บันทึก log ความพยายามยืนยัน OTP ที่ล้มเหลว")
        raise
    return "ยืนยันสำเร็จ"

try:
    verify_otp("1234", "5678")
except ValueError as e:
    print("ระบบหลักจับข้อผิดพลาดได้:", e)

print(verify_otp("1111", "1111"))
```

**เฉลยข้อ 19** (หาข้อผิดพลาด)

โค้ดที่ทำให้ `e.__cause__` เป็น `None` คือบรรทัดที่ 8 ซึ่งเขียน `raise LoanValidationError("ข้อมูลรายได้ไม่ถูกต้อง")`
โดยไม่มี `from e` ต่อท้าย แม้จะดัก `ValueError` ไว้เป็นตัวแปร `e` ที่บรรทัด 7 แล้วก็ตาม แต่ไม่ได้ส่งต่อไปยังคำสั่ง
`raise` เลย รูปแบบที่จะเชื่อมโยงข้อผิดพลาดเดิมไว้ได้จริงคือ `raise LoanValidationError("ข้อมูลรายได้ไม่ถูกต้อง")
from e` ซึ่งจะกำหนด `__cause__` ของ `LoanValidationError` ให้เป็นอ็อบเจกต์ `ValueError` เดิมโดยตรง ผลลัพธ์
ที่ยืนยันจากการรันจริงของโค้ดในรูปแบบที่ให้มา (ไม่มี `from e`) คือ

```python
%reset -f
class LoanValidationError(Exception):
    pass

def validate_income(income_text):
    try:
        income = float(income_text)
    except ValueError as e:
        raise LoanValidationError("ข้อมูลรายได้ไม่ถูกต้อง")
    return income

try:
    validate_income("สามหมื่น")
except LoanValidationError as e:
    print("ตรวจสอบไม่ผ่าน:", e)
    print("สาเหตุเดิม:", e.__cause__)
```

**เฉลยข้อ 20** (เขียนโค้ด)

---

```python
%reset -f
class MissingFieldError(Exception):
    pass

def process_application(form):
    try:
        return f"ผู้สมัคร {form['name']} ตำแหน่ง {form['position']}"
    except KeyError as e:
        raise MissingFieldError(f"ข้อมูลที่จำเป็นขาดหายไป: {e}") from e

try:
    process_application({"name": "สมชาย"})
except MissingFieldError as e:
    print("ยื่นใบสมัครไม่สำเร็จ:", e)
    print("สาเหตุเดิม:", repr(e.__cause__))
```
