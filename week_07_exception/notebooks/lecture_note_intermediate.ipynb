## 6: ส่วน else (เมื่อไม่มีข้อผิดพลาด)

### แนวคิด

โครงสร้างการจัดการข้อผิดพลาดในหลายภาษาอนุญาตให้แนบส่วนเสริมเข้ากับบล็อกทดลองรัน (try) เพื่อระบุชุดคำสั่งที่ควรทำงาน
**เฉพาะเมื่อบล็อกทดลองรันทั้งหมดจบลงโดยไม่มีข้อผิดพลาดขณะรันเกิดขึ้นเลย** ในภาษาที่ไม่มีส่วนนี้โดยตรง ผู้เขียนโปรแกรม
มักต้องใช้ตัวแปรสถานะ (flag) หรือวางคำสั่งไว้ต่อท้ายโครงสร้าง try/except ทั้งหมดแทน ซึ่งอ่านยากกว่าและแยกแยะเจตนา
ของโค้ดได้ไม่ชัดเจนเท่า ภาษา Python มีส่วนที่เรียกว่า `else` แนบท้าย try/except โดยเฉพาะสำหรับจุดประสงค์นี้

ส่วน `else` ทำงานหลังจากบล็อกทดลองรันจบสมบูรณ์โดยไม่มีข้อผิดพลาด และทำงานก่อนส่วน finally (ถ้ามี) ข้อสำคัญที่ต้อง
เข้าใจให้ถูกต้องคือ **บล็อกดักข้อผิดพลาด (except) ที่เขียนคู่กันนั้นคุ้มครองเฉพาะสิ่งที่เกิดขึ้นภายในบล็อกทดลองรันเท่านั้น
ไม่ได้ขยายไปคุ้มครองสิ่งที่เกิดขึ้นภายในส่วน else ด้วย** แม้ข้อผิดพลาดที่เกิดในส่วน else จะเป็นชนิดเดียวกับที่ except
ระบุไว้ก็ตาม การแยกโค้ดเสี่ยงต่อข้อผิดพลาด (ไว้ในบล็อกทดลองรัน) ออกจากโค้ดที่ควรทำงานเฉพาะเมื่อสำเร็จ (ไว้ในส่วน else)
ทำให้อ่านโค้ดแล้วแยกแยะได้ทันทีว่าส่วนใดของโปรแกรมมีความเสี่ยงต่อข้อผิดพลาดจริง ๆ

### ตัวอย่าง Python

#### ตัวอย่างที่ 1 (หลัก): ตรวจสอบคะแนนสอบที่ป้อนเป็นข้อความ

````python
def read_score(text):
    try:
        score = int(text)
    except ValueError:
        print("แปลงคะแนนไม่สำเร็จ:", text)
    else:
        print("แปลงคะแนนสำเร็จ:", score)

read_score("85")
read_score("แปดสิบ")
````

**Output:**
```
แปลงคะแนนสำเร็จ: 85
แปลงคะแนนไม่สำเร็จ: แปดสิบ
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | คำอธิบายเพิ่มเติม |
|---|---|---|---|
| 9 | เรียก `read_score("85")` | text="85" | เข้าสู่ฟังก์ชัน เริ่มบล็อกทดลองรัน |
| 3 | `score = int(text)` | score=85 | แปลงสำเร็จ ไม่มีข้อผิดพลาดเกิดขึ้นในบล็อกทดลองรัน |
| 7 | เข้าสู่ส่วน `else` แล้ว `print(...)` | แสดง `แปลงคะแนนสำเร็จ: 85` | ส่วน else ทำงานเพราะบล็อกทดลองรันจบโดยไม่มีข้อผิดพลาด |
| 10 | เรียก `read_score("แปดสิบ")` | text="แปดสิบ" | |
| 3 | `score = int(text)` | เกิด ValueError | `int()` แปลงข้อความที่ไม่ใช่ตัวเลขไม่ได้ |
| 5 | เข้าสู่ `except ValueError` แล้ว `print(...)` | แสดง `แปลงคะแนนไม่สำเร็จ: แปดสิบ` | เข้าสู่บล็อกดักข้อผิดพลาดทันทีที่เกิดข้อผิดพลาด ส่วน else ไม่ทำงานในกรณีนี้ |

**คำอธิบายภาพรวม:** บรรทัด 1–7 นิยามฟังก์ชัน `read_score` ที่พยายามแปลงข้อความเป็นจำนวนเต็มในบรรทัด 3 หากแปลง
สำเร็จ ควบคุมจะข้ามบล็อก except ไปเข้าส่วน `else` ที่บรรทัด 6–7 ทันที เพื่อพิมพ์ผลลัพธ์ที่แปลงได้ หากแปลงไม่สำเร็จ
(บรรทัด 10) ควบคุมจะกระโดดไปยังบล็อก except ที่บรรทัด 4–5 แทน และส่วน else จะไม่ถูกประมวลผลเลย โครงสร้างนี้แยก
"โค้ดที่เสี่ยงต่อข้อผิดพลาด" (การแปลงชนิดข้อมูลที่บรรทัด 3) ออกจาก "โค้ดที่ควรทำงานเฉพาะเมื่อสำเร็จ" (การแสดงผลที่
บรรทัด 7) อย่างชัดเจน

**จุดสังเกต:**
- ส่วน else ทำงานเฉพาะเมื่อบล็อกทดลองรันทั้งหมดจบลงโดยไม่มีข้อผิดพลาดเกิดขึ้นเลยเท่านั้น
- ตัวแปร `score` ที่สร้างขึ้นในบล็อกทดลองรัน (บรรทัด 3) ยังคงใช้งานได้ในส่วน else (บรรทัด 7) ตามปกติ เพราะทั้งสอง
  ส่วนอยู่ใน scope เดียวกันของฟังก์ชัน ไม่ใช่ scope ที่แยกจากกัน
- หากย้ายคำสั่ง `print(...)` ที่บรรทัด 7 เข้าไปต่อท้ายบรรทัด 3 ภายในบล็อกทดลองรันแทน ผลลัพธ์ที่แสดงจะเหมือนเดิม
  ทุกประการในกรณีนี้ ความแตกต่างระหว่างสองแนวทางจะปรากฏชัดก็ต่อเมื่อโค้ดที่ย้ายเข้าไปนั้นมีโอกาสก่อข้อผิดพลาดขึ้นเอง
  ได้ ดังแสดงในตัวอย่างที่ 3

#### ตัวอย่างที่ 2 (โดเมนต่างออกไป): ค้นหาเงินเดือนพนักงานจากรหัส

````python
salary_table = {"E001": 25000, "E002": 30000}

def show_salary(emp_id):
    try:
        salary = salary_table[emp_id]
    except KeyError:
        print("ไม่พบรหัสพนักงาน:", emp_id)
    else:
        print("เงินเดือนของ", emp_id, "คือ", salary)

show_salary("E001")
show_salary("E999")
````

**Output:**
```
เงินเดือนของ E001 คือ 25000
ไม่พบรหัสพนักงาน: E999
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | คำอธิบายเพิ่มเติม |
|---|---|---|---|
| 11 | เรียก `show_salary("E001")` | emp_id="E001" | เข้าสู่ฟังก์ชัน เริ่มบล็อกทดลองรัน |
| 5 | `salary = salary_table[emp_id]` | salary=25000 | คีย์ "E001" มีอยู่จริงในดิกชันนารี ไม่มีข้อผิดพลาดเกิดขึ้น |
| 9 | เข้าสู่ส่วน `else` แล้ว `print(...)` | แสดง `เงินเดือนของ E001 คือ 25000` | |
| 12 | เรียก `show_salary("E999")` | emp_id="E999" | |
| 5 | `salary = salary_table[emp_id]` | เกิด KeyError | ไม่มีคีย์ "E999" อยู่ในดิกชันนารี |
| 7 | เข้าสู่ `except KeyError` แล้ว `print(...)` | แสดง `ไม่พบรหัสพนักงาน: E999` | ส่วน else ไม่ทำงานในกรณีนี้ |

**คำอธิบายภาพรวม:** บรรทัด 3–9 นิยามฟังก์ชัน `show_salary` ที่พยายามเข้าถึงดิกชันนารี `salary_table` ด้วยคีย์ที่รับเข้า
มา (บรรทัด 5) การเข้าถึงดิกชันนารีด้วยคีย์ที่ไม่มีอยู่จริงก่อให้เกิด `KeyError` (เทียบเคียงได้กับ `IndexError` ของ list
ที่กล่าวถึงในหัวข้อที่ 3 ของ `lecture_note_basic.md`) หากคีย์มีอยู่จริง ควบคุมจะข้ามไปเข้าส่วน else เพื่อแสดงเงินเดือน
หากไม่มี ควบคุมจะเข้าสู่บล็อก except เพื่อแจ้งว่าไม่พบรหัสพนักงานแทน

**จุดสังเกต:**
- `KeyError` เป็นชนิดข้อผิดพลาดขณะรันที่เกิดจากการเข้าถึงดิกชันนารีด้วยคีย์ที่ไม่มีอยู่จริง มีกลไกคล้ายกับ `IndexError`
  ของ list แต่เป็นโครงสร้างข้อมูลคนละชนิดกัน
- การเข้าถึงดิกชันนารีที่เสี่ยงต่อข้อผิดพลาด (บรรทัด 5) อยู่ในบล็อกทดลองรัน ส่วนการแสดงผลลัพธ์ที่ปลอดภัยแล้ว
  (บรรทัด 9) แยกไปไว้ในส่วน else — รูปแบบเดียวกับตัวอย่างที่ 1 แต่ต่างโดเมนข้อมูลกันอย่างชัดเจน (ตัวเลขแปลงข้อความ
  เทียบกับการค้นหาในดิกชันนารี)
- ตัวแปร `salary_table` ประกาศไว้นอกฟังก์ชันที่บรรทัด 1 ฟังก์ชัน `show_salary` จึงเข้าถึงได้โดยไม่ต้องรับเป็น
  พารามิเตอร์ สอดคล้องกับกฎเรื่อง scope ที่กล่าวถึงในหัวข้อที่ 5 ของ `lecture_note_basic.md`

#### ตัวอย่างที่ 3 (edge case): ข้อผิดพลาดที่เกิดขึ้นภายในส่วน else เอง

````python
def convert_temperature(text):
    try:
        celsius = float(text)
    except ValueError:
        print("แปลงอุณหภูมิไม่สำเร็จ:", text)
    else:
        if celsius < -273.15:
            raise ValueError("อุณหภูมิต่ำกว่าศูนย์องศาสัมบูรณ์")
        fahrenheit = celsius * 9 / 5 + 32
        print("องศาฟาเรนไฮต์:", fahrenheit)

convert_temperature("100")
convert_temperature("-300")
````

**Output:**
```
องศาฟาเรนไฮต์: 212.0
Traceback (most recent call last):
  ...
ValueError: อุณหภูมิต่ำกว่าศูนย์องศาสัมบูรณ์
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | คำอธิบายเพิ่มเติม |
|---|---|---|---|
| 12 | เรียก `convert_temperature("100")` | text="100" | |
| 3 | `celsius = float(text)` | celsius=100.0 | แปลงสำเร็จ ไม่มีข้อผิดพลาดในบล็อกทดลองรัน |
| 7 | เข้าสู่ส่วน `else` แล้ว `if celsius < -273.15` | False (100.0 ไม่ต่ำกว่า -273.15) | เงื่อนไขเป็นเท็จ |
| 9 | `fahrenheit = celsius * 9 / 5 + 32` | fahrenheit=212.0 | |
| 10 | `print(...)` | แสดง `องศาฟาเรนไฮต์: 212.0` | |
| 13 | เรียก `convert_temperature("-300")` | text="-300" | |
| 3 | `celsius = float(text)` | celsius=-300.0 | "-300" เป็นรูปแบบตัวเลขที่ถูกต้อง จึงแปลงสำเร็จ ไม่มี ValueError จากบรรทัดนี้ |
| 7 | เข้าสู่ส่วน `else` แล้ว `if celsius < -273.15` | True | เงื่อนไขเป็นจริง |
| 8 | `raise ValueError(...)` | เกิด ValueError ใหม่ที่ถูกโยนขึ้นภายในส่วน else | บล็อก `except ValueError` ที่บรรทัด 4 ดักจับข้อผิดพลาดนี้ไม่ได้ เพราะขอบเขตของ except ครอบคลุมเฉพาะบล็อกทดลองรัน (บรรทัด 2–3) เท่านั้น ไม่ครอบคลุมส่วน else — โปรแกรมจึงหยุดทำงานด้วย traceback |

**คำอธิบายภาพรวม:** การเรียกที่บรรทัด 12 ทำงานตามปกติ คือแปลงค่าในบล็อกทดลองรันสำเร็จ แล้วเข้าสู่ส่วน else เพื่อ
คำนวณและแสดงผลอุณหภูมิ ส่วนการเรียกที่บรรทัด 13 นั้นการแปลงค่าที่บรรทัด 3 สำเร็จเช่นกัน (เพราะ "-300" เป็นข้อความ
ตัวเลขที่ถูกต้องตามรูปแบบ) จึงเข้าสู่ส่วน else ที่บรรทัด 6 ด้วย แต่เงื่อนไขตรวจสอบที่บรรทัด 7 พบว่าอุณหภูมินี้ต่ำกว่าศูนย์
องศาสัมบูรณ์ในทางฟิสิกส์ จึงมีการโยนข้อผิดพลาดใหม่ขึ้นเองที่บรรทัด 8 ด้วยคำสั่ง `raise` (รายละเอียดของ `raise` จะกล่าว
ถึงในหัวข้อที่ 9) ข้อผิดพลาดนี้แม้จะเป็นชนิด `ValueError` ตรงกับที่ระบุไว้ใน except ที่บรรทัด 4 ทุกประการ แต่ except
นั้นดักจับไม่ได้ เพราะขอบเขตการดักจับของ except ผูกอยู่กับบล็อกทดลองรันเท่านั้น ไม่ได้ขยายไปถึงส่วน else โปรแกรมจึง
หยุดทำงานด้วย traceback หลังจากบรรทัด 13

**จุดสังเกต:**
- แม้ข้อผิดพลาดที่เกิดขึ้นที่บรรทัด 8 จะเป็นชนิด `ValueError` ตรงกับชนิดที่ except ที่บรรทัด 4 ดักไว้ทุกประการ แต่
  except นั้นก็ยังดักจับไม่ได้ เพราะขอบเขตการดักจับของ except ครอบคลุมเฉพาะสิ่งที่เกิดขึ้นในบล็อกทดลองรังเท่านั้น
  ไม่ครอบคลุมส่วน else แม้จะเป็นข้อผิดพลาดชนิดเดียวกันก็ตาม
- นี่คือความแตกต่างสำคัญระหว่างการวางโค้ดที่มีโอกาสก่อข้อผิดพลาดไว้ในบล็อกทดลองรันกับการวางไว้ในส่วน else: โค้ดใน
  บล็อกทดลองรันได้รับการคุ้มครองจาก except คู่กัน แต่โค้ดในส่วน else ไม่ได้รับการคุ้มครองจาก except บล็อกเดียวกันเลย
- หากต้องการดักจับข้อผิดพลาดที่อาจเกิดขึ้นภายในส่วน else ด้วย จำเป็นต้องเขียนบล็อกทดลองรัน/ดักข้อผิดพลาดซ้อนอีกชุด
  หนึ่งครอบคลุมเฉพาะส่วนนั้นโดยเฉพาะ (nested try/except) ซึ่งเป็นแนวคิดที่อยู่นอกขอบเขตของเอกสารชุดนี้

**แหล่งอ้างอิง:** W3Schools — "Else", TutorialsPoint — "try-except-else Statement"

## 7: ส่วน finally (รันเสมอ)

### แนวคิด

โครงสร้างการจัดการข้อผิดพลาดในหลายภาษามีกลไกเสริมสำหรับการดำเนินการที่ต้อง "เกิดขึ้นเสมอ" ไม่ว่าการทำงานก่อนหน้า
จะสำเร็จ เกิดข้อผิดพลาดที่ถูกดักจับได้ หรือแม้แต่เกิดข้อผิดพลาดที่ไม่มีการดักจับเลยก็ตาม เช่น การปิดการเชื่อมต่อกับ
อุปกรณ์ภายนอก การคืนทรัพยากรที่ยืมมาใช้ชั่วคราว หรือการบันทึกว่าขั้นตอนตรวจสอบได้ดำเนินการเสร็จสิ้นแล้ว การดำเนินการ
ลักษณะนี้เรียกว่าการล้างข้อมูล (cleanup) ในภาษา Python ส่วนที่ทำหน้าที่นี้เรียกว่า `finally` ซึ่งแนบท้ายโครงสร้าง
try เสมอ

ส่วน finally ทำงานหลังจากบล็อกทดลองรัน (และบล็อกดักข้อผิดพลาดหรือส่วน else ถ้ามีการเข้าสู่ส่วนใดส่วนหนึ่ง) ประมวลผล
เสร็จสิ้นแล้วในทุกกรณี รวมถึงกรณีที่มีคำสั่ง `return` อยู่ภายในบล็อกทดลองรันหรือบล็อกดักข้อผิดพลาดด้วย — ค่าที่จะถูก
ส่งกลับจาก `return` นั้นจะถูก "รอไว้ก่อน" จนกว่าส่วน finally จะประมวลผลเสร็จสิ้นแล้วจึงส่งกลับออกไปจริง ข้อสำคัญ
อีกประการคือ finally **ไม่ได้ทำหน้าที่ดักจับหรือระงับข้อผิดพลาด** ด้วยตัวมันเอง หากบล็อกทดลองรันเกิดข้อผิดพลาดที่ไม่มี
except ใดดักจับได้ ข้อผิดพลาดนั้นจะยังคงแพร่กระจายต่อไปตามปกติหลังจากส่วน finally ทำงานเสร็จแล้ว

### ตัวอย่าง Python

#### ตัวอย่างที่ 1 (หลัก): คำนวณค่าเฉลี่ยคะแนนสองวิชา

````python
def average_of_two(a, b):
    try:
        result = (a + b) / 2
        print("ค่าเฉลี่ย:", result)
    except TypeError:
        print("ชนิดข้อมูลไม่ถูกต้อง")
    finally:
        print("ประมวลผลคะแนนคู่นี้เสร็จสิ้น")

average_of_two(80, 90)
average_of_two(80, "เก้าสิบ")
````

**Output:**
```
ค่าเฉลี่ย: 85.0
ประมวลผลคะแนนคู่นี้เสร็จสิ้น
ชนิดข้อมูลไม่ถูกต้อง
ประมวลผลคะแนนคู่นี้เสร็จสิ้น
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | คำอธิบายเพิ่มเติม |
|---|---|---|---|
| 10 | เรียก `average_of_two(80, 90)` | a=80, b=90 | |
| 3 | `result = (a + b) / 2` | result=85.0 | ไม่มีข้อผิดพลาดเกิดขึ้น |
| 4 | `print(...)` | แสดง `ค่าเฉลี่ย: 85.0` | |
| 8 | เข้าสู่ `finally` แล้ว `print(...)` | แสดง `ประมวลผลคะแนนคู่นี้เสร็จสิ้น` | finally ทำงานหลังบล็อกทดลองรันจบสำเร็จ (ไม่ผ่าน except เพราะไม่มีข้อผิดพลาด) |
| 11 | เรียก `average_of_two(80, "เก้าสิบ")` | a=80, b="เก้าสิบ" | |
| 3 | `result = (a + b) / 2` | เกิด TypeError | บวก int กับ str เข้าด้วยกันไม่ได้ |
| 6 | เข้าสู่ `except TypeError` แล้ว `print(...)` | แสดง `ชนิดข้อมูลไม่ถูกต้อง` | |
| 8 | เข้าสู่ `finally` แล้ว `print(...)` | แสดง `ประมวลผลคะแนนคู่นี้เสร็จสิ้น` | finally ทำงานอีกครั้ง แม้ครั้งนี้จะเกิดข้อผิดพลาดและถูกจัดการโดย except ไปแล้วก็ตาม |

**คำอธิบายภาพรวม:** บรรทัด 1–8 นิยามฟังก์ชัน `average_of_two` ที่คำนวณค่าเฉลี่ยของสองค่าในบล็อกทดลองรัน (บรรทัด
3–4) หากชนิดข้อมูลไม่ถูกต้องจะเข้าสู่บล็อกดักข้อผิดพลาดที่บรรทัด 5–6 ไม่ว่าเส้นทางใดเส้นทางหนึ่งข้างต้นจะทำงาน ส่วน
finally ที่บรรทัด 7–8 จะทำงานเสมอเป็นลำดับสุดท้าย ดังที่เห็นได้จากข้อความ "ประมวลผลคะแนนคู่นี้เสร็จสิ้น" ที่ปรากฏ
ทั้งสองครั้งของการเรียกฟังก์ชัน (บรรทัด 10 และ 11) แม้ผลลัพธ์ก่อนหน้าจะต่างกันโดยสิ้นเชิงก็ตาม

**จุดสังเกต:**
- ข้อความ "ประมวลผลคะแนนคู่นี้เสร็จสิ้น" ถูกพิมพ์ทั้งสองครั้งที่เรียกฟังก์ชัน ไม่ว่าการคำนวณจะสำเร็จ (บรรทัด 10)
  หรือเกิดข้อผิดพลาด (บรรทัด 11) ก็ตาม นี่คือคุณสมบัติเฉพาะของ finally ที่ else ในหัวข้อที่ 6 ไม่มี (else ทำงาน
  เฉพาะกรณีสำเร็จเท่านั้น)
- ลำดับการทำงานคงที่เสมอ: บล็อกทดลองรัน (หรือบล็อกทดลองรันตามด้วยบล็อกดักข้อผิดพลาด) ทำงานก่อน แล้ว finally
  ทำงานทีหลังเสมอ ไม่มีทางที่ finally จะทำงานก่อนบล็อกอื่น
- หากลบ `except TypeError` ออกจากโค้ดนี้ การเรียก `average_of_two(80, "เก้าสิบ")` จะยังคงพิมพ์ "ประมวลผล
  คะแนนคู่นี้เสร็จสิ้น" จาก finally ก่อน แล้วจึงเกิด `TypeError` ที่ไม่มีผู้ใดดักจับ โปรแกรมจะหยุดทำงานด้วย traceback
  ภายหลังจาก finally ทำงานเสร็จแล้ว (finally ไม่ได้ทำให้ข้อผิดพลาดหายไป)

#### ตัวอย่างที่ 2 (โดเมนต่างออกไป): จำลองการเปิด-ปิดการเชื่อมต่อเซนเซอร์

````python
def read_sensor(value_text):
    print("เปิดการเชื่อมต่อเซนเซอร์")
    try:
        reading = float(value_text)
        print("ค่าที่อ่านได้:", reading)
    except ValueError:
        print("ค่าจากเซนเซอร์อ่านไม่ได้:", value_text)
    finally:
        print("ปิดการเชื่อมต่อเซนเซอร์")

read_sensor("36.5")
read_sensor("ERROR")
````

**Output:**
```
เปิดการเชื่อมต่อเซนเซอร์
ค่าที่อ่านได้: 36.5
ปิดการเชื่อมต่อเซนเซอร์
เปิดการเชื่อมต่อเซนเซอร์
ค่าจากเซนเซอร์อ่านไม่ได้: ERROR
ปิดการเชื่อมต่อเซนเซอร์
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | คำอธิบายเพิ่มเติม |
|---|---|---|---|
| 11 | เรียก `read_sensor("36.5")` | value_text="36.5" | |
| 2 | `print(...)` | แสดง `เปิดการเชื่อมต่อเซนเซอร์` | ทำงานก่อนเข้าสู่บล็อกทดลองรัน |
| 4 | `reading = float(value_text)` | reading=36.5 | แปลงสำเร็จ |
| 5 | `print(...)` | แสดง `ค่าที่อ่านได้: 36.5` | |
| 9 | เข้าสู่ `finally` แล้ว `print(...)` | แสดง `ปิดการเชื่อมต่อเซนเซอร์` | |
| 12 | เรียก `read_sensor("ERROR")` | value_text="ERROR" | |
| 2 | `print(...)` | แสดง `เปิดการเชื่อมต่อเซนเซอร์` | |
| 4 | `reading = float(value_text)` | เกิด ValueError | "ERROR" ไม่ใช่รูปแบบตัวเลข |
| 7 | เข้าสู่ `except ValueError` แล้ว `print(...)` | แสดง `ค่าจากเซนเซอร์อ่านไม่ได้: ERROR` | |
| 9 | เข้าสู่ `finally` แล้ว `print(...)` | แสดง `ปิดการเชื่อมต่อเซนเซอร์` | ปิดการเชื่อมต่อแม้การอ่านค่าจะล้มเหลวก็ตาม |

**คำอธิบายภาพรวม:** บรรทัด 1–9 นิยามฟังก์ชัน `read_sensor` ที่จำลองการเปิดการเชื่อมต่อกับอุปกรณ์ก่อนเข้าสู่บล็อก
ทดลองรัน (บรรทัด 2) แล้วพยายามแปลงค่าที่อ่านได้เป็นตัวเลข ไม่ว่าการแปลงจะสำเร็จ (บรรทัด 11) หรือไม่สำเร็จ (บรรทัด
12) ส่วน finally ที่บรรทัด 8–9 รับประกันว่าการเชื่อมต่อจะถูกปิดเสมอ ตัวอย่างนี้แสดงกรณีใช้งานจริงของ finally ที่พบ
บ่อยที่สุด คือการจับคู่ "เปิด" กับ "ปิด" ทรัพยากรให้สมดุลกันเสมอไม่ว่าจะเกิดข้อผิดพลาดระหว่างทางหรือไม่

**จุดสังเกต:**
- ข้อความ "เปิดการเชื่อมต่อเซนเซอร์" กับ "ปิดการเชื่อมต่อเซนเซอร์" ปรากฏเป็นคู่กันเสมอในทุกครั้งที่เรียกฟังก์ชัน
  ไม่ว่าผลลัพธ์ตรงกลางจะเป็นอย่างไร นี่คือรูปแบบการใช้งาน finally ที่พบได้บ่อยในสถานการณ์จริง เช่น การจัดการไฟล์
  หรือการเชื่อมต่อฐานข้อมูล
- โดเมนของตัวอย่างนี้ (การจำลองอุปกรณ์เซนเซอร์) แตกต่างจากตัวอย่างที่ 1 (การคำนวณค่าเฉลี่ยตัวเลข) แต่โครงสร้าง
  โค้ดยังคงใช้รูปแบบเดียวกัน (try / except / finally) แสดงให้เห็นว่ารูปแบบนี้นำไปปรับใช้กับปัญหาที่หลากหลายได้
- แม้ฟังก์ชันนี้จะไม่ได้เชื่อมต่ออุปกรณ์จริง (เป็นเพียงข้อความจำลอง) แต่ลำดับการพิมพ์ข้อความยืนยันหลักการทำงานของ
  finally ได้ตรงกับสถานการณ์การจัดการทรัพยากรจริงทุกประการ

#### ตัวอย่างที่ 3 (edge case): finally ทำงานก่อนที่ค่า return จะถูกส่งออกจริง

````python
def is_password_valid(password):
    try:
        if len(password) < 8:
            return False
        return True
    finally:
        print("ตรวจสอบความยาวรหัสผ่านเสร็จสิ้น")

print("ผลตรวจสอบ 1:", is_password_valid("abc"))
print("ผลตรวจสอบ 2:", is_password_valid("abcdefgh"))
````

**Output:**
```
ตรวจสอบความยาวรหัสผ่านเสร็จสิ้น
ผลตรวจสอบ 1: False
ตรวจสอบความยาวรหัสผ่านเสร็จสิ้น
ผลตรวจสอบ 2: True
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | คำอธิบายเพิ่มเติม |
|---|---|---|---|
| 9 | เรียก `is_password_valid("abc")` | password="abc" | |
| 3 | `if len(password) < 8` | len=3, 3<8 → True | |
| 4 | `return False` | ค่าที่จะคืน = False (ยังไม่ส่งออกจากฟังก์ชันจริง) | Python "รอค่าที่จะ return ไว้ก่อน" แล้วยังไม่ออกจากฟังก์ชันทันที |
| 7 | เข้าสู่ `finally` แล้ว `print(...)` | แสดง `ตรวจสอบความยาวรหัสผ่านเสร็จสิ้น` | finally ทำงานก่อนที่ค่า False จะถูกส่งกลับออกไปจริง |
| 9 | ฟังก์ชันส่งค่า False กลับสู่จุดเรียกใช้ แล้ว `print(...)` | แสดง `ผลตรวจสอบ 1: False` | ค่าที่ return ส่งกลับหลัง finally ทำงานเสร็จแล้วเท่านั้น |
| 10 | เรียก `is_password_valid("abcdefgh")` | password="abcdefgh" | |
| 3 | `if len(password) < 8` | len=8, 8<8 → False | |
| 5 | `return True` | ค่าที่จะคืน = True | |
| 7 | เข้าสู่ `finally` แล้ว `print(...)` | แสดง `ตรวจสอบความยาวรหัสผ่านเสร็จสิ้น` | |
| 10 | ฟังก์ชันส่งค่า True กลับสู่จุดเรียกใช้ แล้ว `print(...)` | แสดง `ผลตรวจสอบ 2: True` | |

**คำอธิบายภาพรวม:** ฟังก์ชัน `is_password_valid` มีคำสั่ง `return` อยู่สองจุดในบล็อกทดลองรัน (บรรทัด 4 และ 5)
เมื่อบรรทัดใดบรรทัดหนึ่งประมวลผล Python จะยังไม่ส่งค่านั้นออกจากฟังก์ชันทันที แต่จะเข้าสู่ finally ที่บรรทัด 6–7 ก่อน
เสมอ หลังจาก finally ประมวลผลเสร็จสิ้นแล้ว ค่าที่ถูก "รอไว้" จากคำสั่ง return จึงถูกส่งกลับออกไปสู่จุดที่เรียกใช้ฟังก์ชัน
จริง ๆ ลำดับผลลัพธ์ที่ปรากฏ (ข้อความจาก finally มาก่อนข้อความที่แสดงค่าที่ return เสมอ) ยืนยันลำดับการทำงานนี้

**จุดสังเกต:**
- แม้คำสั่ง return จะอยู่ในบล็อกทดลองรันและดูเหมือนควรทำให้ฟังก์ชันออกจากการทำงานทันที แต่ finally ยังคงทำงาน
  แทรกก่อนที่ค่านั้นจะถูกส่งกลับออกไปสู่ผู้เรียกใช้จริงเสมอ
- ข้อความจาก finally ("ตรวจสอบความยาวรหัสผ่านเสร็จสิ้น") ปรากฏขึ้นก่อนข้อความที่แสดงผลลัพธ์ของ return ("ผลตรวจ
  สอบ 1: ..." และ "ผลตรวจสอบ 2: ...") ในทุกบรรทัดของผลลัพธ์ ยืนยันลำดับการทำงานนี้อย่างชัดเจน
- ฟังก์ชันนี้ยัง return ค่าเดียวเสมอในทุกเส้นทางการทำงาน (มีเพียง True หรือ False เท่านั้น) สอดคล้องกับรูปแบบ
  ฟังก์ชันป้องกันข้อผิดพลาดที่แสดงใน Slide 5 ของสไลด์ต้นฉบับ (`safe_divide`) แม้จะมีจุดออกจากฟังก์ชันสองจุด
  (บรรทัด 4 และ 5) ก็ตาม

## 8: จับข้อผิดพลาดเป็นตัวแปรด้วย `except ... as` และข้อความข้อผิดพลาดจริง

### แนวคิด

เมื่อเกิดข้อผิดพลาดขณะรัน ภาษาโปรแกรมจำนวนมากไม่เพียงระบุ "ชนิด" ของข้อผิดพลาดเท่านั้น แต่ยังสร้างอ็อบเจกต์ที่บรรจุ
รายละเอียดของข้อผิดพลาดนั้นไว้ด้วย ผู้เขียนโปรแกรมสามารถเข้าถึงรายละเอียดนี้เพื่อนำไปแสดงผล บันทึกเก็บไว้ตรวจสอบ
ภายหลัง หรือใช้ตัดสินใจขั้นตอนถัดไปได้ ในภาษา Python ทำได้โดยเขียน `except ExceptionType as ตัวแปร` ซึ่งจะผูก
อ็อบเจกต์ของข้อผิดพลาดที่เกิดขึ้นจริงเข้ากับตัวแปรที่ตั้งชื่อไว้ ทำให้เรียกใช้ตัวแปรนั้นภายในบล็อกดักข้อผิดพลาดได้เหมือน
ตัวแปรทั่วไป

เมื่อพิมพ์ตัวแปรนี้ผ่าน `print()` โดยตรง Python จะแปลงอ็อบเจกต์เป็นข้อความให้อัตโนมัติ ซึ่งได้ข้อความวินิจฉัยที่ Python
สร้างขึ้นเองขณะเกิดข้อผิดพลาดจริง แตกต่างจากข้อความที่ผู้เขียนโปรแกรมพิมพ์กำหนดเองในบล็อกดักข้อผิดพลาด (ดังที่ทำใน
หัวข้อที่ 2–7) ข้อความจริงนี้มักมีรายละเอียดเชิงเทคนิคมากกว่า จึงเหมาะกับการบันทึกเพื่อการแก้ไขข้อบกพร่อง (debug)
มากกว่าการแสดงต่อผู้ใช้ปลายทางโดยตรง

### ตัวอย่าง Python

#### ตัวอย่างที่ 1 (หลัก): หารจำนวนแล้วแสดงข้อความข้อผิดพลาดจริง

````python
def divide(a, b):
    try:
        result = a / b
        print("ผลหาร:", result)
    except ZeroDivisionError as e:
        print("เกิดข้อผิดพลาด:", e)

divide(10, 2)
divide(10, 0)
````

**Output:**
```
ผลหาร: 5.0
เกิดข้อผิดพลาด: division by zero
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | คำอธิบายเพิ่มเติม |
|---|---|---|---|
| 8 | เรียก `divide(10, 2)` | a=10, b=2 | |
| 3 | `result = a / b` | result=5.0 | |
| 4 | `print(...)` | แสดง `ผลหาร: 5.0` | |
| 9 | เรียก `divide(10, 0)` | a=10, b=0 | |
| 3 | `result = a / b` | เกิด ZeroDivisionError | หารด้วยศูนย์ |
| 5 | `except ZeroDivisionError as e` | e ผูกกับอ็อบเจกต์ข้อผิดพลาดที่เกิดขึ้นจริง | ตัวแปร e เก็บรายละเอียดของข้อผิดพลาดนี้ไว้ |
| 6 | `print("เกิดข้อผิดพลาด:", e)` | แสดง `เกิดข้อผิดพลาด: division by zero` | Python เรียก str(e) ให้อัตโนมัติเมื่อพิมพ์ e ผ่าน print — ข้อความนี้ Python สร้างขึ้นเอง ไม่ใช่ข้อความที่ผู้เขียนโปรแกรมพิมพ์เอง |

**คำอธิบายภาพรวม:** บรรทัด 1–6 นิยามฟังก์ชัน `divide` ที่คำนวณผลหารในบล็อกทดลองรัน หากตัวหารเป็นศูนย์จะเกิด
`ZeroDivisionError` ซึ่งบล็อกดักข้อผิดพลาดที่บรรทัด 5 จับไว้พร้อมผูกอ็อบเจกต์ข้อผิดพลาดนั้นเข้ากับตัวแปร `e` บรรทัด 6
พิมพ์ `e` โดยตรง ได้ข้อความ "division by zero" ซึ่งเป็นข้อความที่ Python สร้างขึ้นเองภายในขณะสร้างอ็อบเจกต์
`ZeroDivisionError` ไม่ใช่ข้อความที่ผู้เขียนโปรแกรมกำหนดเองอย่างในหัวข้อก่อนหน้าทั้งหมด

**จุดสังเกต:**
- ข้อความ "division by zero" ไม่ได้ถูกเขียนโดยผู้เขียนโปรแกรมเลย แต่เป็นข้อความที่ Python สร้างขึ้นเองเมื่อสร้าง
  อ็อบเจกต์ `ZeroDivisionError` นี่คือความแตกต่างจากหัวข้อที่ผ่านมา (หัวข้อที่ 2–7) ที่ข้อความในบล็อกดักข้อผิดพลาด
  ล้วนเป็นข้อความที่ผู้เขียนโปรแกรมกำหนดเองทั้งหมด
- `print("เกิดข้อผิดพลาด:", e)` พิมพ์ `e` ได้โดยตรงโดยไม่ต้องแปลงชนิดข้อมูลเอง เพราะ Python เรียก `str(e)` ให้
  อัตโนมัติเมื่อพิมพ์อ็อบเจกต์ผ่าน `print()`
- ตัวแปร `e` มีขอบเขตจำกัดอยู่เฉพาะภายในบล็อกดักข้อผิดพลาดเท่านั้น (เป็นกฎเฉพาะของ Python 3) เมื่อออกจากบล็อก
  except ไปแล้ว ตัวแปร `e` จะถูกลบทิ้งโดยอัตโนมัติ ไม่สามารถเรียกใช้ `e` ต่อภายนอกบล็อก except ได้ (ยืนยันด้วยการรัน
  จริง: การพิมพ์ `e` นอกบล็อก except ก่อให้เกิด `NameError: name 'e' is not defined`)

#### ตัวอย่างที่ 2 (โดเมนต่างออกไป): ลงทะเบียนอายุผู้ใช้งาน

````python
def register_age(text):
    try:
        age = int(text)
        print("อายุที่ลงทะเบียน:", age)
    except ValueError as e:
        print("ข้อความข้อผิดพลาดจริง:", e)

register_age("25")
register_age("ยี่สิบห้า")
````

**Output:**
```

ข้อความข้อผิดพลาดจริง: invalid literal for int() with base 10: 'ยี่สิบห้า'
```

| 8 | เรียก `register_age("25")` | text="25" | |
| 3 | `age = int(text)` | age=25 | |

| 3 | `age = int(text)` | เกิด ValueError | ข้อความนี้ไม่ใช่รูปแบบตัวเลข |
| 5 | `except ValueError as e` | e ผูกกับอ็อบเจกต์ข้อผิดพลาดที่เกิดขึ้นจริง | |
| 6 | `print(...)` | แสดง `ข้อความข้อผิดพลาดจริง: invalid literal for int() with base 10: 'ยี่สิบห้า'` | ข้อความนี้ Python สร้างขึ้นเอง มีรายละเอียดฐานเลขที่ใช้แปลง (base 10) ปรากฏอยู่ด้วย |

**คำอธิบายภาพรวม:** บรรทัด 1–6 นิยามฟังก์ชัน `register_age` ที่พยายามแปลงข้อความเป็นจำนวนเต็ม เมื่อแปลงไม่สำเร็จ
(บรรทัด 9) จะเกิด `ValueError` ซึ่งถูกผูกเข้ากับตัวแปร `e` ที่บรรทัด 5 ข้อความจริงที่พิมพ์ออกมาที่บรรทัด 6 คือ
"invalid literal for int() with base 10: 'ยี่สิบห้า'" ซึ่งเป็นข้อความวินิจฉัยของ Python เองทั้งประโยค ไม่ใช่ข้อความ
ที่ผู้เขียนโปรแกรมพิมพ์กำหนดเองอย่างในหัวข้อที่ 6 (ตัวอย่างที่ 1 ของหัวข้อนั้นใช้ข้อความ "แปลงคะแนนไม่สำเร็จ" ซึ่งเป็น
ข้อความที่ผู้เขียนโปรแกรมกำหนดเอง)

**จุดสังเกต:**
- ข้อความ "invalid literal for int() with base 10: 'ยี่สิบห้า'" เป็นข้อความวินิจฉัยของ Python เอง มีรายละเอียด
  เชิงเทคนิค (ระบุฐานเลขที่ใช้แปลง) มากกว่าข้อความที่ผู้เขียนโปรแกรมมักเขียนเองในบล็อกดักข้อผิดพลาด จึงเหมาะกับการ
  บันทึกเพื่อการแก้ไขข้อบกพร่องมากกว่าการแสดงต่อผู้ใช้ปลายทางโดยตรง
- ข้อความจริงจาก `e` มีค่าที่ผู้เรียนป้อนเข้ามาปรากฏอยู่ในเครื่องหมายคำพูดเดี่ยว ('ยี่สิบห้า') ซึ่งยืนยันโดยตรงว่า `e`
  มีรายละเอียดของสถานการณ์ที่ทำให้เกิดข้อผิดพลาดจริง ไม่ใช่ข้อความทั่วไปที่ตายตัวเหมือนกันทุกครั้ง

#### ตัวอย่างที่ 3 (edge case): รูปแบบข้อความที่ต่างกันไปตามชนิดข้อผิดพลาด

````python
inventory = {"apple": 10, "banana": 5}

def check_stock(item):
    try:
        quantity = inventory[item]
        print("จำนวนคงเหลือ:", quantity)
    except KeyError as e:
        print("ชนิดข้อผิดพลาด:", type(e).__name__)
        print("ข้อความจาก str(e):", str(e))
        print("ข้อความจาก e.args:", e.args)

check_stock("apple")
check_stock("mango")
````

**Output:**
```
จำนวนคงเหลือ: 10
ชนิดข้อผิดพลาด: KeyError
ข้อความจาก str(e): 'mango'
ข้อความจาก e.args: ('mango',)
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | คำอธิบายเพิ่มเติม |
|---|---|---|---|
| 12 | เรียก `check_stock("apple")` | item="apple" | |
| 5 | `quantity = inventory[item]` | quantity=10 | |
| 6 | `print(...)` | แสดง `จำนวนคงเหลือ: 10` | |
| 13 | เรียก `check_stock("mango")` | item="mango" | |
| 5 | `quantity = inventory[item]` | เกิด KeyError | ไม่มีคีย์ "mango" ในดิกชันนารี |
| 7 | `except KeyError as e` | e ผูกกับอ็อบเจกต์ KeyError | |
| 8 | `print(...)` | แสดง `ชนิดข้อผิดพลาด: KeyError` | `type(e).__name__` คืนชื่อชนิดของอ็อบเจกต์เป็นข้อความ |
| 9 | `print(...)` | แสดง `ข้อความจาก str(e): 'mango'` | str(e) ของ KeyError แสดงเพียงคีย์ที่ไม่พบ (มีเครื่องหมายคำพูดล้อมรอบ) ไม่ใช่ประโยคอธิบายเต็มรูปแบบ |
| 10 | `print(...)` | แสดง `ข้อความจาก e.args: ('mango',)` | e.args คือ tuple ของอาร์กิวเมนต์ที่ใช้สร้างอ็อบเจกต์ข้อผิดพลาดนี้ |

**คำอธิบายภาพรวม:** บรรทัด 3–10 นิยามฟังก์ชัน `check_stock` ที่ตรวจสอบจำนวนสินค้าคงเหลือจากดิกชันนารี `inventory`
เมื่อไม่พบสินค้าที่ระบุ (บรรทัด 13) จะเกิด `KeyError` ซึ่งถูกผูกเข้ากับตัวแปร `e` แล้วแสดงรายละเอียดสามรูปแบบ ได้แก่
ชนิดของข้อผิดพลาด (บรรทัด 8), ข้อความจาก `str(e)` (บรรทัด 9) และค่าดิบจาก `e.args` (บรรทัด 10) ผลลัพธ์แสดงให้
เห็นว่า `str(e)` ของ `KeyError` ให้เพียงคีย์ที่ไม่พบในรูปแบบ `repr` (มีเครื่องหมายคำพูดล้อมรอบ) ต่างจาก
`ZeroDivisionError` และ `ValueError` ในตัวอย่างที่ 1 และ 2 ที่ให้ประโยคอธิบายเต็มรูปแบบ

**จุดสังเกต:**
- `str(e)` ของ `KeyError` แสดงผลเป็น `'mango'` (มีเครื่องหมายคำพูดล้อมรอบ) ไม่ใช่ประโยคอธิบายแบบ
  `ZeroDivisionError` ("division by zero") หรือ `ValueError` ("invalid literal...") ในตัวอย่างก่อนหน้า เพราะ
  `KeyError` เก็บเฉพาะคีย์ที่ไม่พบไว้ใน args แล้วแสดงด้วยรูปแบบ `repr` ของคีย์นั้น ไม่ได้สร้างประโยคอธิบายแบบ
  ข้อผิดพลาดชนิดอื่น
- `e.args` คือ tuple ที่เก็บอาร์กิวเมนต์ทั้งหมดที่ถูกส่งเข้าไปตอนสร้างอ็อบเจกต์ข้อผิดพลาด ในกรณีนี้มีสมาชิกเพียงตัว
  เดียวคือคีย์ที่ไม่พบ `('mango',)` การเรียก `str(e)` ภายในคือการแปลง `args` เหล่านี้ให้เป็นข้อความตามรูปแบบเฉพาะ
  ของข้อผิดพลาดแต่ละชนิด ซึ่งแตกต่างกันไปตามชนิด
- ตัวอย่างนี้แสดงให้เห็นว่า "ข้อความข้อผิดพลาดจริง" ที่ได้จาก `except ... as e` ไม่มีรูปแบบเดียวกันตายตัวในทุกชนิด
  ของข้อผิดพลาด ผู้เขียนโปรแกรมจึงไม่ควรคาดหวังรูปแบบข้อความที่ตายตัวเมื่อออกแบบระบบที่ต้องแยกวิเคราะห์ข้อความเหล่านี้

## 9: การโยนข้อผิดพลาดเองด้วย `raise`

### แนวคิด

นอกเหนือจากข้อผิดพลาดที่เกิดขึ้นเองตามธรรมชาติจากการทำงานของภาษาโปรแกรม (เช่น การหารด้วยศูนย์ หรือการแปลงชนิด
ข้อมูลผิดพลาด) ผู้เขียนโปรแกรมยังสามารถกำหนดเงื่อนไขการตรวจสอบ (validation) เฉพาะของระบบที่ตนพัฒนาขึ้นเองได้ เช่น
อายุต้องไม่ติดลบ หรือจำนวนสินค้าที่สั่งซื้อต้องไม่เกินจำนวนคงเหลือในคลัง เมื่อพบว่าเงื่อนไขเหล่านี้ไม่เป็นจริง ภาษาโปรแกรม
จำนวนมากมีคำสั่งเฉพาะสำหรับ "โยน" ข้อผิดพลาดขึ้นมาเองในจุดนั้นทันที เพื่อส่งต่อควบคุมไปยังตัวจัดการข้อผิดพลาดที่ใกล้
ที่สุดในสายการเรียกใช้งาน เสมือนว่าเป็นข้อผิดพลาดที่เกิดขึ้นเองตามธรรมชาติ ในภาษา Python คำสั่งนี้คือ `raise`

รูปแบบการเขียนคือ `raise ExceptionType("ข้อความ")` โดยเลือกชนิดข้อผิดพลาดที่มีอยู่แล้วในภาษา (เช่น `ValueError`
สำหรับกรณีค่าที่ได้รับไม่ถูกต้องตามเงื่อนไข) พร้อมข้อความอธิบายที่ผู้เขียนโปรแกรมกำหนดขึ้นเอง หากไม่มีบล็อกดัก
ข้อผิดพลาดใดในสายการเรียกใช้งานดักจับข้อผิดพลาดที่ถูกโยนนี้ไว้ได้ ข้อผิดพลาดจะแพร่กระจายขึ้นไปเรื่อย ๆ จนถึงจุดเริ่มต้น
ของโปรแกรมแล้วทำให้โปรแกรมหยุดทำงานพร้อมแสดง traceback เช่นเดียวกับข้อผิดพลาดที่เกิดขึ้นเองตามธรรมชาติทุกประการ
`raise` จึงไม่ได้เป็นกลไกที่แยกจาก try/except ที่เคยเรียนมา แต่เป็นอีกช่องทางหนึ่งในการสร้างข้อผิดพลาดที่ใช้กลไกการ
ดักจับและแพร่กระจายเดียวกันทั้งหมด

#### ตัวอย่างที่ 1 (หลัก): ตรวจสอบอายุก่อนลงทะเบียนผู้ใช้งาน

````python
def register_user(age):
    if age < 0:
        raise ValueError("อายุต้องไม่ติดลบ")
    print("ลงทะเบียนสำเร็จ อายุ:", age)

register_user(20)
try:
    register_user(-5)
except ValueError as e:
    print("ลงทะเบียนไม่สำเร็จ:", e)
````

**Output:**
```
ลงทะเบียนสำเร็จ อายุ: 20
```

**คำอธิบายภาพรวม:** บรรทัด 1–4 นิยามฟังก์ชัน `register_user` ที่ตรวจสอบเงื่อนไขทางธุรกิจของระบบนี้เอง คืออายุต้อง
ไม่ติดลบ หากไม่เป็นไปตามเงื่อนไข จะโยน `ValueError` ขึ้นเองที่บรรทัด 3 พร้อมข้อความที่ผู้เขียนโปรแกรมกำหนดขึ้นเอง
บรรทัด 6 เรียกฟังก์ชันด้วยค่าที่ถูกต้องจึงทำงานตามปกติ ส่วนบรรทัด 7–10 เรียกฟังก์ชันด้วยค่าที่ไม่ถูกต้องภายใน try/except
ทำให้ข้อผิดพลาดที่ถูกโยนออกมาถูกดักจับและแสดงข้อความที่กำหนดไว้แทนที่จะทำให้โปรแกรมหยุดทำงาน

**จุดสังเกต:**
- ข้อผิดพลาดที่เกิดจาก `raise` ที่บรรทัด 3 ถูกจัดการด้วยกลไก try/except เดียวกันทุกประการกับข้อผิดพลาดที่เกิดขึ้น
  เองตามธรรมชาติ (เช่น `ZeroDivisionError` ในหัวข้อที่ 8) ไม่มีวากยสัมพันธ์พิเศษเพิ่มเติมสำหรับการดักจับข้อผิดพลาด
  ที่เกิดจาก `raise`
- ข้อความข้อผิดพลาดในกรณีนี้ ("อายุต้องไม่ติดลบ") เป็นข้อความที่ผู้เขียนโปรแกรมกำหนดขึ้นเองทั้งหมด ต่างจากหัวข้อที่
  8 ที่ข้อความมาจาก Python โดยตรง นี่คือประโยชน์หลักของ `raise` คือช่วยให้ผู้เขียนโปรแกรมสื่อสารเงื่อนไขทางธุรกิจที่
  Python เองไม่มีทางรู้ล่วงหน้าได้
- เมื่อเกิด `raise` ที่บรรทัด 3 คำสั่งที่บรรทัด 4 จะไม่ถูกประมวลผลเลย เพราะการโยนข้อผิดพลาดทำให้ควบคุมออกจาก
  ฟังก์ชันทันที เช่นเดียวกับพฤติกรรมของข้อผิดพลาดตามธรรมชาติที่เคยเรียนในหัวข้อที่ 2

#### ตัวอย่างที่ 2 (โดเมนต่างออกไป): ตรวจสอบจำนวนสินค้าคงเหลือก่อนสั่งซื้อ

````python
stock = {"เสื้อยืด": 3}

def place_order(item, quantity):
    if quantity > stock.get(item, 0):
        raise ValueError("สินค้าในคลังไม่พอสำหรับคำสั่งซื้อนี้")
    print("สั่งซื้อสำเร็จ:", item, "จำนวน", quantity)

place_order("เสื้อยืด", 2)
try:
    place_order("เสื้อยืด", 10)
except ValueError as e:
    print("สั่งซื้อไม่สำเร็จ:", e)
````

**Output:**
```
สั่งซื้อสำเร็จ: เสื้อยืด จำนวน 2
สั่งซื้อไม่สำเร็จ: สินค้าในคลังไม่พอสำหรับคำสั่งซื้อนี้
```

|---|---|---|---|
| 8 | เรียก `place_order("เสื้อยืด", 2)` | item="เสื้อยืด", quantity=2 | |
| 4 | `if quantity > stock.get(item, 0)` | 2 > 3 → False | stock.get คืนค่า 3 |
| 6 | `print(...)` | แสดง `สั่งซื้อสำเร็จ: เสื้อยืด จำนวน 2` | |
| 10 | เรียก `place_order("เสื้อยืด", 10)` ภายในบล็อกทดลองรัน | item="เสื้อยืด", quantity=10 | |
| 4 | `if quantity > stock.get(item, 0)` | 10 > 3 → True | |

**คำอธิบายภาพรวม:** บรรทัด 3–6 นิยามฟังก์ชัน `place_order` ที่ตรวจสอบว่าจำนวนที่สั่งซื้อเกินจำนวนคงเหลือในคลังหรือ
ไม่ โดยใช้ `stock.get(item, 0)` แทนการเข้าถึงด้วย `[]` โดยตรง เพื่อป้องกัน `KeyError` ในกรณีที่สินค้านั้นไม่มีอยู่
ในคลังเลย (คืนค่า 0 แทน) หากจำนวนที่สั่งซื้อเกินกว่าจำนวนคงเหลือ จะโยน `ValueError` ขึ้นเองที่บรรทัด 5 บรรทัด 9–12
เรียกฟังก์ชันด้วยจำนวนที่เกินคงคลังภายใน try/except ทำให้ข้อผิดพลาดถูกดักจับและแสดงข้อความที่กำหนดไว้แทน

**จุดสังเกต:**
- `stock.get(item, 0)` ใช้เมธอด `get()` แทนการเข้าถึงด้วย `[]` โดยตรง เพื่อป้องกัน `KeyError` ในกรณีที่สินค้านั้น
  ไม่มีอยู่ในคลังเลย แสดงให้เห็นว่าการเขียนโปรแกรมเชิงป้องกัน (defensive programming) และการโยนข้อผิดพลาดเองด้วย
  `raise` สามารถใช้ร่วมกันในฟังก์ชันเดียวกันได้
- เงื่อนไข validation ในตัวอย่างนี้ (จำนวนที่สั่งซื้อต้องไม่เกินจำนวนคงเหลือ) เป็นกฎทางธุรกิจที่เฉพาะเจาะจงกับระบบนี้
  เท่านั้น ไม่มีทางที่ Python จะรู้ล่วงหน้าหรือสร้างข้อผิดพลาดให้เองโดยอัตโนมัติ ต้องอาศัย `raise` เท่านั้นจึงจะแจ้งเตือน
  กรณีนี้ได้
- โดเมนของตัวอย่างนี้ (ระบบสั่งซื้อสินค้า) แตกต่างจากตัวอย่างที่ 1 (การลงทะเบียนผู้ใช้งาน) อย่างชัดเจน แต่ใช้รูปแบบ
  โค้ดเดียวกันทุกประการ (ตรวจเงื่อนไข → raise พร้อมข้อความ → ดักจับด้วย try/except as e) แสดงให้เห็นว่ารูปแบบนี้
  นำไปใช้ซ้ำได้กับปัญหาที่หลากหลาย

#### ตัวอย่างที่ 3 (edge case): ข้อผิดพลาดที่ถูกโยนแต่ไม่มีการดักจับ แล้วแก้ไข

ก่อนพิจารณาตัวอย่างที่สมบูรณ์ ให้พิจารณาโค้ดต่อไปนี้ซึ่ง **ไม่มีการดักจับ** ข้อผิดพลาดที่ถูกโยนออกมาโดยเจตนา เพื่อสาธิต
ว่าจะเกิดอะไรขึ้น:

````python
def validate_temperature(celsius):
    if celsius < -273.15:
        raise ValueError("อุณหภูมิต่ำกว่าศูนย์องศาสัมบูรณ์")
    print("อุณหภูมิถูกต้อง:", celsius)

validate_temperature(25)
validate_temperature(-300)
print("บรรทัดนี้จะไม่ถูกประมวลผล")
````

โค้ดนี้ผ่านการรันจริงแล้ว (ไม่ใช่การคาดเดา) Output จริงคือ:

```

Traceback (most recent call last):
  ...
ValueError: อุณหภูมิต่ำกว่าศูนย์องศาสัมบูรณ์
```

บรรทัดที่ 8 (`print("บรรทัดนี้จะไม่ถูกประมวลผล")`) ไม่ถูกประมวลผลเลย เพราะข้อผิดพลาดที่ถูกโยนออกมาจากบรรทัด 7 ไม่มี
ผู้ใดดักจับ จึงแพร่กระจายขึ้นไปจนถึงจุดเริ่มต้นของโปรแกรมและทำให้โปรแกรมทั้งหมดหยุดทำงานทันที ปัญหานี้แก้ไขได้ด้วยการ
ครอบการเรียกฟังก์ชันด้วย try/except ดังนี้ (นี่คือตัวอย่างฉบับสมบูรณ์ที่นับเป็นตัวอย่างที่ 3):

````python
def validate_temperature(celsius):
    if celsius < -273.15:
        raise ValueError("อุณหภูมิต่ำกว่าศูนย์องศาสัมบูรณ์")
    print("อุณหภูมิถูกต้อง:", celsius)

validate_temperature(25)
try:
    validate_temperature(-300)
except ValueError as e:
    print("ตรวจพบค่าที่ไม่ถูกต้อง:", e)
print("โปรแกรมทำงานต่อได้ตามปกติ")
````

**Output:**
```

ตรวจพบค่าที่ไม่ถูกต้อง: อุณหภูมิต่ำกว่าศูนย์องศาสัมบูรณ์

```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | คำอธิบายเพิ่มเติม |

| 6 | เรียก `validate_temperature(25)` | celsius=25 | |

| 8 | เรียก `validate_temperature(-300)` ภายในบล็อกทดลองรัน | celsius=-300 | |
| 2 | `if celsius < -273.15` | True | |
| 3 | `raise ValueError(...)` | โยน ValueError ขึ้นเอง | ในโค้ดชุดแรก (ไม่มี try/except) ข้อผิดพลาดนี้ไม่มีผู้ใดดักจับ โปรแกรมจึงหยุดทำงานทันทีหลังบรรทัดนี้ |
| 9 | `except ValueError as e` | e ผูกกับอ็อบเจกต์ที่ถูกโยน | ในโค้ดชุดที่สอง (มี try/except) ข้อผิดพลาดถูกดักจับไว้ที่นี่แทน |
| 10 | `print(...)` | แสดง `ตรวจพบค่าที่ไม่ถูกต้อง: อุณหภูมิต่ำกว่าศูนย์องศาสัมบูรณ์` | |
| 11 | `print(...)` | แสดง `โปรแกรมทำงานต่อได้ตามปกติ` | บรรทัดนี้ประมวลผลได้สำเร็จ เพราะข้อผิดพลาดถูกดักจับไว้แล้วก่อนหน้า ต่างจากโค้ดชุดแรกที่บรรทัดถัดจากจุดเกิดข้อผิดพลาดไม่ถูกประมวลผลเลย |

**คำอธิบายภาพรวม:** โค้ดชุดแรก (`validate_temperature` ที่เรียกโดยไม่มี try/except ครอบ) แสดงให้เห็นผลลัพธ์จริง
เมื่อข้อผิดพลาดที่ถูกโยนด้วย `raise` ไม่มีผู้ใดดักจับ คือโปรแกรมทั้งหมดหยุดทำงานทันทีที่บรรทัด 7 พร้อมแสดง traceback
บรรทัดที่ 8 ซึ่งอยู่ถัดจากจุดเกิดข้อผิดพลาดจึงไม่ถูกประมวลผลเลย ส่วนโค้ดชุดที่สองแก้ไขปัญหานี้ด้วยการครอบการเรียก
ฟังก์ชันด้วย try/except ที่บรรทัด 7–10 ทำให้ข้อผิดพลาดที่ถูกโยนออกมาถูกดักจับไว้ได้ โปรแกรมจึงทำงานต่อไปได้ตามปกติ
จนถึงบรรทัดสุดท้าย (บรรทัด 11) แสดงให้เห็นว่า `raise` เพียงอย่างเดียวไม่ได้ทำให้โปรแกรมปลอดภัยขึ้นโดยอัตโนมัติ ยังคง
ต้องจับคู่กับ try/except เพื่อดักจับเสมอ

**จุดสังเกต:**
- ข้อผิดพลาดที่เกิดจาก `raise` ที่ไม่มีการดักจับมีพฤติกรรมเหมือนข้อผิดพลาดตามธรรมชาติที่ไม่มีการดักจับทุกประการ คือ
  ทำให้โปรแกรมทั้งหมดหยุดทำงานทันที ไม่ใช่แค่ฟังก์ชันที่เรียก `raise` เท่านั้น
- คำสั่งที่อยู่ถัดจากจุดเกิดข้อผิดพลาดที่ไม่มีการดักจับ (บรรทัด 8 ของโค้ดชุดแรก) จะไม่ถูกประมวลผลเลยไม่ว่าจะอยู่ใน
  ฟังก์ชันเดียวกันหรือเป็นคำสั่งระดับบนสุดของโปรแกรมก็ตาม
- การเปรียบเทียบโค้ดสองชุดในตัวอย่างนี้แสดงให้เห็นชัดเจนว่า `raise` กับ `try/except` เป็นกลไกที่ต้องใช้คู่กันเสมอ
  หากต้องการให้โปรแกรมทำงานต่อได้หลังพบเงื่อนไขที่ไม่ถูกต้อง — `raise` เพียงอย่างเดียวทำหน้าที่แจ้งปัญหา ส่วน
  `try/except` ทำหน้าที่รับมือกับปัญหานั้นเพื่อให้โปรแกรมทำงานต่อได้

### หัวข้อที่ 6: ส่วน else (เมื่อไม่มีข้อผิดพลาด) (ข้อ 1–5)

**ข้อ 1 (อธิบายแนวคิด)** จงอธิบายว่าเหตุใดการวางคำสั่งที่ไม่มีโอกาสก่อข้อผิดพลาดไว้ในส่วน `else` แทนที่จะวาง
ต่อท้ายในบล็อกทดลองรัน (try) จึงช่วยแยกแยะ "โค้ดที่เสี่ยงต่อข้อผิดพลาด" ออกจาก "โค้ดที่ควรทำงานเฉพาะเมื่อ
สำเร็จ" ได้ชัดเจนขึ้น พร้อมระบุว่าบล็อกดักข้อผิดพลาด (except) ที่เขียนคู่กันคุ้มครองส่วนใดของโครงสร้างนี้บ้าง

**ข้อ 2 (ทายผลลัพธ์)** กำหนดโค้ดต่อไปนี้ จงทายผลลัพธ์ที่จะปรากฏบนหน้าจอ
```python
01  def borrow_book(book_id):
02      catalog = {"B001": "มีอยู่", "B002": "ถูกยืมแล้ว"}
03      try:
04          status = catalog[book_id]
05      except KeyError:
06          print("ไม่พบรหัสหนังสือ:", book_id)
07      else:
08          print("สถานะหนังสือ", book_id, ":", status)
09
10  borrow_book("B001")
11  borrow_book("B999")
```

**ข้อ 3 (หาข้อผิดพลาด)** โค้ดต่อไปนี้ตั้งใจใช้ส่วน `else` เพื่อแสดงสถานะที่นั่ง แต่กลับทำให้โปรแกรมหยุดทำงาน
เมื่อเรียกด้วยตำแหน่งที่นั่งที่ไม่มีอยู่จริง จงระบุว่าเหตุใดบล็อก `except IndexError` ที่เขียนไว้จึงดักจับข้อผิดพลาด
นี้ไม่ได้
```python
01  def check_seat(seat_map, seat_no):
02      try:
03          pass
04      except IndexError:
05          print("ไม่มีที่นั่งนี้")
06      else:
07          seat = seat_map[seat_no]
08          print("สถานะที่นั่ง:", seat)
09
10  check_seat(["ว่าง", "จอง"], 0)
11  check_seat(["ว่าง", "จอง"], 5)
12  print("จบโปรแกรม")
```

**ข้อ 4 (เขียนโค้ด)** จงเขียนฟังก์ชันชื่อ `check_ticket_number(text)` ที่ใช้ `try`/`except` แปลงข้อความเป็น
จำนวนเต็มในบล็อกทดลองรัน หากแปลงไม่สำเร็จให้พิมพ์ข้อความแจ้งว่าหมายเลขตั๋วไม่ถูกต้อง หากแปลงสำเร็จให้ใช้
ส่วน `else` พิมพ์ข้อความยืนยันการรับตั๋ว แล้วทดสอบเรียกใช้ด้วยข้อความ `"45"` และ `"สี่สิบห้า"`

**ข้อ 5 (ทายผลลัพธ์)** กำหนดโค้ดต่อไปนี้ จงทายผลลัพธ์ที่จะปรากฏบนหน้าจอ
```python
01  exchange_rate = {"USD": 36.5, "EUR": 39.2}
02
03  def show_rate(currency):
04      try:
05          rate = exchange_rate[currency]
06      except KeyError:
07          print("ไม่มีอัตราแลกเปลี่ยนของสกุลเงิน:", currency)
08      else:
09          print("อัตราแลกเปลี่ยนของ", currency, "คือ", rate)
10
11  show_rate("USD")
12  show_rate("JPY")
```

### หัวข้อที่ 7: ส่วน finally (รันเสมอ) (ข้อ 6–10)

**ข้อ 6 (อธิบายแนวคิด)** จงอธิบายว่าส่วน `finally` ต่างจากส่วน `else` (หัวข้อที่ 6) อย่างไรในแง่ของเงื่อนไขที่
ทำให้แต่ละส่วนทำงาน พร้อมระบุว่า `finally` มีบทบาทในการดักจับหรือระงับข้อผิดพลาดด้วยตัวเองหรือไม่

**ข้อ 7 (ทายผลลัพธ์)** กำหนดโค้ดต่อไปนี้ จงทายผลลัพธ์ที่จะปรากฏบนหน้าจอ (สังเกตว่าโค้ดนี้ไม่มีบล็อกดักข้อผิดพลาด)
```python
01  def print_receipt(total, num_items):
02      try:
03          per_item = total / num_items
04          print("ราคาต่อชิ้น:", per_item)
05      finally:
06          print("พิมพ์ใบเสร็จเสร็จสิ้น")
07
08  print_receipt(300, 3)
09  print_receipt(300, 0)
10  print("จบโปรแกรม")
```

**ข้อ 8 (หาข้อผิดพลาด)** ฟังก์ชันต่อไปนี้มี `finally` ครอบอยู่ แต่การเรียกใช้ครั้งที่สองยังคงทำให้โปรแกรมหยุด
ทำงานด้วย traceback จงระบุว่าเหตุใด `finally` จึงไม่ได้ป้องกันไม่ให้โปรแกรมหยุดทำงาน และ `finally` ทำหน้าที่
อะไรกันแน่ในสถานการณ์นี้
```python
01  def withdraw(balance, amount):
02      try:
03          new_balance = balance - amount
04          if new_balance < 0:
05              raise ValueError("ยอดเงินไม่พอ")
06          return new_balance
07      finally:
08          print("บันทึกรายการถอนเงินเสร็จสิ้น")
09
10  print(withdraw(1000, 500))
11  print(withdraw(1000, 2000))
```

**ข้อ 9 (เขียนโค้ด)** จงเขียนฟังก์ชันชื่อ `process_payment(amount)` ที่คำนวณค่าธรรมเนียมต่อหน่วยด้วย
`1000 / amount` ในบล็อกทดลองรัน ดักจับ `ZeroDivisionError` โดยคืนค่า `None` แทนเมื่อ `amount` เป็นศูนย์
และใช้ `finally` พิมพ์ข้อความว่าปิดการเชื่อมต่อระบบชำระเงินเสมอไม่ว่าผลลัพธ์จะเป็นอย่างไร แล้วทดสอบเรียกใช้
ด้วยค่า `200` และ `0`

**ข้อ 10 (หาข้อผิดพลาด)** ฟังก์ชันต่อไปนี้ดักจับเฉพาะ `TypeError` แต่การเรียกใช้ครั้งที่สองยังคงทำให้โปรแกรม
หยุดทำงาน จงระบุว่าชนิดข้อผิดพลาดที่เกิดขึ้นจริงคือชนิดใด และ `finally` ทำงานก่อนหรือหลังโปรแกรมหยุดทำงาน
```python
01  def shipping_cost(weight_kg, zone):
02      rate_table = {"A": 20, "B": 35}
03      try:
04          rate = rate_table[zone]
05          cost = weight_kg * rate
06          return cost
07      except TypeError:
08          return None
09      finally:
10          print("คำนวณค่าจัดส่งเสร็จสิ้น")
11
12  print(shipping_cost(5, "A"))
13  print(shipping_cost(5, "C"))
```

### หัวข้อที่ 8: จับข้อผิดพลาดเป็นตัวแปรด้วย `except ... as` (ข้อ 11–15)

**ข้อ 11 (อธิบายแนวคิด)** จงอธิบายความแตกต่างระหว่างข้อความข้อผิดพลาดที่ผู้เขียนโปรแกรมกำหนดเองไว้ในบล็อก
ดักข้อผิดพลาด กับข้อความที่ได้จากการผูกตัวแปรด้วย `except ... as e` แล้วพิมพ์ตัวแปรนั้นโดยตรง พร้อมระบุขอบเขต
(scope) ของตัวแปรที่ผูกไว้ด้วยลักษณะนี้

**ข้อ 12 (ทายผลลัพธ์)** กำหนดโค้ดต่อไปนี้ จงทายผลลัพธ์ที่จะปรากฏบนหน้าจอ
```python
01  def get_answer(answers, index):
02      try:
03          return answers[index]
04      except IndexError as e:
05          print("ข้อผิดพลาด:", e)
06          return None
07
08  print(get_answer(["A", "B", "C"], 1))
09  print(get_answer(["A", "B", "C"], 5))
```

**ข้อ 13 (หาข้อผิดพลาด)** ฟังก์ชันต่อไปนี้ทำงานได้ตามปกติกับการเรียกใช้ครั้งแรก แต่การเรียกใช้ครั้งที่สองกลับ
ทำให้โปรแกรมหยุดทำงานด้วยข้อผิดพลาดอีกชนิดหนึ่งที่ไม่ใช่ `ValueError` จงระบุว่าเหตุใดจึงเป็นเช่นนั้น
```python
01  def read_temperature(text):
02      try:
03          return float(text)
04      except ValueError as e:
05          print("อ่านค่าไม่สำเร็จ")
06      print("รายละเอียดข้อผิดพลาด:", e)
07      return None
08
09  print(read_temperature("36.5"))
10  print(read_temperature("ร้อน"))
```

**ข้อ 14 (เขียนโค้ด)** จงเขียนฟังก์ชันชื่อ `parse_quantity(text)` ที่ใช้ `try`/`except ValueError as e` แปลง
ข้อความเป็นจำนวนเต็ม หากสำเร็จให้พิมพ์และคืนค่าจำนวนนั้น หากไม่สำเร็จให้พิมพ์ทั้งชนิดของข้อผิดพลาด
(`type(e).__name__`) และข้อความข้อผิดพลาดจริงจาก `e` แล้วคืนค่า `None` แทน แล้วทดสอบเรียกใช้ด้วยข้อความ
`"10"` และ `"ten"`

**ข้อ 15 (เขียนโค้ด)** จงเขียนฟังก์ชันชื่อ `lookup_price(price_table, product)` ที่ค้นหาราคาสินค้าจากดิกชันนารี
`price_table` ในบล็อกทดลองรัน ดักจับ `KeyError as e` โดยพิมพ์ทั้งข้อความแจ้งว่าไม่พบสินค้าและค่าจาก `e.args`
แล้วคืนค่า `None` แทน แล้วทดสอบเรียกใช้กับดิกชันนารี `{"ปากกา": 15, "ยางลบ": 5}` โดยค้นหา `"ปากกา"` และ
`"ไม้บรรทัด"`

### หัวข้อที่ 9: การโยนข้อผิดพลาดเองด้วย `raise` (ข้อ 16–20)

**ข้อ 16 (อธิบายแนวคิด)** จงอธิบายว่าคำสั่ง `raise` มีบทบาทอย่างไรในการตรวจสอบเงื่อนไขทางธุรกิจ (validation)
ที่ภาษาโปรแกรมไม่สามารถตรวจพบได้เอง และเพราะเหตุใดข้อผิดพลาดที่เกิดจาก `raise` จึงถูกดักจับด้วยกลไก
try/except ชุดเดียวกันทุกประการกับข้อผิดพลาดที่เกิดขึ้นเองตามธรรมชาติ

**ข้อ 17 (ทายผลลัพธ์)** กำหนดโค้ดต่อไปนี้ จงทายผลลัพธ์ที่จะปรากฏบนหน้าจอ
```python
01  def validate_score(score):
02      if score < 0 or score > 100:
03          raise ValueError("คะแนนต้องอยู่ระหว่าง 0 ถึง 100")
04      print("คะแนนถูกต้อง:", score)
05
06  validate_score(85)
07  try:
08      validate_score(150)
09  except ValueError as e:
10      print("พบข้อผิดพลาด:", e)
```

**ข้อ 18 (หาข้อผิดพลาด)** โค้ดต่อไปนี้ตรวจสอบน้ำหนักพัสดุด้วย `raise` แต่การเรียกใช้ครั้งที่สองทำให้โปรแกรม
ทั้งหมดหยุดทำงาน จงระบุว่าเหตุใดโปรแกรมจึงหยุดทำงาน และคำสั่งใดที่ไม่ถูกประมวลผลเลย
```python
01  def validate_weight(weight_kg):
02      if weight_kg > 30:
03          raise ValueError("น้ำหนักเกินขีดจำกัดการจัดส่ง")
04      print("น้ำหนักอยู่ในเกณฑ์ที่รับส่งได้:", weight_kg)
05
06  validate_weight(10)
07  validate_weight(50)
08  print("เตรียมพัสดุเพื่อจัดส่ง")
```

**ข้อ 19 (เขียนโค้ด)** จงเขียนฟังก์ชันชื่อ `set_thermostat(temp)` ที่โยน `ValueError` ด้วย `raise` เมื่อ `temp`
น้อยกว่า 16 หรือมากกว่า 30 พร้อมข้อความอธิบายที่กำหนดขึ้นเอง หากค่าถูกต้องให้พิมพ์ข้อความยืนยันการตั้งค่า
แล้วทดสอบเรียกใช้ภายใน try/except ด้วยค่า `24` และ `40`

**ข้อ 20 (อธิบายแนวคิด)** จงอธิบายว่าเหตุใดข้อผิดพลาดที่เกิดจากคำสั่ง `raise` กับข้อผิดพลาดที่เกิดขึ้นเองตาม
ธรรมชาติจากการทำงานของภาษา (เช่น `ZeroDivisionError` หรือ `KeyError`) จึงถูกดักจับด้วยกลไกเดียวกันทุก
ประการ โดยยกตัวอย่างสถานการณ์หนึ่งที่ต่างจากเนื้อหาข้างต้นเพื่อประกอบคำอธิบาย
