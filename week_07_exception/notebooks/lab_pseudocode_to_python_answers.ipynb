```python
def parse_free_storage(text):
    try:
        free_mb = int(text)
        return free_mb
    except ValueError:
        print("ข้อมูลพื้นที่ว่างไม่ถูกต้อง")
        return 0


print(parse_free_storage("2048"))
print(parse_free_storage("สองพันสี่สิบแปด"))
```

#### เฉลยข้อที่ 2

```python
def calculate_laundry_fee(num_loads_text, price_per_load):
    try:
        num_loads = int(num_loads_text)
        total_fee = num_loads * price_per_load
        return total_fee
    except ValueError:
        return "จำนวนถังผ้าไม่ถูกต้อง"
    except TypeError:
        return "ราคาต่อถังไม่ถูกต้อง"


print(calculate_laundry_fee("3", 40))
print(calculate_laundry_fee("สาม", 40))
print(calculate_laundry_fee("3", None))
```

```python
def get_daily_steps(step_list, day_index):
    try:
        return step_list[day_index]
    except IndexError:
        return 0


weekly_steps = [8000, 10500, 6200, 7300, 9100, 12000, 5000]
print(get_daily_steps(weekly_steps, 3))
print(get_daily_steps(weekly_steps, 9))
```

#### เฉลยข้อที่ 4

```python
def estimate_minutes_to_full(current_percent, charge_rate_per_minute):
    try:
        remaining_percent = 100 - current_percent
        return remaining_percent / charge_rate_per_minute
    except ZeroDivisionError:
        return -1


print(estimate_minutes_to_full(60, 0.5))
print(estimate_minutes_to_full(60, 0))
```

แบบเดียว) กรณีแรกแบตเตอรี่เหลืออีก 40% ต้องชาร์จอีกด้วยอัตรา 0.5% ต่อนาที จึงใช้เวลา 80.0 นาที ส่วนกรณีที่สอง
อัตราการชาร์จเป็น 0 (สายชาร์จหลุด) ฟังก์ชันจึงดักจับ `ZeroDivisionError` แล้วคืนค่า -1 แทนเพื่อสื่อว่าไม่สามารถ
ประมาณเวลาได้ในสถานการณ์นี้ แทนที่จะปล่อยให้โปรแกรมหยุดทำงาน

```python
def redeem_free_drink(points_text):
    try:
        points = int(points_text)
    except ValueError:
        print("ข้อมูลแต้มสะสมไม่ถูกต้อง")
    else:
        if points >= 50:
            print("แลกเครื่องดื่มฟรีได้สำเร็จ")
        else:
            print("แต้มสะสมไม่พอสำหรับแลกเครื่องดื่มฟรี")


redeem_free_drink("60")
redeem_free_drink("30")
redeem_free_drink("หกสิบ")
```

```python
scan_count = 0


def scan_ticket(code_text):
    global scan_count
    try:
        code = int(code_text)
        print("รับตั๋วหมายเลข", code, "เรียบร้อย")
    except ValueError:
        print("ตั๋วไม่ถูกต้อง")
    finally:
        scan_count += 1


scan_ticket("101")
scan_ticket("ABC")
scan_ticket("102")
print("จำนวนครั้งที่สแกนทั้งหมด:", scan_count)
```

```python
def enter_class(remaining_hours):
    if remaining_hours <= 0:
        raise ValueError("ชั่วโมงเรียนคงเหลือหมดแล้ว ไม่สามารถเข้าเรียนได้")
    return "เข้าเรียนสำเร็จ ชั่วโมงเรียนคงเหลือ " + str(remaining_hours) + " ชั่วโมง"


print(enter_class(5))
try:
    print(enter_class(0))
except ValueError as e:
    print("เข้าเรียนไม่สำเร็จ:", e)
```

### ระดับสูง

#### เฉลยข้อที่ 1

```python
class BidTooLowError(Exception):
    pass


def place_bid(current_highest, new_bid):
    if new_bid <= current_highest:
        raise BidTooLowError("ราคาประมูลต้องสูงกว่าราคาปัจจุบัน")
    return new_bid


try:
    place_bid(500, 400)
except BidTooLowError as e:
    print("ประมูลไม่สำเร็จ:", e)

print(place_bid(500, 600))
```

คำอธิบายเพิ่มเติม (ขอบเขตนอกเหนือจากสไลด์): ข้อผิดพลาดที่ผู้ใช้กำหนดเอง (custom exception) เป็นแนวคิดขั้นสูง
กว่า `slides/07_exception.md` ที่สอนจริงในสัปดาห์ 7 มาก ดูรายละเอียดใน `lecture_note_advanced.md` หัวข้อที่ 10
`BidTooLowError` สืบทอดจาก `Exception` โดยตรงโดยไม่เพิ่มพฤติกรรมใด ๆ ของตัวเอง (`pass`) เช่นเดียวกับตัวอย่างที่ 1
ของหัวข้อ 10 ในเอกสารนั้น

#### เฉลยข้อที่ 2

```python
daily_specials = {
    "จันทร์": ["ผัดไทย", "ข้าวผัดกุ้ง"],
    "อังคาร": ["ต้มยำกุ้ง"],
}


def get_special_menu(day, menu_index):
    try:
        return daily_specials[day][menu_index]
    except LookupError as e:
        print("ไม่พบเมนู (" + type(e).__name__ + "):", e)
        return None


print(get_special_menu("จันทร์", 0))
print(get_special_menu("จันทร์", 5))
print(get_special_menu("อาทิตย์", 0))
```

คำอธิบายเพิ่มเติม (ขอบเขตนอกเหนือจากสไลด์): ลำดับชั้นของข้อผิดพลาด (exception hierarchy) เป็นแนวคิดขั้นสูงกว่า
`slides/07_exception.md` มาก ดูรายละเอียดใน `lecture_note_advanced.md` หัวข้อที่ 11 `IndexError` และ `KeyError`
ต่างสืบทอดจาก `LookupError` เช่นเดียวกัน บล็อกดักข้อผิดพลาดเพียงบล็อกเดียว (`except LookupError as e`) จึงจับได้
ทั้งสองกรณี แต่ `type(e).__name__` ยังคงเผยให้เห็นชนิดที่แท้จริงของแต่ละกรณีไว้ให้ต่างกัน

คำอธิบายเพิ่มเติม (ขอบเขตนอกเหนือจากสไลด์): การยืนยันเงื่อนไข (assert) เป็นแนวคิดขั้นสูงกว่า `slides/07_exception.md`
มาก ดูรายละเอียดใน `lecture_note_advanced.md` หัวข้อที่ 12 การเรียกครั้งที่สอง (`start_match(15)`) ไม่มีบล็อก

เต็มรูปแบบออกมาแล้วโปรแกรมหยุดทำงาน (exit code ไม่เท่ากับ 0) ซึ่งเป็นพฤติกรรมที่ถูกต้องตามที่ออกแบบไว้

#### เฉลยข้อที่ 4

```python
def connect_printer(is_online):
    if not is_online:
        raise ConnectionError("ไม่สามารถเชื่อมต่อเครื่องพิมพ์ได้ เครื่องพิมพ์ไม่ได้เปิดอยู่")
    print("เชื่อมต่อเครื่องพิมพ์สำเร็จ")


def check_paper(pages_requested, pages_remaining):
    if pages_requested > pages_remaining:
        raise ValueError("จำนวนหน้าที่ขอพิมพ์เกินจำนวนกระดาษที่เหลือในถาด")
    print("ตรวจสอบกระดาษผ่าน เหลือเพียงพอสำหรับพิมพ์")


def print_document(pages_requested):
    print("กำลังพิมพ์เอกสารจำนวน", pages_requested, "หน้า")


def process_print_job(is_online, pages_requested, pages_remaining):
    try:
        try:
            connect_printer(is_online)
            check_paper(pages_requested, pages_remaining)
            print_document(pages_requested)
        except Exception as e:
            print("บันทึก log ข้อผิดพลาด:", type(e).__name__, "-", e)
            raise
    finally:
        print("ปิดการเชื่อมต่อเครื่องพิมพ์แล้ว")


process_print_job(True, 5, 10)
print("---")
try:
    process_print_job(False, 5, 10)
except ConnectionError as e:
    print("งานพิมพ์ล้มเหลว:", e)
print("---")
try:
    process_print_job(True, 20, 10)
except ValueError as e:
    print("งานพิมพ์ล้มเหลว:", e)
```

คำอธิบายเพิ่มเติม (ขอบเขตนอกเหนือจากสไลด์): ตัวอย่างเฉลยนี้เป็นเพียงหนึ่งในหลายวิธีออกแบบที่เป็นไปได้ (pain point
ไม่มีคำตอบตายตัวเพียงแบบเดียว) ผสมผสานสามแนวคิดขั้นสูงเข้าด้วยกัน คือ การซ้อนบล็อกทดลองรัน/บล็อกดักข้อผิดพลาด/
ส่วน finally (หัวข้อที่ 13 ใน `lecture_note_advanced.md`), การโยนข้อผิดพลาดซ้ำด้วย `raise` เพียงคำเดียวหลังบันทึก
log (หัวข้อที่ 15 ในเอกสารเดียวกัน) และการชั่งน้ำหนักระหว่างการดักข้อผิดพลาดแบบระบุชนิดกับแบบไม่ระบุชนิด (bare
except หัวข้อที่ 14) ซึ่งตัวอย่างนี้เลือกใช้ `except Exception as e` (ระบุชนิดกว้างแต่ยังคงตรวจสอบชนิดที่แท้จริง
ผ่าน `type(e).__name__` ได้) แทนการใช้ `except:` แบบไม่ระบุชนิดเลย เพื่อยังคงรักษาความสามารถในการแยกแยะชนิด
ข้อผิดพลาดไว้ที่ชั้นนอกสุดผ่าน `except ConnectionError` และ `except ValueError` แยกกันตามชนิดที่แท้จริง ส่วน
`finally` ของบล็อกซ้อนชั้นนอกทำงานเสมอไม่ว่าขั้นตอนภายในจะสำเร็จหรือไม่ก็ตาม สอดคล้องกับพฤติกรรมที่ยืนยันด้วย
การรันจริงข้างต้น
