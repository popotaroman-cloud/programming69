## เฉลยแบบฝึกหัด

> เฉลยข้อทายผลลัพธ์ ข้อหาข้อผิดพลาด และข้อเขียนโค้ดทุกข้อในไฟล์นี้ถูกรันจริงผ่าน `nbclient` เพื่อยืนยันผลลัพธ์ (ไม่มีข้อใดที่เดา output) ข้อที่ตั้งใจให้เกิดข้อผิดพลาดที่ไม่ถูกดักจับ (uncaught exception/traceback) เป็นพฤติกรรมที่ถูกต้องตามโจทย์ ไม่ใช่ข้อผิดพลาดของสมุดบันทึกนี้

**เฉลยข้อ 1:** ข้อผิดพลาดทางไวยากรณ์ (syntax error) คือข้อผิดพลาดที่ตรวจพบได้จากการตรวจสอบโครงสร้างของโค้ดเพียงอย่างเดียว
ในขั้นตอนการแปลงโค้ดต้นฉบับ (parsing) ก่อนที่โปรแกรมจะเริ่มทำงานแม้แต่บรรทัดเดียว ตัวอย่างเช่น การเยื้องบรรทัด
ที่ไม่ถูกต้อง ส่วนข้อผิดพลาดขณะรัน (exception) คือข้อผิดพลาดที่เกิดขึ้นระหว่างที่โปรแกรมกำลังทำงานอยู่จริง เช่น
การหารด้วยศูนย์ หรือการเข้าถึงตำแหน่งดัชนีที่ไม่มีอยู่จริงในรายการ ข้อผิดพลาดประเภทนี้ไม่สามารถตรวจพบได้ล่วงหน้า
จากการอ่านโครงสร้างของโค้ดเพียงอย่างเดียว ต้องรอให้โปรแกรมทำงานมาถึงบรรทัดนั้นจริงจึงจะทราบได้

**เฉลยข้อ 6:**

````python
fee_text = "ยี่สิบบาท"
try:
    fee = int(fee_text)
    print("ค่าจอดรถคือ", fee, "บาท")
except ValueError:
    print("ข้อมูลค่าจอดรถไม่ถูกต้อง")
print("ระบบจอดรถทำงานต่อ")
````

**ผลลัพธ์ที่ยืนยันจากการรันจริง**
```

ระบบจอดรถทำงานต่อ
```
คำอธิบาย: `int("ยี่สิบบาท")` แปลงไม่ได้เพราะเป็นข้อความภาษาไทย เกิด `ValueError` ซึ่งถูกดักจับได้ตามที่ระบุไว้
ในบรรทัดที่ 5 โปรแกรมจึงทำงานต่อได้จนถึงบรรทัดสุดท้าย

**เฉลยข้อ 7:**

````python
distance_km = 10
try:
    fuel_needed = distance_km / "0"
    print(fuel_needed)
except ZeroDivisionError:
    print("ระยะทางเป็นศูนย์")
````

ชนิดข้อผิดพลาดที่เกิดขึ้นจริงคือ `TypeError` ไม่ใช่ `ZeroDivisionError` เพราะบรรทัดที่ 3 พยายามหารตัวเลข
(`10`) ด้วยข้อความ (`"0"` ซึ่งเป็นสายอักขระ ไม่ใช่ตัวเลข 0) ไพทอนไม่สามารถดำเนินการหารระหว่างชนิด `int` กับ
`str` ได้เลยไม่ว่าค่าของข้อความจะเป็นอะไรก็ตาม จึงเกิด `TypeError` ขึ้นก่อนที่จะพิจารณาว่าค่าที่หารเป็นศูนย์
หรือไม่ด้วยซ้ำ เนื่องจากบล็อก `except` ในบรรทัดที่ 5 ระบุไว้เฉพาะ `ZeroDivisionError` จึงไม่สามารถดักจับ
`TypeError` นี้ได้ โปรแกรมจึงหยุดทำงานพร้อม traceback ที่ยืนยันจากการรันจริงดังนี้
```
Traceback (most recent call last):
  File "exercise.py", line 3, in <module>
    fuel_needed = distance_km / "0"
                  ~~~~~~~~~~~~^~~~~
TypeError: unsupported operand type(s) for /: 'int' and 'str'
```

**เฉลยข้อ 8:**

````python
def average_water_usage(total_liters, days):
    try:
        return total_liters / days
    except ZeroDivisionError:
        return "ไม่สามารถคำนวณได้เนื่องจากจำนวนวันเป็นศูนย์"


print(average_water_usage(300, 3))
print(average_water_usage(300, 0))
````

**Output (รันจริงยืนยันแล้ว)**
```
100.0
ไม่สามารถคำนวณได้เนื่องจากจำนวนวันเป็นศูนย์
```

**เฉลยข้อ 9:** - `ZeroDivisionError` เกิดเมื่อมีการหารด้วยศูนย์
- `ValueError` เกิดเมื่อแปลงชนิดข้อมูลด้วยค่าที่ไม่สามารถแปลงได้ตามชนิดที่ต้องการ (ชนิดข้อมูลตั้งต้นถูกต้อง แต่ค่า
  ไม่เหมาะสม เช่น ข้อความที่ไม่ใช่ตัวเลข)
- `TypeError` เกิดเมื่อดำเนินการระหว่างชนิดข้อมูลที่ไม่สอดคล้องกันโดยตรง (ชนิดข้อมูลไม่ถูกต้องตั้งแต่แรกสำหรับ
  การดำเนินการนั้น)
- `IndexError` เกิดเมื่อเข้าถึงตำแหน่งดัชนีของรายการที่ไม่มีอยู่จริง

ความแตกต่างระหว่าง `ValueError` กับ `TypeError`: `ValueError` เกี่ยวข้องกับ "ค่า" ที่ไม่เหมาะสมแม้ชนิดข้อมูล
จะถูกต้องแล้ว (เช่น ข้อความ "abc" เป็น `str` ที่ถูกต้อง แต่แปลงเป็นตัวเลขไม่ได้) ในขณะที่ `TypeError` เกี่ยวข้อง
กับ "ชนิด" ของข้อมูลที่ไม่รองรับการดำเนินการนั้นตั้งแต่ต้น (เช่น การบวก `int` กับ `str` โดยตรง)

**เฉลยข้อ 13:** การเขียนบล็อก `except` หลายบล็อกช่วยให้ฟังก์ชันตอบสนองต่อสาเหตุของปัญหาแต่ละชนิดได้อย่างเจาะจง เช่น แจ้ง
ข้อความที่แตกต่างกันสำหรับกรณีหารด้วยศูนย์ กรณีข้อมูลแปลงชนิดไม่ได้ และกรณีดัชนีเกินขอบเขต ในขณะที่การใช้
บล็อก `except` เดียวรับมือกับทุกข้อผิดพลาดด้วยข้อความเดียวกันจะทำให้ผู้ใช้งานฟังก์ชัน (หรือผู้พัฒนาที่ตรวจสอบ
ปัญหา) ไม่ทราบว่าสาเหตุที่แท้จริงของปัญหาคืออะไร ทำให้ยากต่อการแก้ไขหรือปรับปรุงข้อมูลนำเข้าให้ถูกต้อง

**เฉลยข้อ 14:**

````python
def get_fuel_cost(price_texts, distance, index):
    try:
        prices = [float(text) for text in price_texts]
        total = sum(prices)
        cost_per_km = total / distance
        print("ต้นทุนต่อกิโลเมตรคือ", cost_per_km)
        return prices[index]
    except ZeroDivisionError:
        return "ระยะทางเป็นศูนย์"
    except ValueError:
        return "พบข้อความที่ไม่ใช่ตัวเลขในรายการราคาน้ำมัน"
    except IndexError:
        return "ไม่มีราคาลำดับนี้"


print(get_fuel_cost(["30", "32", "31"], 100, 0))
print(get_fuel_cost(["30", "32", "31"], 0, 0))
````

**ผลลัพธ์ที่ยืนยันจากการรันจริง**
```
ต้นทุนต่อกิโลเมตรคือ 0.93
30.0
ระยะทางเป็นศูนย์
```
คำอธิบาย: การเรียกครั้งแรกคำนวณสำเร็จทั้งหมด (รวมราคา 93 หารด้วยระยะทาง 100 กม. ได้ 0.93 แล้วคืนค่าที่ดัชนี 0
คือ 30.0) การเรียกครั้งที่สอง `distance` เป็น 0 ทำให้เกิด `ZeroDivisionError` ในบรรทัดที่ 5 จึงถูกดักจับและคืน
ข้อความแจ้งเตือนแทน

**เฉลยข้อ 15:**

````python
def get_temperature_report(temp_texts, day_index):
    try:
        temps = [float(t) for t in temp_texts]
        average = sum(temps) / len(temps)
        print("อุณหภูมิเฉลี่ยคือ", average)
        return temps[day_index]
    except ZeroDivisionError:
        return "ไม่มีข้อมูลอุณหภูมิ"
    except ValueError:
        return "พบข้อความที่ไม่ใช่ตัวเลขในรายการอุณหภูมิ"


print(get_temperature_report(["30.5", "31.0", "29.8"], 1))
print(get_temperature_report(["30.5", "31.0", "29.8"], 10))
````

การเรียกใช้ครั้งที่สอง (บรรทัดที่ 14) ส่งค่า `day_index` เป็น 10 ในขณะที่รายการ `temps` มีสมาชิกเพียง 3 ตัว
ทำให้ `temps[10]` ในบรรทัดที่ 6 เกิด `IndexError` แต่ฟังก์ชันนี้มีเพียงบล็อก `except ZeroDivisionError`
(บรรทัดที่ 7–8) และ `except ValueError` (บรรทัดที่ 9–10) เท่านั้น ไม่มีบล็อก `except IndexError` เลย
ข้อผิดพลาดจึงแพร่กระจายออกจากฟังก์ชันไปยังจุดที่เรียกใช้ ทำให้โปรแกรมหยุดทำงาน ผลลัพธ์ที่ยืนยันจากการรันจริง
คือ
```

31.0
อุณหภูมิเฉลี่ยคือ 30.433333333333334
Traceback (most recent call last):
  File "exercise.py", line 14, in <module>
    print(get_temperature_report(["30.5", "31.0", "29.8"], 10))
  File "exercise.py", line 6, in get_temperature_report
    return temps[day_index]
IndexError: list index out of range
```

**เฉลยข้อ 16:**

````python
def get_medicine_stock(stock_texts, index):
    try:
        stocks = [int(text) for text in stock_texts]
        return stocks[index]
    except ValueError:
        return "พบข้อความที่ไม่ใช่ตัวเลขในรายการสต็อกยา"
    except IndexError:
        return "ไม่มีรายการยาลำดับนี้"


print(get_medicine_stock(["100", "50", "20"], 1))
print(get_medicine_stock(["100", "ห้าสิบ", "20"], 1))
print(get_medicine_stock(["100", "50", "20"], 5))
````

**Output (รันจริงยืนยันแล้ว)**
```
50
พบข้อความที่ไม่ใช่ตัวเลขในรายการสต็อกยา
ไม่มีรายการยาลำดับนี้
```

**เฉลยข้อ 17:** ฟังก์ชันเชิงป้องกัน (defensive function) คือฟังก์ชันที่ออกแบบให้ใช้ `try`/`except` ห่อหุ้มการดำเนินการที่อาจ
ก่อให้เกิดข้อผิดพลาดไว้ภายในตัวฟังก์ชันเอง เพื่อให้ฟังก์ชันนั้น return ค่าออกมาเป็นค่าเดียวเสมอไม่ว่าจะเรียกใช้ด้วย
ข้อมูลนำเข้าแบบใดก็ตาม ไม่ปล่อยให้ข้อผิดพลาดแพร่กระจายออกไปทำให้โปรแกรมที่เรียกใช้หยุดทำงาน ความสัมพันธ์
กับ `safe_divide` คือ `safe_divide` เป็นตัวอย่างต้นแบบของฟังก์ชันเชิงป้องกันในสัปดาห์นี้ ซึ่งใช้ `try`/`except`
ดักจับ `ZeroDivisionError` แล้วคืนค่า 0 แทนการปล่อยให้โปรแกรมหยุดทำงาน

**เฉลยข้อ 18:**

````python
def safe_get_discount_percent(original_price, discounted_price):
    try:
        return (original_price - discounted_price) / original_price * 100
    except ZeroDivisionError:
        return 0


print(safe_get_discount_percent(500, 400))
print(safe_get_discount_percent(0, 0))
````

**ผลลัพธ์ที่ยืนยันจากการรันจริง**
```
20.0
0
```
คำอธิบาย: การเรียกครั้งแรกคำนวณ `(500-400)/500*100` ได้ 20.0 ส่วนการเรียกครั้งที่สอง `original_price`
เป็น 0 ทำให้เกิดการหารด้วยศูนย์ในบรรทัดที่ 3 ซึ่งถูกดักจับและคืนค่า 0 แทน

**เฉลยข้อ 19:**

````python
def safe_convert_weight(text):
    try:
        return float(text)
    except ValueError:
        return None


print(safe_convert_weight("65.5"))
print(safe_convert_weight("หกสิบห้า"))
print(safe_convert_weight(None))
````

การเรียกใช้ครั้งที่สาม (บรรทัดที่ 10) ส่งค่า `None` เข้าไปแทนที่จะเป็นข้อความ ทำให้ `float(None)` ในบรรทัดที่ 3
ก่อให้เกิด `TypeError` ไม่ใช่ `ValueError` เนื่องจากบล็อก `except` ในบรรทัดที่ 4 ระบุไว้เฉพาะ `ValueError`
เท่านั้น จึงไม่สามารถดักจับ `TypeError` นี้ได้ โปรแกรมจึงหยุดทำงาน แสดงให้เห็นว่าคำว่า "safe" ในชื่อฟังก์ชันสื่อถึง
ความปลอดภัยที่จำกัดอยู่เฉพาะชนิดข้อผิดพลาดที่ระบุไว้ในบล็อก `except` เท่านั้น ผลลัพธ์ที่ยืนยันจากการรันจริงคือ

**เฉลยข้อ 20:**

````python
def safe_get_item_price(price_list, index):
    try:
        return price_list[index]
    except IndexError:
        return None


print(safe_get_item_price([100, 200, 300], 1))
print(safe_get_item_price([100, 200, 300], 10))
````
