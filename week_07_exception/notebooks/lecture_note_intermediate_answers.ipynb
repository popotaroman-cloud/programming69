### เฉลยข้อ 4

**โค้ดตัวอย่าง**

````python
def check_ticket_number(text):
    try:
        ticket = int(text)
    except ValueError:
        print("หมายเลขตั๋วไม่ถูกต้อง:", text)
    else:
        print("รับตั๋วหมายเลข", ticket, "เรียบร้อยแล้ว")

check_ticket_number("45")
check_ticket_number("สี่สิบห้า")
````

**Output (รันจริงยืนยันแล้ว)**
```
รับตั๋วหมายเลข 45 เรียบร้อยแล้ว
หมายเลขตั๋วไม่ถูกต้อง: สี่สิบห้า
```

### เฉลยข้อ 6

ส่วน `else` ทำงานเฉพาะเมื่อบล็อกทดลองรันจบลงโดยไม่มีข้อผิดพลาดเกิดขึ้นเลยเท่านั้น หากเกิดข้อผิดพลาดขึ้น
ส่วน `else` จะไม่ทำงานเลย ในทางตรงข้าม ส่วน `finally` ทำงานเสมอในทุกกรณี ไม่ว่าบล็อกทดลองรันจะสำเร็จ เกิด
ข้อผิดพลาดที่ถูกดักจับได้ หรือแม้แต่เกิดข้อผิดพลาดที่ไม่มีผู้ใดดักจับเลยก็ตาม `finally` ไม่ได้ทำหน้าที่ดักจับหรือ
ระงับข้อผิดพลาดด้วยตัวเอง หากบล็อกทดลองรันเกิดข้อผิดพลาดที่ไม่มี except ใดดักจับได้ ข้อผิดพลาดนั้นจะยังคง
แพร่กระจายต่อไปตามปกติหลังจากส่วน finally ทำงานเสร็จแล้ว

### เฉลยข้อ 9

**โค้ดตัวอย่าง**

````python
def process_payment(amount):
    try:
        fee = 1000 / amount
        print("ค่าธรรมเนียมต่อหน่วย:", fee)
        return fee
    except ZeroDivisionError:
        print("จำนวนเงินต้องไม่เป็นศูนย์")
        return None
    finally:
        print("ปิดการเชื่อมต่อระบบชำระเงิน")

print(process_payment(200))
print(process_payment(0))
````

**Output (รันจริงยืนยันแล้ว)**
```
ค่าธรรมเนียมต่อหน่วย: 5.0
ปิดการเชื่อมต่อระบบชำระเงิน
5.0
จำนวนเงินต้องไม่เป็นศูนย์
ปิดการเชื่อมต่อระบบชำระเงิน
None
```

### เฉลยข้อ 14

**โค้ดตัวอย่าง**

````python
def parse_quantity(text):
    try:
        quantity = int(text)
        print("จำนวนที่แปลงได้:", quantity)
        return quantity
    except ValueError as e:
        print("ชนิดข้อผิดพลาด:", type(e).__name__)
        print("ข้อความข้อผิดพลาดจริง:", e)
        return None

parse_quantity("10")
parse_quantity("ten")
````

**Output (รันจริงยืนยันแล้ว)**
```
จำนวนที่แปลงได้: 10
ชนิดข้อผิดพลาด: ValueError
ข้อความข้อผิดพลาดจริง: invalid literal for int() with base 10: 'ten'
```

### เฉลยข้อ 15

**โค้ดตัวอย่าง**

````python
def lookup_price(price_table, product):
    try:
        price = price_table[product]
        print("ราคาของ", product, "คือ", price)
        return price
    except KeyError as e:
        print("ไม่พบสินค้าในตาราง")
        print("ข้อความจาก e.args:", e.args)
        return None

price_table = {"ปากกา": 15, "ยางลบ": 5}
lookup_price(price_table, "ปากกา")
lookup_price(price_table, "ไม้บรรทัด")
````

**Output (รันจริงยืนยันแล้ว)**
```
ราคาของ ปากกา คือ 15
ไม่พบสินค้าในตาราง
ข้อความจาก e.args: ('ไม้บรรทัด',)
```

ไม่สามารถสร้างข้อผิดพลาดให้เองโดยอัตโนมัติ เมื่อพบว่าเงื่อนไขไม่เป็นจริง ผู้เขียนโปรแกรมจึงโยนข้อผิดพลาดขึ้นเอง
ด้วย `raise` ข้อผิดพลาดที่เกิดจาก `raise` ถูกจัดการด้วยกลไก try/except เดียวกันทุกประการกับข้อผิดพลาดที่เกิด
ขึ้นเองตามธรรมชาติ เพราะ `raise` เพียงแต่สร้างอ็อบเจกต์ข้อผิดพลาดขึ้นมาแล้วส่งต่อควบคุมไปยังตัวจัดการข้อผิดพลาด
ที่ใกล้ที่สุดในสายการเรียกใช้งานเหมือนกันทุกประการ ไม่มีวากยสัมพันธ์พิเศษเพิ่มเติมสำหรับการดักจับข้อผิดพลาดที่
เกิดจาก `raise` โดยเฉพาะ

### เฉลยข้อ 19

**โค้ดตัวอย่าง**

````python
def set_thermostat(temp):
    if temp < 16 or temp > 30:
        raise ValueError("อุณหภูมิต้องอยู่ระหว่าง 16 ถึง 30 องศา")
    print("ตั้งอุณหภูมิเครื่องปรับอากาศเป็น:", temp)

try:
    set_thermostat(24)
    set_thermostat(40)
except ValueError as e:
    print("ตั้งค่าไม่สำเร็จ:", e)
````

**Output (รันจริงยืนยันแล้ว)**
```
ตั้งอุณหภูมิเครื่องปรับอากาศเป็น: 24
ตั้งค่าไม่สำเร็จ: อุณหภูมิต้องอยู่ระหว่าง 16 ถึง 30 องศา
```
