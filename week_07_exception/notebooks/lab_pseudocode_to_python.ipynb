**Pseudocode**:

```
FUNCTION redeem_free_drink(points_text)
    TRY
        points = CONVERT points_text TO INTEGER
    EXCEPT ValueError
        PRINT "ข้อมูลแต้มสะสมไม่ถูกต้อง"
    ELSE
        IF points >= 50 THEN
            PRINT "แลกเครื่องดื่มฟรีได้สำเร็จ"
        ELSE
            PRINT "แต้มสะสมไม่พอสำหรับแลกเครื่องดื่มฟรี"
        END IF
    END TRY
END FUNCTION

CALL redeem_free_drink WITH ("60")
CALL redeem_free_drink WITH ("30")
CALL redeem_free_drink WITH ("หกสิบ")
```

### ข้อที่ 1 (Pseudocode)

**โจทย์**: เว็บไซต์ประมูลสินค้าออนไลน์แห่งหนึ่งกำหนดว่าราคาที่ผู้ประมูลรายใหม่เสนอเข้ามาต้องสูงกว่าราคาประมูล
สูงสุดปัจจุบันเสมอ มิฉะนั้นถือว่าการประมูลครั้งนั้นไม่สำเร็จ

**Pseudocode**:

```
CLASS BidTooLowError INHERITS FROM Exception
END CLASS

FUNCTION place_bid(current_highest, new_bid)
    IF new_bid <= current_highest THEN
        RAISE BidTooLowError("ราคาประมูลต้องสูงกว่าราคาปัจจุบัน")
    END IF
    RETURN new_bid
END FUNCTION

TRY
    CALL place_bid WITH (500, 400)
EXCEPT BidTooLowError AS e
    PRINT "ประมูลไม่สำเร็จ:", e
END TRY

PRINT CALL place_bid WITH (500, 600)
```
