```python
def classify_battery_level(percent):
    if percent >= 80:
        return "แบตเตอรี่เต็ม"
    if percent >= 20:
        return "แบตเตอรี่ปกติ"
    return "แบตเตอรี่ต่ำ"

print(classify_battery_level(95))
print(classify_battery_level(50))
print(classify_battery_level(10))
```

#### เฉลยข้อที่ 4

```python
def calculate_parking_fee(hours):
    if hours <= 2:
        fee = 0
    else:
        fee = (hours - 2) * 20
    return fee

print(calculate_parking_fee(5))
print(calculate_parking_fee(1))
```

หมายเหตุ: ตัวอย่างเฉลยนี้เป็นเพียงหนึ่งในหลายวิธีออกแบบที่เป็นไปได้ (pain point ไม่มีคำตอบตายตัวเพียงแบบเดียว)
จอด 5 ชั่วโมงเสียค่าจอดเฉพาะ 3 ชั่วโมงที่เกินมา (5 - 2 = 3) คูณ 20 บาทได้ 60 บาท ส่วนจอด 1 ชั่วโมงยังอยู่ในช่วงฟรี
2 ชั่วโมงแรก จึงไม่เสียค่าใช้จ่ายเลย (0)

#### เฉลยข้อที่ 2

```python
def sum_up_to(n):
    if n == 1:
        return 1
    return n + sum_up_to(n - 1)

print(sum_up_to(5))
```

#### เฉลยข้อที่ 3

```python
def average_rating(*ratings):
    total = 0
    for r in ratings:
        total = total + r
    average = total / len(ratings)
    return average

print(average_rating(5, 4, 5))
print(average_rating(3, 4, 2, 5, 4))
```

#### เฉลยข้อที่ 4

```python
def add_song_to_playlist(playlist, song_name):
    playlist.append(song_name)
    return playlist

my_playlist = ["เพลงที่ 1"]
add_song_to_playlist(my_playlist, "เพลงที่ 2")
print(my_playlist)
```

หมายเหตุ: ตัวอย่างเฉลยนี้เป็นเพียงหนึ่งในหลายวิธีออกแบบที่เป็นไปได้ ฟังก์ชัน `add_song_to_playlist` ไม่จำเป็นต้องคืนค่า `playlist` กลับมาก็ได้
(เพราะ list เป็นชนิดข้อมูลที่เปลี่ยนแปลงค่าได้ หรือ mutable และ Python ส่งค่าตัวแปร list แบบอ้างอิงไปยังอ็อบเจกต์เดียวกัน
การเรียก `playlist.append(...)` ภายในฟังก์ชันจึงแก้ไข `my_playlist` ตัวเดิมนอกฟังก์ชันได้ทันทีโดยไม่ต้อง return ก็เห็นผลเช่นกัน)
บรรทัด 6 ในตัวอย่างนี้เลือก return ค่ากลับมาด้วยเพื่อให้ฟังก์ชันยังคง return ค่าเดียวเสมอตามธรรมเนียมของบทเรียนนี้
แต่ไม่ได้ใช้ค่าที่คืนมานั้นต่อ

### ระดับสูง

#### เฉลยข้อที่ 1

```python
employees = [
    {"name": "สมชาย", "years": 5},
    {"name": "สมหญิง", "years": 2},
    {"name": "วิชัย", "years": 8},
]

get_years = lambda employee: employee["years"]

sorted_employees = sorted(employees, key=get_years)

for employee in sorted_employees:
    print(employee["name"], employee["years"])
```

หมายเหตุ (ขอบเขตนอกเหนือจากสไลด์): `lambda` และการเรียงลำดับด้วย `key=` เป็นแนวคิดขั้นสูงกว่า `slides/06_function.md`
ที่สอนจริงในสัปดาห์ 6 มาก (ดูรายละเอียดใน `lecture_note_advanced.md` หัวข้อที่ 11) พนักงานแต่ละคนเก็บไว้ใน dictionary
(ไม่ใช่ tuple) เพื่อหลีกเลี่ยงการต้องใช้ tuple unpacking ซึ่งยังไม่ได้สอนในคอร์สนี้เช่นกัน และ `get_years` ยังคง return
ค่าเดียวเสมอตามกฎของสไลด์ 6 (ไม่ได้ขัดกับกฎ "return ค่าเดียวเสมอ")

#### เฉลยข้อที่ 2

```python
def apply_workout_intensity(duration, intensity_function):
    return intensity_function(duration)

def light_workout(duration):
    return duration * 5

def intense_workout(duration):
    return duration * 12

print(apply_workout_intensity(30, light_workout))
print(apply_workout_intensity(30, intense_workout))
```

หมายเหตุ (ขอบเขตนอกเหนือจากสไลด์): การส่งฟังก์ชัน (`light_workout`, `intense_workout`) เป็นอาร์กิวเมนต์ให้อีกฟังก์ชันหนึ่ง
เป็นแนวคิดฟังก์ชันชั้นหนึ่ง (first-class function) และฟังก์ชันอันดับสูง (higher-order function) ซึ่งไม่ได้สอนใน
`slides/06_function.md` — ดูรายละเอียดใน `lecture_note_advanced.md` หัวข้อที่ 10 และ 12 ทุกฟังก์ชันในตัวอย่างนี้ยัง
return ค่าเดียวเสมอ ไม่ขัดกับกฎของสไลด์ 6

#### เฉลยข้อที่ 3

```python
def calculate_weighted_average(courses):
    total_weighted = 0
    total_credit = 0
    for course in courses:
        total_weighted = total_weighted + (course["score"] * course["credit"])
        total_credit = total_credit + course["credit"]
    average = total_weighted / total_credit
    return average

courses = [
    {"score": 3.5, "credit": 3},
    {"score": 4.0, "credit": 2},
    {"score": 3.0, "credit": 1},
]
result = calculate_weighted_average(courses)
print(result)
```

หมายเหตุ (ขอบเขตนอกเหนือจากสไลด์): แนวคิดฟังก์ชันบริสุทธิ์ (pure function) และผลข้างเคียง (side effect) ไม่มีอยู่ใน
`slides/06_function.md` — ดูรายละเอียดใน `lecture_note_advanced.md` หัวข้อที่ 13 สังเกตว่า `calculate_weighted_average`
ไม่มีคำสั่ง `print` อยู่ภายในฟังก์ชันเลย และไม่ได้แก้ไขลิสต์ `courses` ที่รับเข้ามา (ไม่มีผลข้างเคียง) การพิมพ์ผลลัพธ์
ทำที่บรรทัด 16 ซึ่งอยู่นอกฟังก์ชัน สอดคล้องกับนิยามของฟังก์ชันบริสุทธิ์ และฟังก์ชันนี้ยัง return ค่าเดียวเสมอตามกฎของสไลด์ 6

#### เฉลยข้อที่ 4

```python
def validate_num_correct(num_correct, total_questions):
    return 0 <= num_correct <= total_questions

def calculate_percentage(num_correct, total_questions):
    return (num_correct / total_questions) * 100

def display_quiz_result(student_name, percentage):
    print(student_name + ": ทำได้ " + str(percentage) + "%")

def process_quiz_result(student_name, num_correct, total_questions):
    if not validate_num_correct(num_correct, total_questions):
        print("ข้อมูลไม่ถูกต้อง: ตอบถูก " + str(num_correct) + " จาก " + str(total_questions) + " ข้อ")
        return None
    percentage = calculate_percentage(num_correct, total_questions)
    display_quiz_result(student_name, percentage)
    return percentage

process_quiz_result("มานี", 8, 10)
process_quiz_result("ชูใจ", 15, 10)
```

หมายเหตุ (ขอบเขตนอกเหนือจากสไลด์): หลักความรับผิดชอบเดียว (Single Responsibility Principle) และแนวคิดลายเซ็นฟังก์ชัน/สัญญา
(function signature/contract) ไม่มีอยู่ใน `slides/06_function.md` — ดูรายละเอียดใน `lecture_note_advanced.md` หัวข้อที่ 14
และ 15 ตัวอย่างเฉลยนี้เป็นเพียงหนึ่งในหลายวิธีออกแบบที่เป็นไปได้ (pain point ไม่มีคำตอบตายตัวเพียงแบบเดียว) โดยแยกออกเป็น
4 ฟังก์ชันย่อยที่แต่ละฟังก์ชันรับผิดชอบหน้าที่เดียว (ตรวจสอบความถูกต้อง / คำนวณเปอร์เซ็นต์ / แสดงผล) แล้วให้ `process_quiz_result`
ทำหน้าที่เป็นฟังก์ชันประสาน (orchestrator) เรียกใช้ฟังก์ชันย่อยเหล่านั้นตามลำดับ ทุกฟังก์ชันในตัวอย่างนี้ยัง return
ค่าเดียวเสมอ (หรือไม่ return เลยในกรณีของ `display_quiz_result`) ไม่ขัดกับกฎของสไลด์ 6
