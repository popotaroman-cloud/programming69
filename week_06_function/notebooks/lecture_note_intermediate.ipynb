## 6: By Value vs By Reference (กลไกการส่งค่าให้ฟังก์ชันของ Python)

### แนวคิด

เมื่อเรียกฟังก์ชันโดยส่งอาร์กิวเมนต์เข้าไป ภาษาโปรแกรมแต่ละภาษามีกลไกการส่งค่าที่แตกต่างกัน โดยทั่วไปแบ่งเป็นสองแบบหลัก
คือ **การส่งค่าแบบสำเนา (pass by value)** ซึ่งฟังก์ชันจะได้รับสำเนาของข้อมูล การแก้ไขค่าภายในฟังก์ชันจะไม่กระทบต้นฉบับ
และ **การส่งค่าแบบอ้างอิง (pass by reference)** ซึ่งฟังก์ชันจะได้รับที่อยู่อ้างอิงของข้อมูลต้นฉบับโดยตรง การแก้ไขค่าภายใน
ฟังก์ชันจะกระทบต้นฉบับด้วย

Python ไม่ได้ใช้กลไกทั้งสองแบบนี้ตรง ๆ แต่ใช้แบบที่เรียกว่า **การส่งผ่านการอ้างอิงอ็อบเจกต์ (pass-by-object-reference)**
กล่าวคือ พารามิเตอร์ในฟังก์ชันจะได้รับ "การอ้างอิง" ไปยังอ็อบเจกต์เดียวกับที่ตัวแปรต้นทางชี้อยู่ (ไม่ใช่สำเนาข้อมูลทั้งก้อน)
แต่พฤติกรรมที่ผู้เขียนโปรแกรมสังเกตเห็นจะขึ้นอยู่กับว่าอ็อบเจกต์นั้น **เปลี่ยนแปลงค่าได้ (mutable)** หรือ **เปลี่ยนแปลงค่าไม่ได้
(immutable)**: ถ้าเป็นชนิดข้อมูลที่ไม่สามารถแก้ไขค่าได้ (เช่น `int`, `str`, `tuple`) การกำหนดค่าใหม่ให้พารามิเตอร์ภายใน
ฟังก์ชันจะสร้างอ็อบเจกต์ใหม่ขึ้นมา ไม่กระทบตัวแปรต้นทาง พฤติกรรมจึงดูคล้าย pass by value แต่ถ้าเป็นชนิดข้อมูลที่แก้ไขค่าได้
(เช่น `list`, `dict`, `set`) การเรียกเมธอดที่แก้ไขอ็อบเจกต์แบบ in-place (เช่น `append`) จะกระทบตัวแปรต้นทางด้วย เพราะทั้ง
พารามิเตอร์และตัวแปรต้นทางชี้ไปยังอ็อบเจกต์เดียวกันในหน่วยความจำ

### ตัวอย่าง Python

#### ตัวอย่างที่ 1 (หลัก): ส่งค่าชนิด immutable (int)

````python
def change_value(x):
    x = x + 10
    print("ในฟังก์ชัน:", x)

num = 5
change_value(num)
print("นอกฟังก์ชัน:", num)
````

**Output:**
```
ในฟังก์ชัน: 15
นอกฟังก์ชัน: 5
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | หมายเหตุ |
|---|---|---|---|
| 5 | `num = 5` | num อ้างอิงอ็อบเจกต์ int 5 | สร้างตัวแปรใน scope หลัก |
| 6 | เรียก `change_value(num)` | x อ้างอิงอ็อบเจกต์เดียวกับ num (5) | เข้าสู่ฟังก์ชัน พารามิเตอร์ x และ num ชี้ไปยังอ็อบเจกต์เดียวกันชั่วคราว |
| 2 | `x = x + 10` | x อ้างอิงอ็อบเจกต์ใหม่ 15 | int เป็น immutable คำสั่งนี้สร้างอ็อบเจกต์ใหม่แล้วให้ x ชี้ไปที่อ็อบเจกต์ใหม่ ไม่กระทบ num |
| 3 | `print(...)` | แสดง `ในฟังก์ชัน: 15` | พิมพ์ค่า x ภายในฟังก์ชันหลังการกำหนดค่าใหม่ |
| 7 | `print(...)` | แสดง `นอกฟังก์ชัน: 5` | num ยังคงอ้างอิงอ็อบเจกต์เดิม 5 เพราะไม่เคยถูกแก้ไข |

**คำอธิบายภาพรวม:** บรรทัด 1–3 นิยามฟังก์ชัน `change_value` ที่รับพารามิเตอร์ `x` แล้วพยายามเปลี่ยนค่า บรรทัด 5–6
สร้างตัวแปร `num` แล้วเรียกฟังก์ชันโดยส่ง `num` เข้าไป เนื่องจาก `int` เป็นชนิดข้อมูล immutable การกำหนดค่าใหม่ที่บรรทัด 2
จึงเป็นการสร้างอ็อบเจกต์ใหม่ให้ `x` ชี้ไป แทนที่จะแก้ไขอ็อบเจกต์เดิม ผลคือค่าของ `num` ภายนอกฟังก์ชัน (บรรทัด 7) ไม่เปลี่ยนแปลง
แม้ภายในฟังก์ชันจะแสดงค่าที่เปลี่ยนไปแล้วก็ตาม (บรรทัด 3)

#### ตัวอย่างที่ 2 (edge case): ส่งค่าชนิด mutable (list)

````python
def add_item(items):
    items.append("B")
    print("ในฟังก์ชัน:", items)

my_list = ["A"]
add_item(my_list)
print("นอกฟังก์ชัน:", my_list)
````

**Output:**
```
ในฟังก์ชัน: ['A', 'B']
นอกฟังก์ชัน: ['A', 'B']
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | หมายเหตุ |
|---|---|---|---|
| 5 | `my_list = ["A"]` | my_list อ้างอิง list `["A"]` | สร้าง list ซึ่งเป็นชนิด mutable |
| 6 | เรียก `add_item(my_list)` | items อ้างอิงอ็อบเจกต์เดียวกับ my_list | ทั้ง items และ my_list ชี้ไปยัง list ก้อนเดียวกันในหน่วยความจำ |
| 2 | `items.append("B")` | items อ้างอิง `["A", "B"]` | แก้ไข list เดิมแบบ in-place ไม่ได้สร้าง list ใหม่ |
| 3 | `print(...)` | แสดง `ในฟังก์ชัน: ['A', 'B']` | พิมพ์ค่าภายในฟังก์ชันหลังแก้ไข |
| 7 | `print(...)` | แสดง `นอกฟังก์ชัน: ['A', 'B']` | my_list เปลี่ยนตามด้วย เพราะชี้ไปยังอ็อบเจกต์เดียวกับ items |

**คำอธิบายภาพรวม:** บรรทัด 1–3 นิยามฟังก์ชัน `add_item` ที่รับพารามิเตอร์ `items` แล้วเรียกเมธอด `append` ซึ่งแก้ไข
อ็อบเจกต์แบบ in-place บรรทัด 5–6 สร้าง `my_list` แล้วส่งเข้าไปในฟังก์ชัน เนื่องจาก `list` เป็นชนิดข้อมูล mutable และ
`append` ไม่ได้สร้างอ็อบเจกต์ใหม่แต่แก้ไขอ็อบเจกต์เดิม ผลคือ `my_list` ภายนอกฟังก์ชัน (บรรทัด 7) เปลี่ยนแปลงตามไปด้วย
ต่างจากตัวอย่างที่ 1 อย่างชัดเจน แม้กลไกเบื้องหลังจะเป็น "การส่งผ่านการอ้างอิงอ็อบเจกต์" แบบเดียวกันทั้งคู่

#### ตัวอย่างที่ 3 (โดเมนต่างออกไป): ส่งค่าชนิดข้อความ (str) — อีกกรณีของ immutable

````python
def shout(message):
    message = message.upper()
    print("ในฟังก์ชัน:", message)

greeting_text = "hello world"
shout(greeting_text)
print("นอกฟังก์ชัน:", greeting_text)
````

**Output:**
```
ในฟังก์ชัน: HELLO WORLD
นอกฟังก์ชัน: hello world
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | หมายเหตุ |
|---|---|---|---|
| 5 | `greeting_text = "hello world"` | greeting_text อ้างอิง str `"hello world"` | สร้างตัวแปรชนิดข้อความ |
| 6 | เรียก `shout(greeting_text)` | message อ้างอิงอ็อบเจกต์เดียวกับ greeting_text | เข้าสู่ฟังก์ชัน message และ greeting_text ชี้ไปยัง str เดิมชั่วคราว |
| 2 | `message = message.upper()` | message อ้างอิงอ็อบเจกต์ใหม่ `"HELLO WORLD"` | `str` เป็น immutable เมธอด `.upper()` ไม่ได้แก้ไข str เดิม แต่คืนค่า str ใหม่ แล้วบรรทัดนี้ให้ message ชี้ไปที่อ็อบเจกต์ใหม่แทน |
| 3 | `print(...)` | แสดง `ในฟังก์ชัน: HELLO WORLD` | พิมพ์ค่า message ภายในฟังก์ชันหลังกำหนดค่าใหม่ |
| 7 | `print(...)` | แสดง `นอกฟังก์ชัน: hello world` | greeting_text ยังคงอ้างอิง str เดิม เพราะไม่เคยถูกแก้ไข |

**คำอธิบายภาพรวม:** บรรทัด 1–3 นิยามฟังก์ชัน `shout` ที่รับพารามิเตอร์ `message` แล้วพยายามแปลงเป็นตัวพิมพ์ใหญ่ บรรทัด 5–6
สร้างตัวแปร `greeting_text` ชนิดข้อความแล้วเรียกฟังก์ชันโดยส่งเข้าไป เนื่องจาก `str` เป็นชนิดข้อมูล immutable เช่นเดียวกับ
`int` ในตัวอย่างที่ 1 เมธอด `.upper()` ที่บรรทัด 2 จึงไม่ได้แก้ไขข้อความเดิม แต่สร้าง str ใหม่ขึ้นมาแล้วให้ `message` ชี้ไปที่
อ็อบเจกต์ใหม่นั้นแทน ผลคือ `greeting_text` ภายนอกฟังก์ชัน (บรรทัด 7) ยังคงเป็นตัวพิมพ์เล็กเหมือนเดิม ตัวอย่างนี้แสดงให้เห็นว่า
พฤติกรรม "คล้าย pass by value" ของ Python ไม่ได้จำกัดเฉพาะชนิดตัวเลข แต่ครอบคลุมชนิดข้อมูล immutable ทุกชนิด รวมถึง
ข้อความ ซึ่งเป็นโดเมนที่ต่างจากตัวเลข (ตัวอย่างที่ 1) และรายการ (ตัวอย่างที่ 2) อย่างชัดเจน

object-reference model)

## 7: อาร์กิวเมนต์ค่าเริ่มต้น (Default Parameters)

### แนวคิด

**อาร์กิวเมนต์ค่าเริ่มต้น (default argument)** คือค่าที่กำหนดไว้ล่วงหน้าให้กับพารามิเตอร์ในตอนนิยามฟังก์ชัน
เพื่อให้ฟังก์ชันสามารถถูกเรียกโดยไม่ต้องระบุอาร์กิวเมนต์สำหรับพารามิเตอร์นั้นก็ได้ หากผู้เรียกไม่ส่งค่ามา ฟังก์ชันจะใช้
ค่าเริ่มต้นที่กำหนดไว้โดยอัตโนมัติ แต่ถ้าผู้เรียกส่งค่ามา ค่านั้นจะเข้ามาแทนที่ค่าเริ่มต้น แนวคิดนี้ช่วยให้ฟังก์ชันมีความยืดหยุ่น
มากขึ้น เพราะพารามิเตอร์บางตัวกลายเป็น "ทางเลือก" (optional) แทนที่จะบังคับให้ผู้เรียกต้องระบุครบทุกตัวเสมอ

กฎสำคัญของอาร์กิวเมนต์ค่าเริ่มต้นในภาษาที่รองรับ (รวมถึง Python) คือพารามิเตอร์ที่ไม่มีค่าเริ่มต้นต้องมาก่อนพารามิเตอร์ที่มี
ค่าเริ่มต้นเสมอในการนิยามฟังก์ชัน นอกจากนี้ยังมีข้อควรระวังเฉพาะของ Python คือ **ค่าเริ่มต้นจะถูกสร้างขึ้นเพียงครั้งเดียว
ตอนที่ประมวลผลคำสั่ง `def`** ไม่ใช่ทุกครั้งที่ฟังก์ชันถูกเรียก ซึ่งหากค่าเริ่มต้นเป็นชนิดข้อมูล mutable (เช่น `list` หรือ
`dict`) อาจก่อให้เกิดพฤติกรรมที่ไม่คาดคิดได้ ดังแสดงในตัวอย่างที่ 2

### ตัวอย่าง Python

#### ตัวอย่างที่ 1 (หลัก): พารามิเตอร์พร้อมค่าเริ่มต้น

````python
def greet(name, greeting="สวัสดี"):
    print(greeting + " " + name)

greet("อาทิตย์")
greet("จันทร์", "หวัดดี")
````

**Output:**
```
สวัสดี อาทิตย์
หวัดดี จันทร์
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | หมายเหตุ |
|---|---|---|---|
| 4 | เรียก `greet("อาทิตย์")` | name="อาทิตย์", greeting="สวัสดี" (ค่าเริ่มต้น) | ไม่ได้ส่งอาร์กิวเมนต์ตัวที่สอง จึงใช้ค่าเริ่มต้น |
| 2 | `print(...)` | แสดง `สวัสดี อาทิตย์` | |
| 5 | เรียก `greet("จันทร์", "หวัดดี")` | name="จันทร์", greeting="หวัดดี" | ส่งอาร์กิวเมนต์ตามตำแหน่งครบสองตัว ค่าที่ส่งมาแทนที่ค่าเริ่มต้น |
| 2 | `print(...)` | แสดง `หวัดดี จันทร์` | |

**คำอธิบายภาพรวม:** บรรทัด 1–2 นิยามฟังก์ชัน `greet` ที่มีพารามิเตอร์ `greeting` กำหนดค่าเริ่มต้นเป็น `"สวัสดี"`
บรรทัด 4 เรียกฟังก์ชันโดยส่งอาร์กิวเมนต์เพียงตัวเดียว ทำให้ `greeting` ใช้ค่าเริ่มต้น ส่วนบรรทัด 5 ส่งอาร์กิวเมนต์ตามตำแหน่ง
ครบทั้งสองตัว ค่าที่ส่งมาจึงทับค่าเริ่มต้น แสดงให้เห็นว่าอาร์กิวเมนต์ค่าเริ่มต้นทำให้พารามิเตอร์นั้นกลายเป็นทางเลือกโดยไม่ทำให้
ฟังก์ชันเสียความยืดหยุ่นในการรับค่าที่แตกต่างกัน

#### ตัวอย่างที่ 2 (edge case): ค่าเริ่มต้นแบบ mutable ที่ก่อปัญหา และวิธีแก้

````python
def add_item_bad(item, basket=[]):
    basket.append(item)
    return basket

result1 = add_item_bad("apple")
result2 = add_item_bad("banana")
print("ครั้งที่ 1:", result1)
print("ครั้งที่ 2:", result2)

def add_item_good(item, basket=None):
    if basket is None:
        basket = []
    basket.append(item)
    return basket

result3 = add_item_good("apple")
result4 = add_item_good("banana")
print("แก้ไขแล้ว ครั้งที่ 1:", result3)
print("แก้ไขแล้ว ครั้งที่ 2:", result4)
````

**Output:**
```
ครั้งที่ 1: ['apple', 'banana']
ครั้งที่ 2: ['apple', 'banana']
แก้ไขแล้ว ครั้งที่ 1: ['apple']
แก้ไขแล้ว ครั้งที่ 2: ['banana']
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | หมายเหตุ |
|---|---|---|---|
| 1 | นิยาม `add_item_bad` | ค่าเริ่มต้น `basket=[]` ถูกสร้างเพียงครั้งเดียวตอนนิยามฟังก์ชัน | list ว่างนี้ผูกอยู่กับตัวฟังก์ชันเอง ไม่ได้สร้างใหม่ทุกครั้งที่เรียก |
| 5 | เรียก `add_item_bad("apple")` | item="apple", basket=[] (อ็อบเจกต์เดิมที่ผูกกับฟังก์ชัน) | ไม่ได้ส่ง basket มา จึงใช้ list เริ่มต้นร่วมกัน |
| 2 | `basket.append(item)` | basket=["apple"] | แก้ไข list เดิมแบบถาวร |
| 6 | เรียก `add_item_bad("banana")` | item="banana", basket=["apple"] (อ็อบเจกต์เดิมจากการเรียกครั้งก่อน) | basket ไม่ถูกสร้างใหม่ จึงพก "apple" ค้างมาจากการเรียกก่อนหน้า |
| 2 | `basket.append(item)` | basket=["apple", "banana"] | |
| 7–8 | `print(...)` | แสดง `['apple', 'banana']` ทั้งสองครั้ง | result1 และ result2 อ้างอิง list เดียวกัน — นี่คือปัญหา |
| 16 | เรียก `add_item_good("apple")` | item="apple", basket=None → สร้าง list ใหม่ที่บรรทัด 12 | ตรวจพบ None จึงสร้าง list ว่างใหม่ทุกครั้งที่เรียก |
| 13 | `basket.append(item)` | basket=["apple"] | |
| 17 | เรียก `add_item_good("banana")` | item="banana", basket=None → สร้าง list ใหม่อีกชุดหนึ่ง | list ชุดใหม่ ไม่เกี่ยวข้องกับการเรียกก่อนหน้า |
| 13 | `basket.append(item)` | basket=["banana"] | |
| 18–19 | `print(...)` | แสดง `['apple']` และ `['banana']` แยกกัน | ยืนยันว่าแก้ปัญหาสำเร็จ |

**คำอธิบายภาพรวม:** บรรทัด 1–8 สาธิตข้อผิดพลาดที่พบบ่อยเมื่อใช้ค่าเริ่มต้นเป็น mutable object (`basket=[]`)
เนื่องจาก Python สร้างอ็อบเจกต์ค่าเริ่มต้นเพียงครั้งเดียวตอนประมวลผล `def` ทุกการเรียกที่ไม่ได้ส่ง `basket` มาเองจึงใช้
list ก้อนเดียวกันร่วมกัน ทำให้ข้อมูลจากการเรียกครั้งก่อนตกค้างอยู่ในการเรียกครั้งถัดไป บรรทัด 10–19 แสดงวิธีแก้ไขมาตรฐาน
คือกำหนดค่าเริ่มต้นเป็น `None` แล้วสร้าง list ว่างใหม่ภายในฟังก์ชันเมื่อตรวจพบว่าไม่มีการส่งค่ามา (บรรทัด 11–12)
ทำให้แต่ละการเรียกได้ list อิสระของตัวเอง

#### ตัวอย่างที่ 3 (โดเมนต่างออกไป): คำนวณราคารวมพร้อมภาษี

````python
def calculate_total(price, tax_rate=0.07):
    total = price + (price * tax_rate)
    return total

print(calculate_total(100))
print(calculate_total(100, 0.10))
print(calculate_total(200, tax_rate=0))
````

**Output:**
```
107.0
110.0
200
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | หมายเหตุ |
|---|---|---|---|
| 5 | เรียก `calculate_total(100)` | price=100, tax_rate=0.07 (ค่าเริ่มต้น) | ไม่ได้ส่งอาร์กิวเมนต์ตัวที่สอง จึงใช้ค่าเริ่มต้น |
| 2 | `total = price + (price * tax_rate)` | total = 100 + 7.0 = 107.0 | ผลลัพธ์เป็น float เพราะ tax_rate เป็น float |
| 3 | `return total` | คืนค่า 107.0 | |
| 5 | `print(...)` | แสดง 107.0 | |
| 6 | เรียก `calculate_total(100, 0.10)` | price=100, tax_rate=0.10 | ส่ง tax_rate ตามตำแหน่ง ค่าที่ส่งมาทับค่าเริ่มต้น |
| 2 | `total = ...` | total = 100 + 10.0 = 110.0 | |
| 6 | `print(...)` | แสดง 110.0 | |
| 7 | เรียก `calculate_total(200, tax_rate=0)` | price=200, tax_rate=0 (ระบุแบบ keyword) | ส่ง tax_rate=0 แบบระบุชื่อ ค่าที่ส่งมาทับค่าเริ่มต้นเช่นกัน |
| 2 | `total = ...` | total = 200 + 0 = 200 | ไม่มีการบวกกับ float จึงคงชนิด int ไว้ตามอาร์กิวเมนต์ที่ส่งเข้ามาจริง |
| 7 | `print(...)` | แสดง 200 | |

**คำอธิบายภาพรวม:** บรรทัด 1–3 นิยามฟังก์ชัน `calculate_total` ที่มีพารามิเตอร์ `tax_rate` กำหนดค่าเริ่มต้นเป็น 0.07
(อัตราภาษี 7%) บรรทัด 5 เรียกฟังก์ชันโดยไม่ระบุ tax_rate จึงใช้ค่าเริ่มต้น ส่วนบรรทัด 6–7 ส่ง tax_rate เข้าไปเอง ทั้งแบบ
ตามตำแหน่งและแบบระบุชื่อ (keyword) ตัวอย่างนี้แสดงให้เห็นว่าอาร์กิวเมนต์ค่าเริ่มต้นมีประโยชน์ในบริบทการคำนวณตัวเลขเชิงธุรกิจ
(เช่น ภาษี ส่วนลด ค่าธรรมเนียม) ซึ่งเป็นโดเมนที่ต่างจากข้อความทักทาย (ตัวอย่างที่ 1) และตะกร้าสินค้า (ตัวอย่างที่ 2)
นอกจากนี้ยังสังเกตได้ว่าผลลัพธ์ที่บรรทัด 7 เป็น int ไม่ใช่ float เพราะ Python รักษาชนิดข้อมูลตามอาร์กิวเมนต์ที่ส่งเข้ามาจริง
ไม่ได้ยึดตามชนิดของค่าเริ่มต้นเสมอไป

## 8: อาร์กิวเมนต์จำนวนไม่จำกัด (*args / **kwargs)

### แนวคิด

บางครั้งผู้ออกแบบฟังก์ชันไม่ทราบล่วงหน้าว่าผู้เรียกจะส่งอาร์กิวเมนต์เข้ามากี่ตัว แนวคิด **อาร์กิวเมนต์จำนวนไม่จำกัด
(arbitrary arguments)** ช่วยให้ฟังก์ชันรับอาร์กิวเมนต์ได้ตามจำนวนที่ผู้เรียกต้องการ โดยแบ่งเป็นสองรูปแบบตามชนิดของ
อาร์กิวเมนต์: อาร์กิวเมนต์ตามตำแหน่งจำนวนไม่จำกัด ซึ่งฟังก์ชันจะรวบรวมค่าที่ส่งมาไว้เป็นกลุ่มเดียว (ใน Python คือ tuple
ผ่านสัญลักษณ์ `*`) และอาร์กิวเมนต์แบบระบุชื่อจำนวนไม่จำกัด ซึ่งฟังก์ชันจะรวบรวมคู่ชื่อ-ค่าไว้เป็นกลุ่มเดียว (ใน Python
คือ dict ผ่านสัญลักษณ์ `**`) ในภาษา Python ธรรมเนียมการตั้งชื่อพารามิเตอร์เหล่านี้คือ `*args` และ `**kwargs`
แต่ในทางเทคนิคสามารถตั้งชื่ออื่นได้ ตัวสัญลักษณ์ `*` และ `**` ต่างหากที่กำหนดพฤติกรรม ไม่ใช่ชื่อพารามิเตอร์

เมื่อฟังก์ชันหนึ่งต้องการรับทั้งพารามิเตอร์ปกติ อาร์กิวเมนต์ตามตำแหน่งจำนวนไม่จำกัด และอาร์กิวเมนต์แบบระบุชื่อจำนวนไม่จำกัด
พร้อมกัน จะต้องเรียงลำดับในการนิยามฟังก์ชันตามกฎ: พารามิเตอร์ปกติ (positional/default) มาก่อน ตามด้วย `*args`
แล้วจึงตามด้วย `**kwargs` เป็นลำดับสุดท้ายเสมอ

### ตัวอย่าง Python

#### ตัวอย่างที่ 1 (หลัก): `*args` รวบรวมอาร์กิวเมนต์ตามตำแหน่งจำนวนไม่จำกัด

````python
def total(*numbers):
    result = 0
    for n in numbers:
        result = result + n
    return result

print(total(3, 5))
print(total(1, 2, 3, 4))
````

**Output:**
```
8
10
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | หมายเหตุ |
|---|---|---|---|
| 7 | เรียก `total(3, 5)` | numbers=(3, 5) | อาร์กิวเมนต์ทั้งหมดถูกรวบเป็น tuple ผ่าน `*numbers` |
| 2 | `result = 0` | result=0 | |
| 3–4 | ลูป `for n in numbers` | n=3 → result=3; n=5 → result=8 | สะสมผลรวมทีละตัว |
| 5 | `return result` | คืนค่า 8 | |
| 7 | `print(...)` | แสดง 8 | |
| 8 | เรียก `total(1, 2, 3, 4)` | numbers=(1, 2, 3, 4) | เรียกด้วยจำนวนอาร์กิวเมนต์ที่ต่างจากครั้งแรก |
| 3–4 | ลูป `for n in numbers` | n=1→1, n=2→3, n=3→6, n=4→10 | |
| 5 | `return result` | คืนค่า 10 | |
| 8 | `print(...)` | แสดง 10 | |

**คำอธิบายภาพรวม:** บรรทัด 1–5 นิยามฟังก์ชัน `total` ที่ใช้ `*numbers` รับอาร์กิวเมนต์ตามตำแหน่งได้ไม่จำกัดจำนวน
แล้ววนลูปสะสมผลรวมด้วยตัวแปร `result` บรรทัด 7–8 แสดงให้เห็นว่าฟังก์ชันเดียวกันสามารถเรียกด้วยจำนวนอาร์กิวเมนต์ที่ต่างกัน
ได้โดยไม่ต้องนิยามฟังก์ชันหลายเวอร์ชัน ซึ่งเป็นข้อดีหลักของอาร์กิวเมนต์จำนวนไม่จำกัดเมื่อเทียบกับพารามิเตอร์ตายตัว

#### ตัวอย่างที่ 2 (edge case): ผสม `*args` กับ `**kwargs` ในฟังก์ชันเดียว

````python
def student_report(name, *subjects, **details):
    print("ชื่อ:", name)
    print("วิชาที่ลง:", subjects)
    print("ข้อมูลเพิ่มเติม:", details)

student_report("แพร", "คณิตศาสตร์", "วิทยาศาสตร์", year=2, gpa=3.5)
````

**Output:**
```
ชื่อ: แพร
วิชาที่ลง: ('คณิตศาสตร์', 'วิทยาศาสตร์')
ข้อมูลเพิ่มเติม: {'year': 2, 'gpa': 3.5}
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | หมายเหตุ |
|---|---|---|---|
| 6 | เรียก `student_report("แพร", "คณิตศาสตร์", "วิทยาศาสตร์", year=2, gpa=3.5)` | name="แพร", subjects=("คณิตศาสตร์", "วิทยาศาสตร์"), details={"year": 2, "gpa": 3.5} | อาร์กิวเมนต์ตามตำแหน่งที่เกินจาก `name` ถูกรวบเป็น tuple ใน subjects ส่วนอาร์กิวเมนต์แบบระบุชื่อถูกรวบเป็น dict ใน details |
| 2 | `print(...)` | แสดง `ชื่อ: แพร` | |
| 3 | `print(...)` | แสดง subjects เป็น tuple | |
| 4 | `print(...)` | แสดง details เป็น dict | |

**คำอธิบายภาพรวม:** บรรทัด 1 นิยามฟังก์ชัน `student_report` ที่มีพารามิเตอร์ปกติ (`name`) ตามด้วย `*subjects`
และ `**details` ตามลำดับที่ Python กำหนดไว้ บรรทัด 6 เรียกฟังก์ชันโดยส่งอาร์กิวเมนต์ตามตำแหน่งสองตัวเพิ่มเติมและ
อาร์กิวเมนต์แบบระบุชื่อสองตัว Python จะจัดสรรอาร์กิวเมนต์ตามตำแหน่งส่วนเกินให้ `subjects` (เป็น tuple) และอาร์กิวเมนต์
แบบระบุชื่อทั้งหมดให้ `details` (เป็น dict) โดยอัตโนมัติ ตัวอย่างนี้แสดงให้เห็นว่าฟังก์ชันหนึ่งสามารถผสมพารามิเตอร์ปกติกับ
อาร์กิวเมนต์จำนวนไม่จำกัดทั้งสองแบบพร้อมกันได้ ซึ่งเป็นกรณีที่พบได้บ่อยในไลบรารีจริง

#### ตัวอย่างที่ 3 (โดเมนต่างออกไป): ออกใบเสร็จร้านอาหารด้วย `*args` และ `**kwargs`

````python
def print_receipt(customer, *items, **options):
    print("ลูกค้า:", customer)
    print("รายการอาหาร:", items)
    for key, value in options.items():
        print(f"{key}: {value}")

print_receipt("สมชาย", "ผัดไทย", "ส้มตำ", "น้ำแข็ง", discount=50, delivery="ด่วนพิเศษ")
````

**Output:**
```
ลูกค้า: สมชาย
รายการอาหาร: ('ผัดไทย', 'ส้มตำ', 'น้ำแข็ง')
discount: 50
delivery: ด่วนพิเศษ
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | หมายเหตุ |
|---|---|---|---|
| 7 | เรียก `print_receipt("สมชาย", "ผัดไทย", "ส้มตำ", "น้ำแข็ง", discount=50, delivery="ด่วนพิเศษ")` | customer="สมชาย", items=("ผัดไทย", "ส้มตำ", "น้ำแข็ง"), options={"discount": 50, "delivery": "ด่วนพิเศษ"} | อาร์กิวเมนต์ตามตำแหน่งตัวแรกเข้า customer ที่เหลือรวบเป็น tuple ใน items ส่วนอาร์กิวเมนต์แบบระบุชื่อรวบเป็น dict ใน options |
| 2 | `print(...)` | แสดง `ลูกค้า: สมชาย` | |
| 3 | `print(...)` | แสดง items เป็น tuple 3 รายการ | |
| 4 | เริ่มลูป `for key, value in options.items()` | key="discount", value=50 | รอบแรกของลูป |
| 5 | `print(...)` | แสดง `discount: 50` | |
| 4 | ลูปรอบถัดไป | key="delivery", value="ด่วนพิเศษ" | dict คงลำดับตามที่ใส่คีย์เวิร์ดตอนเรียกฟังก์ชัน |
| 5 | `print(...)` | แสดง `delivery: ด่วนพิเศษ` | สิ้นสุดลูป |

**คำอธิบายภาพรวม:** บรรทัด 1–5 นิยามฟังก์ชัน `print_receipt` ที่ผสมพารามิเตอร์ปกติ (`customer`) กับอาร์กิวเมนต์จำนวนไม่จำกัด
ทั้งสองแบบ เพื่อจำลองสถานการณ์จริงของการออกใบเสร็จร้านอาหาร ซึ่งจำนวนรายการอาหารไม่แน่นอนและอาจมีตัวเลือกเสริม (เช่น
ส่วนลด วิธีจัดส่ง) ที่ไม่แน่นอนเช่นกัน บรรทัด 7 เรียกฟังก์ชันด้วยรายการอาหาร 3 อย่างและตัวเลือกเสริม 2 รายการ แสดงให้เห็นว่า
`*items` และ `**options` ทำงานร่วมกับสถานการณ์ในชีวิตจริงได้ดี ซึ่งเป็นโดเมนที่ต่างจากการคำนวณผลรวมตัวเลข (ตัวอย่างที่ 1)
และรายงานผลการเรียนของนักเรียน (ตัวอย่างที่ 2)

## 9: การเรียกซ้ำ (Recursion)

### แนวคิด

**การเรียกซ้ำ (recursion)** คือเทคนิคที่ฟังก์ชันหนึ่งเรียกใช้ตัวเองภายในนิยามของมันเอง เพื่อแก้ปัญหาที่สามารถแบ่งย่อยเป็น
ปัญหาย่อยที่มีลักษณะเดียวกันแต่มีขนาดเล็กลงเรื่อย ๆ ได้ แนวคิดนี้เป็นทางเลือกหนึ่งของการวนลูป (loop) ในการทำงานซ้ำ ๆ
ทุกฟังก์ชันแบบเรียกซ้ำที่ถูกต้องต้องมีองค์ประกอบสำคัญสองส่วนเสมอ ได้แก่ **กรณีฐาน (base case)** ซึ่งเป็นเงื่อนไขที่ทำให้
ฟังก์ชันหยุดเรียกตัวเองและคืนค่าออกไปโดยตรง และ **กรณีเรียกซ้ำ (recursive case)** ซึ่งเป็นส่วนที่ฟังก์ชันเรียกตัวเองด้วย
อาร์กิวเมนต์ที่มีขนาดเล็กลงหรือใกล้เคียงกรณีฐานมากขึ้นในทุกครั้งที่เรียก

หากฟังก์ชันเรียกซ้ำขาดกรณีฐาน หรือกรณีฐานไม่มีวันถูกไปถึง ฟังก์ชันจะเรียกตัวเองไปเรื่อย ๆ ไม่มีที่สิ้นสุด ซึ่งในทางปฏิบัติ
เครื่องคอมพิวเตอร์มีขีดจำกัดของหน่วยความจำที่ใช้เก็บสถานะการเรียกฟังก์ชันแต่ละครั้ง (call stack) เมื่อการเรียกซ้ำลึกเกินขีดจำกัด
นี้ โปรแกรมจะหยุดทำงานด้วยข้อผิดพลาด ในภาษา Python ข้อผิดพลาดนี้คือ `RecursionError: maximum recursion depth
exceeded`

### ตัวอย่าง Python

#### ตัวอย่างที่ 1 (หลัก): หาค่าแฟกทอเรียลด้วยการเรียกซ้ำ

````python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)

print(factorial(4))
````

**Output:**
```
24
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | หมายเหตุ |
|---|---|---|---|
| 7 | เรียก `factorial(4)` | n=4 | เข้าสู่ call ที่ 1 — n ≠ 0 จึงยังไม่ถึงกรณีฐาน |
| 5 | `return 4 * factorial(3)` | n=4 (รอผลจาก factorial(3)) | เรียกซ้ำ เข้าสู่ call ที่ 2 |
| 5 | `return 3 * factorial(2)` | n=3 (รอผลจาก factorial(2)) | เรียกซ้ำ เข้าสู่ call ที่ 3 |
| 5 | `return 2 * factorial(1)` | n=2 (รอผลจาก factorial(1)) | เรียกซ้ำ เข้าสู่ call ที่ 4 |
| 5 | `return 1 * factorial(0)` | n=1 (รอผลจาก factorial(0)) | เรียกซ้ำ เข้าสู่ call ที่ 5 |
| 2–3 | `if n == 0: return 1` | n=0 → คืนค่า 1 | ถึงกรณีฐาน (base case) — หยุดการเรียกซ้ำ ไม่มี call ที่ 6 |
| 5 | call ที่ 5 คืนค่ากลับสู่ call ที่ 4 | 1 * 1 = 1 | เริ่มคลี่ (unwind) การเรียกซ้ำย้อนกลับ |
| 5 | call ที่ 4 คืนค่ากลับสู่ call ที่ 3 | 2 * 1 = 2 | |
| 5 | call ที่ 3 คืนค่ากลับสู่ call ที่ 2 | 3 * 2 = 6 | |
| 5 | call ที่ 2 คืนค่ากลับสู่ call ที่ 1 | 4 * 6 = 24 | |
| 7 | `print(...)` | แสดง 24 | ผลลัพธ์สุดท้ายที่ call ที่ 1 คืนกลับสู่จุดเรียกเริ่มต้น |

**คำอธิบายภาพรวม:** บรรทัด 1–5 นิยามฟังก์ชัน `factorial` ที่มีกรณีฐานอยู่ที่บรรทัด 2–3 (เมื่อ `n == 0` คืนค่า 1 ทันที)
และกรณีเรียกซ้ำอยู่ที่บรรทัด 5 (คูณ `n` กับผลลัพธ์ของ `factorial(n - 1)`) การเรียก `factorial(4)` ที่บรรทัด 7 ทำให้เกิด
การเรียกซ้ำทั้งหมด 5 ครั้ง (call ที่ 1–5) โดยแต่ละครั้งค่า `n` ลดลงทีละ 1 จนถึงกรณีฐานที่ `n = 0` จากนั้นแต่ละ call
จะคืนค่ากลับตามลำดับย้อนกลับ (unwind) พร้อมคูณค่าสะสมไปเรื่อย ๆ จนได้ผลลัพธ์สุดท้าย 24 ซึ่งเป็นรูปแบบการทำงานทั่วไปของ
ฟังก์ชันเรียกซ้ำที่คืนค่าตัวเลข

#### ตัวอย่างที่ 2 (edge case): ฟังก์ชันเรียกซ้ำที่ขาดกรณีฐาน แล้วแก้ไข



ก่อนพิจารณาตัวอย่างที่สมบูรณ์ ให้พิจารณาฟังก์ชันต่อไปนี้ซึ่ง **ขาดกรณีฐาน** โดยเจตนา เพื่อสาธิตว่าจะเกิดอะไรขึ้น:

```python
01  def countdown_broken(n):
02      print(n)
03      countdown_broken(n - 1)
04
05  countdown_broken(3)
```

ฟังก์ชันนี้ผ่านการรันจริงแล้ว (ไม่ใช่การคาดเดา) ผลลัพธ์จริงคือค่า `n` จะลดลงเรื่อย ๆ ไม่มีเงื่อนไขใดหยุดมันไว้ จนกระทั่ง
เกิน "ขีดจำกัดความลึกของการเรียกซ้ำ" ที่ Python กำหนดไว้ (ค่าเริ่มต้นประมาณ 1000 ระดับ) แล้วโปรแกรมจึงหยุดทำงานด้วย
ข้อผิดพลาด — Output จริง (ตัดตอนเนื่องจากยาวมาก):

```
3
2
1
0
-1
...
-994
Traceback (most recent call last):
  ...
RecursionError: maximum recursion depth exceeded while calling a Python object
```

ปัญหาของ `countdown_broken` คือไม่มีเงื่อนไขใดในฟังก์ชันที่ทำให้การเรียกซ้ำหยุดลง (ไม่มีกรณีฐาน) จึงต้องแก้ไขโดยเพิ่ม
เงื่อนไขตรวจสอบและ `return` ออกจากฟังก์ชันเมื่อถึงเงื่อนไขนั้น ดังนี้ (นี่คือตัวอย่างฉบับสมบูรณ์ที่นับเป็นตัวอย่างที่ 2):

````python
def countdown_fixed(n):
    if n == 0:
        print("จบการนับถอยหลัง")
        return
    print(n)
    countdown_fixed(n - 1)

countdown_fixed(3)
````

**Output:**
```
3
2
1
จบการนับถอยหลัง
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | หมายเหตุ |
|---|---|---|---|
| 8 | เรียก `countdown_fixed(3)` | n=3 | เข้าสู่ call ที่ 1 — n ≠ 0 ยังไม่ถึงกรณีฐาน |
| 5 | `print(n)` | แสดง 3 | |
| 6 | เรียก `countdown_fixed(2)` | n=2 | เรียกซ้ำ เข้าสู่ call ที่ 2 |
| 5 | `print(n)` | แสดง 2 | |
| 6 | เรียก `countdown_fixed(1)` | n=1 | เรียกซ้ำ เข้าสู่ call ที่ 3 |
| 5 | `print(n)` | แสดง 1 | |
| 6 | เรียก `countdown_fixed(0)` | n=0 | เรียกซ้ำ เข้าสู่ call ที่ 4 |
| 2–4 | `if n == 0: print(...); return` | แสดง `จบการนับถอยหลัง` แล้วคืนค่า None | ถึงกรณีฐาน — หยุดการเรียกซ้ำ ไม่มี call ที่ 5 |
| 6 | call ที่ 4 คืนค่ากลับสู่ call ที่ 3, 2, 1 ตามลำดับ | ไม่มีการคำนวณเพิ่มเติม | ฟังก์ชันนี้เป็น procedure (ไม่ return ค่าที่ใช้ต่อ) จึงเพียงคลี่ call stack กลับโดยไม่มีผลลัพธ์สะสม |

**คำอธิบายภาพรวม:** ตัวอย่าง `countdown_broken` (บรรทัด 1–5 ของโค้ดชุดแรก) แสดงให้เห็นผลลัพธ์จริงเมื่อฟังก์ชันเรียกซ้ำ
ขาดกรณีฐาน คือค่า `n` ลดลงไม่มีที่สิ้นสุดจนกระทั่งเกิด `RecursionError` เมื่อ call stack เต็ม ส่วน `countdown_fixed`
(บรรทัด 1–8 ของโค้ดชุดที่สอง) แก้ปัญหาด้วยการเพิ่มกรณีฐานที่บรรทัด 2–4: เมื่อ `n` ลดลงจนถึง 0 ฟังก์ชันจะพิมพ์ข้อความ
จบการทำงานแล้ว `return` ออกทันที ไม่เรียกตัวเองต่อ ทำให้การเรียกซ้ำหยุดลงอย่างถูกต้องหลังจาก 4 ครั้ง (call ที่ 1–4)
ตัวอย่างนี้ตอกย้ำว่ากรณีฐานไม่ใช่ทางเลือก แต่เป็นส่วนที่ขาดไม่ได้ของฟังก์ชันเรียกซ้ำทุกฟังก์ชัน

#### ตัวอย่างที่ 3 (โดเมนต่างออกไป): กลับข้อความด้วยการเรียกซ้ำ

````python
def reverse_text(text):
    if len(text) == 0:
        return text
    else:
        return reverse_text(text[1:]) + text[0]

print(reverse_text("Python"))
````

**Output:**
```
nohtyP
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | หมายเหตุ |
|---|---|---|---|
| 7 | เรียก `reverse_text("Python")` | text="Python" | เข้าสู่ call ที่ 1 — len(text)=6 ≠ 0 ยังไม่ถึงกรณีฐาน |
| 5 | `return reverse_text("ython") + "P"` | text="Python" (รอผลจาก call ถัดไป) | เรียกซ้ำด้วยข้อความที่ตัดตัวอักษรแรกออก เข้าสู่ call ที่ 2 |
| 5 | `return reverse_text("thon") + "y"` | text="ython" | เรียกซ้ำ เข้าสู่ call ที่ 3 |
| 5 | `return reverse_text("hon") + "t"` | text="thon" | เรียกซ้ำ เข้าสู่ call ที่ 4 |
| 5 | `return reverse_text("on") + "h"` | text="hon" | เรียกซ้ำ เข้าสู่ call ที่ 5 |
| 5 | `return reverse_text("n") + "o"` | text="on" | เรียกซ้ำ เข้าสู่ call ที่ 6 |
| 5 | `return reverse_text("") + "n"` | text="n" | เรียกซ้ำ เข้าสู่ call ที่ 7 |
| 2–3 | `if len(text) == 0: return text` | text="" → คืนค่า "" | ถึงกรณีฐาน (call ที่ 8) — หยุดการเรียกซ้ำ |
| 5 | call ที่ 7 คืนค่ากลับสู่ call ที่ 6 | "" + "n" = "n" | เริ่มคลี่ (unwind) การเรียกซ้ำย้อนกลับ |
| 5 | call ที่ 6 คืนค่ากลับสู่ call ที่ 5 | "n" + "o" = "no" | |
| 5 | call ที่ 5 คืนค่ากลับสู่ call ที่ 4 | "no" + "h" = "noh" | |
| 5 | call ที่ 4 คืนค่ากลับสู่ call ที่ 3 | "noh" + "t" = "noht" | |
| 5 | call ที่ 3 คืนค่ากลับสู่ call ที่ 2 | "noht" + "y" = "nohty" | |
| 5 | call ที่ 2 คืนค่ากลับสู่ call ที่ 1 | "nohty" + "P" = "nohtyP" | ผลลัพธ์สุดท้าย |
| 7 | `print(...)` | แสดง `nohtyP` | |

**คำอธิบายภาพรวม:** บรรทัด 1–5 นิยามฟังก์ชัน `reverse_text` ที่มีกรณีฐานอยู่ที่บรรทัด 2–3 (เมื่อข้อความว่างเปล่า คืนค่า
ข้อความว่างทันที) และกรณีเรียกซ้ำอยู่ที่บรรทัด 5 ซึ่งเรียกตัวเองด้วยข้อความที่ตัดตัวอักษรตัวแรกออก แล้วนำตัวอักษรตัวแรกนั้นไป
ต่อท้ายผลลัพธ์ที่ได้ การเรียก `reverse_text("Python")` ที่บรรทัด 7 ทำให้เกิดการเรียกซ้ำทั้งหมด 7 ครั้งก่อนถึงกรณีฐาน
(call ที่ 8) จากนั้นแต่ละ call จะคืนค่ากลับตามลำดับย้อนกลับ (unwind) พร้อมต่อท้ายตัวอักษรไปเรื่อย ๆ จนได้ผลลัพธ์สุดท้าย
`"nohtyP"` ซึ่งเป็นข้อความที่กลับด้านจาก `"Python"` ตัวอย่างนี้แสดงให้เห็นว่าการเรียกซ้ำใช้ได้กับโดเมนข้อมูลประเภทข้อความ
(str) ไม่ได้จำกัดอยู่แค่การคำนวณเชิงคณิตศาสตร์ (ตัวอย่างที่ 1) หรือการนับถอยหลังด้วยตัวเลข (ตัวอย่างที่ 2) เท่านั้น

### 6: By Value vs By Reference

**ข้อ 1 (เขียนอธิบายแนวคิด):** อธิบายว่าทำไมการส่งตัวแปรชนิด `int` เข้าไปในฟังก์ชันแล้วแก้ไขค่าภายในฟังก์ชัน จึงไม่กระทบ
ตัวแปรต้นฉบับภายนอกฟังก์ชัน ทั้งที่ Python ใช้กลไก "pass-by-object-reference" แบบเดียวกันกับกรณี `list`

**ข้อ 2 (ทายผลลัพธ์):** โค้ดต่อไปนี้พิมพ์ผลลัพธ์อะไร

```python
def modify(data):
    data[0] = 99

nums = [1, 2, 3]
modify(nums)
print(nums)
```

**ข้อ 3 (ทายผลลัพธ์):** โค้ดต่อไปนี้พิมพ์ผลลัพธ์อะไร

```python
def update(x):
    x += 100
    return x

value = 5
result = update(value)
print(value, result)
```

**ข้อ 4 (หาข้อผิดพลาด):** ผู้เขียนโค้ดต้องการให้ฟังก์ชัน `add_grade` เพิ่มเกรด `"D"` เข้าไปใน list `report` โดยตรง
แต่เมื่อรันแล้ว `report` กลับไม่เปลี่ยนแปลง จงบอกว่าโค้ดผิดพลาดตรงไหน และแก้ไขอย่างไร

```python
def add_grade(grades):
    grades = grades + ["D"]

report = ["A", "B", "C"]
add_grade(report)
print(report)
```

**ข้อ 5 (เขียนโค้ด):** เขียนฟังก์ชัน `clear_cart(cart)` ที่รับ list แล้วล้าง (empty) list นั้นแบบ in-place (ให้ list เดิม
กลายเป็น list ว่างจริง ไม่ใช่แค่ตัวแปรภายในฟังก์ชันเปลี่ยน) ทดสอบด้วย `cart = ["apple", "banana"]` แล้วเรียก
`clear_cart(cart)` จากนั้น `print(cart)` ควรได้ `[]`

**ข้อ 6 (เขียนโค้ด):** เขียนฟังก์ชัน `double_number(n)` รับตัวเลข คืนค่าตัวเลขที่คูณสอง (ไม่ต้องแก้ไขตัวแปรต้นฉบับ เพราะ
`int` เป็น immutable อยู่แล้ว) ทดสอบเรียกด้วย `x = 7` แล้ว `print(x, double_number(x))`

### 7: อาร์กิวเมนต์ค่าเริ่มต้น

**ข้อ 7 (เขียนอธิบายแนวคิด):** อธิบายว่าทำไมการใช้ mutable object (เช่น `list`) เป็นค่าเริ่มต้นของพารามิเตอร์ในฟังก์ชัน
Python จึงอาจก่อให้เกิดพฤติกรรมที่ไม่คาดคิด และวิธีแก้ไขมาตรฐานคืออะไร

**ข้อ 8 (ทายผลลัพธ์):** โค้ดต่อไปนี้พิมพ์ผลลัพธ์อะไร

```python
def power(base, exponent=2):
    return base ** exponent

print(power(3))
```

### 9: การเรียกซ้ำ (Recursion)

**ข้อ 19 (เขียนอธิบายแนวคิด):** อธิบายว่าทำไมฟังก์ชันเรียกซ้ำทุกฟังก์ชันต้องมีกรณีฐาน (base case) และจะเกิดอะไรขึ้นถ้า
ฟังก์ชันเรียกซ้ำไม่มีกรณีฐานหรือกรณีฐานไม่มีวันถูกไปถึง

**ข้อ 20 (ทายผลลัพธ์):** โค้ดต่อไปนี้พิมพ์ผลลัพธ์อะไร

```python
def count_down_sum(n):
    if n == 0:
        return 0
    return n + count_down_sum(n - 1)

print(count_down_sum(5))
```

**ข้อ 21 (ทายผลลัพธ์):** โค้ดต่อไปนี้พิมพ์ผลลัพธ์อะไร

```python
    if times == 0:
        return ""
    return ch + repeat_char(ch, times - 1)

print(repeat_char("*", 4))
```

**ข้อ 22 (หาข้อผิดพลาด):** โค้ดต่อไปนี้ทำให้เกิด `RecursionError` เมื่อเรียก `sum_to_n(0)` จงบอกว่าปัญหาคืออะไร และ
แก้ไขกรณีฐานอย่างไรให้ถูกต้อง (ต้องใช้ได้ทั้งกรณี `sum_to_n(0)` และ `sum_to_n(5)`)

```python
def sum_to_n(n):
    if n == 1:
        return 1
    return n + sum_to_n(n - 1)

print(sum_to_n(0))
```
