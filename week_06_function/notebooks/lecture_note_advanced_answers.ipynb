````python
def apply_all(funcs, value):
    results = []
    for f in funcs:
        results.append(f(value))
    return results

def square(x):
    return x * x

def negate(x):
    return -x

print(apply_all([square, negate, abs], -4))
````

ของลำดับข้อมูล) และ `filter()` (รับฟังก์ชันเป็นอาร์กิวเมนต์เพื่อคัดกรองสมาชิกที่ผ่านเงื่อนไข) รวมถึง `sorted()` เมื่อใช้พารามิเตอร์ `key`

**เฉลยข้อ 11.** ปัญหานี้เกิดจาก **closure จำ "ตัวแปร" ไม่ใช่ "ค่า ณ ขณะสร้าง"** — lambda ทุกตัวที่สร้างขึ้นในลูปต่างอ้างอิงตัวแปร
`name` ตัวเดียวกันของ scope ฟังก์ชัน `make_greeters` เมื่อลูปทำงานจบ `name` จะเหลือค่าสุดท้ายคือ `"ชูใจ"` ดังนั้นเมื่อเรียกฟังก์ชัน
ทั้งสามตัวในภายหลัง ทุกตัวจึงอ่านค่า `name` ล่าสุดที่เหลืออยู่เหมือนกันหมด

Output จริงของโค้ดที่มีปัญหา:
```
สวัสดี ชูใจ
สวัสดี ชูใจ
สวัสดี ชูใจ
```

วิธีแก้คือผูกค่าปัจจุบันของ `name` ไว้เป็นค่าเริ่มต้นของพารามิเตอร์เพิ่มเติมใน lambda (ประเมินค่าและผูกไว้ทันทีตอนสร้าง ไม่ใช่ตอนเรียก):

````python
def make_greeters(names):
    greeters = []
    for name in names:
        greeters.append(lambda name=name: print(f"สวัสดี {name}"))
    return greeters

funcs = make_greeters(["มานะ", "ปิติ", "ชูใจ"])
for f in funcs:
    f()
````

Output หลังแก้ไข:
```
สวัสดี มานะ
สวัสดี ปิติ
สวัสดี ชูใจ
```

**เฉลยข้อ 12.**

````python
def compose(f, g):
    def composed(x):
        return f(g(x))
    return composed

def add_one(x):
    return x + 1

def double(x):
    return x * 2

add_then_double = compose(double, add_one)
print(add_then_double(5))
````

Output:
```
12
```
คำอธิบาย: `compose(double, add_one)` คืนฟังก์ชันใหม่ `composed` ที่เมื่อถูกเรียกด้วย `x` จะคำนวณ `double(add_one(x))`
เมื่อเรียก `add_then_double(5)` จะได้ `add_one(5)=6` แล้ว `double(6)=12`

### หัวข้อ 13: ฟังก์ชันบริสุทธิ์ กับ ผลข้างเคียง

**เฉลยข้อ 13.** ฟังก์ชันบริสุทธิ์ต้องมีคุณสมบัติสองข้อพร้อมกัน คือ (1) เมื่อเรียกด้วยอาร์กิวเมนต์ชุดเดียวกัน จะได้ผลลัพธ์เดียวกันเสมอ
และ (2) ไม่ก่อให้เกิดการเปลี่ยนแปลงใด ๆ ที่มองเห็นได้จากภายนอกขอบเขตของตัวเอง ตัวอย่างของผลข้างเคียงที่พบบ่อย ได้แก่
(1) การแก้ไขตัวแปรส่วนกลาง (global variable) (2) การพิมพ์ข้อความออกทางหน้าจอหรือเขียนไฟล์ และ (3) การแก้ไขข้อมูลที่เปลี่ยนแปลงได้
(mutable) ที่รับเข้ามาเป็นอาร์กิวเมนต์โดยตรง เช่น การ `append` เข้า list ที่ผู้เรียกส่งเข้ามา

**เฉลยข้อ 14.**
```
1
2
2
```
คำอธิบาย: ทุกครั้งที่เรียก `increment()` ฟังก์ชันจะแก้ไขตัวแปรส่วนกลาง `counter` ผ่านคำสั่ง `global counter` แล้วคืนค่าที่เพิ่มขึ้น
การเรียกครั้งแรกได้ 1 ครั้งที่สองได้ 2 และเมื่อพิมพ์ `counter` ในบรรทัดสุดท้ายก็ยังคงเป็น 2 เพราะค่าถูกแก้ไขไว้แล้วจากการเรียกฟังก์ชัน
ทั้งสองครั้งก่อนหน้า (ตัวอย่างของฟังก์ชันที่ไม่บริสุทธิ์เพราะพึ่งพาและแก้ไขสถานะภายนอก)

**เฉลยข้อ 15.** สาเหตุคือ `cart=[]` เป็น **ค่าเริ่มต้นของพารามิเตอร์ที่เป็น mutable object (list)** ซึ่งใน Python จะถูกสร้างขึ้น
**เพียงครั้งเดียว** ตอนที่ฟังก์ชันถูกนิยาม (ไม่ใช่ทุกครั้งที่ถูกเรียก) ดังนั้นทุกการเรียก `add_item(...)` ที่ไม่ได้ระบุ `cart` เอง
จะใช้ list เดิมร่วมกันซ้ำ ๆ ทำให้ค่าที่เพิ่มไว้จากการเรียกครั้งก่อนยังค้างอยู่ในการเรียกครั้งถัดไป

Output จริงของโค้ดที่มีปัญหา:
```
['apple']
['apple', 'banana']
```

วิธีแก้ตามหลักการของฟังก์ชันบริสุทธิ์คือไม่ใช้ mutable object เป็นค่าเริ่มต้นของพารามิเตอร์ ให้ใช้ `None` แทนแล้วสร้าง list ใหม่
ภายในฟังก์ชันทุกครั้งที่ไม่ได้รับ `cart` มา:

````python
def add_item(item, cart=None):
    if cart is None:
        cart = []
    cart.append(item)
    return cart

print(add_item("apple"))
print(add_item("banana"))
````

Output หลังแก้ไข:
```
['apple']
['banana']
```

วิธีแก้คือเพิ่ม `else` เพื่อให้ทุกเส้นทางคืนค่าที่มีชนิดสอดคล้องกัน (ตัวเลขเสมอ):

````python
def get_discount_rate(member_level):
    if member_level == "gold":
        return 0.2
    elif member_level == "silver":
        return 0.1
    else:
        return 0.0

rate = get_discount_rate("bronze")
price = 100 * (1 - rate)
print(price)
````

**เฉลยข้อ 23.** ฟังก์ชัน `register_user` รวมสามความรับผิดชอบไว้ในที่เดียว คือ (1) ตรวจสอบความถูกต้องของอายุ (2) บันทึกชื่อผู้ใช้ลง
`users` และ (3) แสดงข้อความแจ้งผล ปัญหาคือเมื่อพบว่าอายุไม่ถูกต้อง (บรรทัด `if age < 0: print(...)`) โค้ดกลับไม่มีคำสั่ง `return`
เพื่อหยุดการทำงานทันที จึงทำงานต่อไปจนถึงบรรทัด `users.append(username)` และ `print("ลงทะเบียน...สำเร็จ")` ราวกับว่าการลงทะเบียน
สำเร็จทั้งที่ข้อมูลไม่ถูกต้อง ปัญหานี้เกิดง่ายเพราะฟังก์ชันทำหลายหน้าที่ปนกัน ทำให้ตรวจสอบเงื่อนไขแล้ว "ลืม" หยุดการทำงานของ
ส่วนอื่นที่ไม่เกี่ยวข้อง

Output จริงของโค้ดที่มีปัญหา:
```
อายุไม่ถูกต้อง
ลงทะเบียน มานะ สำเร็จ
['มานะ']
```

วิธีแก้ตามหลัก SRP คือแยกเป็นฟังก์ชันย่อยตามความรับผิดชอบ (ตรวจสอบ / บันทึก / แจ้งผล) แล้วให้ฟังก์ชันหลักคืนออกจากการทำงานทันที
เมื่อตรวจสอบไม่ผ่าน:

````python
users = []

def is_valid_age(age):
    return age >= 0

def save_user(username):
    users.append(username)

def notify_registration(username):
    print(f"ลงทะเบียน {username} สำเร็จ")

def register_user(username, age):
    if not is_valid_age(age):
        print("อายุไม่ถูกต้อง")
        return
    save_user(username)
    notify_registration(username)

register_user("มานะ", -5)
print(users)
````

Output หลังแก้ไข:
```
อายุไม่ถูกต้อง
[]
```

````python
def validate_email(email):
    return "@" in email and "." in email

def send_welcome_message(email):
    print(f"ส่งข้อความต้อนรับไปยัง {email}")

test_email = "student@example.com"
if validate_email(test_email):
    send_welcome_message(test_email)
````
