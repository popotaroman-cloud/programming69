**เฉลยข้อ 1:** เพราะ `int` เป็นชนิดข้อมูล **immutable (เปลี่ยนแปลงค่าไม่ได้)** เมื่อเข้าสู่ฟังก์ชัน พารามิเตอร์จะอ้างอิง
อ็อบเจกต์เดียวกับตัวแปรต้นทางก็จริง แต่การกำหนดค่าใหม่ให้พารามิเตอร์ (เช่น `x = x + 1`) ไม่ได้แก้ไขอ็อบเจกต์เดิม แต่เป็นการ
สร้างอ็อบเจกต์ใหม่แล้วให้พารามิเตอร์ชี้ไปที่อ็อบเจกต์ใหม่นั้นแทน ตัวแปรต้นทางภายนอกฟังก์ชันจึงยังคงชี้ไปยังอ็อบเจกต์เดิม
ไม่เปลี่ยนแปลง ต่างจาก `list` ซึ่งเป็น mutable — การเรียกเมธอดที่แก้ไขแบบ in-place (เช่น `append`) จะแก้ไขอ็อบเจกต์เดิม
โดยตรง จึงกระทบตัวแปรต้นทางด้วย

`value` เป็น `int` (immutable) การกระทำ `x += 100` ภายในฟังก์ชันสร้างอ็อบเจกต์ใหม่ให้ `x` ไม่กระทบ `value` เดิม
ฟังก์ชันต้อง `return x` ออกมาแล้วรับด้วยตัวแปร `result` ค่าที่เปลี่ยนแปลงจึงไปปรากฏที่ `result` แทน

**เฉลยข้อ 4:** โค้ดผิดพลาดตรงบรรทัด `grades = grades + ["D"]` ซึ่งเป็นการ**สร้าง list ใหม่**จากการต่อ (`+`) แล้วให้
พารามิเตอร์ `grades` ชี้ไปยัง list ใหม่นั้น ไม่ได้แก้ไข list เดิมแบบ in-place ตัวแปร `report` ภายนอกฟังก์ชันจึงยังชี้ไปยัง
list เดิมที่ไม่มี `"D"` วิธีแก้คือใช้เมธอดที่แก้ไขแบบ in-place แทน เช่น `grades.append("D")`:

````python
def add_grade(grades):
    grades.append("D")

report = ["A", "B", "C"]
add_grade(report)
print(report)
````

Output ที่รันจริงหลังแก้ไข:
```
['A', 'B', 'C', 'D']
```

**เฉลยข้อ 17:**

````python
def find_max(*numbers):
    if len(numbers) == 0:
        return None
    result = numbers[0]
    for n in numbers:
        if n > result:
            result = n
    return result

print(find_max(3, 7, 2, 9, 4))
print(find_max())
````

Output:
```
9
None
```

**เฉลยข้อ 22:** ปัญหาคือกรณีฐานตรวจสอบแค่ `n == 1` เท่านั้น เมื่อเรียก `sum_to_n(0)` ค่า `n` จะลดลงจาก 0 เป็น -1, -2,
-3, ... ไปเรื่อย ๆ โดยไม่มีวันเท่ากับ 1 พอดี ทำให้เกิดการเรียกซ้ำไม่มีที่สิ้นสุดจนเกิด `RecursionError` (ยืนยันจากการรันจริง)
วิธีแก้คือเปลี่ยนเงื่อนไขกรณีฐานเป็น `n <= 0` แทน `n == 1` เพื่อให้ครอบคลุมทั้งกรณี `n = 0` และค่าติดลบ:

````python
def sum_to_n(n):
    if n <= 0:
        return 0
    return n + sum_to_n(n - 1)

print(sum_to_n(0))
print(sum_to_n(5))
````

Output ที่รันจริงหลังแก้ไข:
```
0
15
```

**เฉลยข้อ 23:**

````python
def count_vowels(text):
    if len(text) == 0:
        return 0
    first = 1 if text[0] in "aeiou" else 0
    return first + count_vowels(text[1:])

print(count_vowels("education"))
````
