### ข้อที่ 1 (Pseudocode)

**โจทย์**: ฝ่ายบุคคลของบริษัทมีรายชื่อพนักงานเก็บเป็นชุดข้อมูล (แต่ละคนมีฟิลด์ `name` และ `years` คืออายุงานเป็นปี)
ต้องการเรียงลำดับพนักงานจากอายุงานน้อยไปมาก โดยใช้ฟังก์ชันนิรนาม (lambda) เป็น "key function" ระบุว่าจะเรียงลำดับตามฟิลด์ใด

**Pseudocode**:

```
SET employees TO [
    { name: "สมชาย", years: 5 },
    { name: "สมหญิง", years: 2 },
    { name: "วิชัย", years: 8 }
]

SET get_years TO LAMBDA (employee) RETURNS employee.years

SET sorted_employees TO SORT employees BY get_years IN ASCENDING ORDER

FOR EACH employee IN sorted_employees
    PRINT employee.name, employee.years
END FOR EACH
```

ร่วมกับฟังก์ชันเรียงลำดับของ Python) แล้วรันเพื่อตรวจคำตอบ (คำเฉลยอยู่ในหมวด "เฉลย" ท้ายไฟล์)

**โจทย์**: ระบบบันทึกเกรดต้องการฟังก์ชันคำนวณเกรดเฉลี่ยแบบถ่วงน้ำหนัก (weighted average) จากรายวิชาที่นักศึกษาลงทะเบียน
โดยฟังก์ชันต้องเป็นฟังก์ชันบริสุทธิ์ (pure function) คือคำนวณแล้วคืนค่าออกไปเท่านั้น ห้ามแก้ไขข้อมูลเดิมหรือพิมพ์ค่าใด ๆ
