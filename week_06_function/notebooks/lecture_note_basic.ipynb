## 1: จุดประสงค์ของฟังก์ชัน (DRY, Decomposition)

### แนวคิด

ฟังก์ชัน (function) คือกลุ่มคำสั่งที่ถูกตั้งชื่อไว้และสามารถเรียกใช้ซ้ำได้หลายครั้งโดยไม่ต้องเขียนคำสั่งชุดเดิมซ้ำใหม่ทุกครั้ง
จุดประสงค์หลักของการใช้ฟังก์ชันมีสองประการที่สัมพันธ์กัน ประการแรกคือหลักการ **DRY (Don't Repeat Yourself)**
ซึ่งหมายถึงการเขียนตรรกะการทำงานหนึ่งอย่างไว้เพียงจุดเดียวในโปรแกรม แล้วเรียกใช้ซ้ำเมื่อจำเป็น
แทนที่จะคัดลอกวางคำสั่งชุดเดิมไปไว้หลายจุด เพราะหากต้องแก้ไขตรรกะนั้นในภายหลัง การมีอยู่จุดเดียวจะทำให้แก้ไขได้ง่ายและลดโอกาสเกิดข้อผิดพลาด
ประการที่สองคือ **decomposition** หรือการแตกปัญหาใหญ่ที่ซับซ้อนให้กลายเป็นปัญหาย่อยหลายส่วน โดยแต่ละส่วนเขียนเป็นฟังก์ชันของตนเอง
ทำให้โปรแกรมอ่านง่ายขึ้น ทดสอบแยกส่วนได้ และดูแลรักษาง่ายขึ้นในระยะยาว

ในภาษา Python ฟังก์ชันถูกสร้างขึ้นด้วยคำสั่ง `def` ตามด้วยชื่อฟังก์ชันและวงเล็บ ดังจะเห็นในตัวอย่างถัดไป
ซึ่งแสดงให้เห็นว่าเมื่อมีฟังก์ชันแล้ว การนำตรรกะเดิมไปใช้ซ้ำทำได้เพียงเรียกชื่อฟังก์ชัน ไม่ต้องเขียนสูตรคำนวณซ้ำใหม่ทุกครั้ง

### ตัวอย่าง Python

#### ตัวอย่างที่ 1.1 — แนวคิดหลัก: ใช้ฟังก์ชันเดียวคำนวณซ้ำหลายครั้ง (DRY)

````python
def calculate_area(width, height):
    return width * height

print(calculate_area(4, 5))
print(calculate_area(6, 2))
print(calculate_area(3, 3))
````

**Output:**
```

9
```

| 4 | `print(...)` | แสดงผล 20 | พิมพ์ผลลัพธ์ครั้งที่ 1 |

| 2 | `return width * height` | คืนค่า 12 | คำนวณพื้นที่และส่งกลับ |

แสดงหลักการ DRY อย่างชัดเจน คือเขียนตรรกะการคำนวณพื้นที่ (`width * height`) ไว้ที่เดียว แล้วนำกลับมาใช้ซ้ำได้เรื่อย ๆ
แทนที่จะคัดลอกสูตรไปเขียนซ้ำสามแห่ง ซึ่งหากภายหลังต้องแก้สูตรคำนวณ ก็แก้เพียงจุดเดียวคือบรรทัด 2

#### ตัวอย่างที่ 1.2 — บริบทอื่นที่ต่างโดเมน: DRY กับงานจัดรูปแบบข้อความ (ไม่ใช่แค่คำนวณเลข)

````python
def format_receipt_line(item, price):
    return f"{item} ราคา {price} บาท"

print(format_receipt_line("กาแฟ", 45))
print(format_receipt_line("ขนมปัง", 25))
print(format_receipt_line("น้ำส้ม", 30))
````

(ซึ่งเป็นโดเมนตัวเลขเชิงเรขาคณิต) ใช้ได้ผลเช่นเดียวกันกับงานด้านข้อความ (text) เช่นการจัดรูปแบบรายการใบเสร็จ ไม่ได้จำกัดอยู่แค่การคำนวณตัวเลขเท่านั้น

#### ตัวอย่างที่ 1.3 — ข้อผิดพลาดที่พบบ่อย: ไม่ทำ DRY แล้วคัดลอกสูตรผิด

````python
def celsius_to_fahrenheit_v1(celsius):
    return celsius * 9 / 5 + 32

def celsius_to_fahrenheit_v2(celsius):
    return celsius * 9 / 5 + 3  # พิมพ์ผิด (32 กลายเป็น 3) เพราะคัดลอกสูตรไปเขียนซ้ำ

temp = 25
print(celsius_to_fahrenheit_v1(temp))
print(celsius_to_fahrenheit_v2(temp))
````

**Output:**
```
77.0
48.0
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | หมายเหตุ |
|---|---|---|---|
| 7 | `temp = 25` | temp=25 | กำหนดค่าตั้งต้น |
| 8 | เรียก `celsius_to_fahrenheit_v1(25)` | celsius=25 | เข้าฟังก์ชันเวอร์ชันที่ถูกต้อง |
| 2 | `return celsius * 9 / 5 + 32` | คืนค่า 77.0 | สูตรแปลงองศาที่ถูกต้อง |
| 8 | `print(...)` | แสดงผล 77.0 | ผลลัพธ์ที่คาดหวัง |
| 9 | เรียก `celsius_to_fahrenheit_v2(25)` | celsius=25 | เข้าฟังก์ชันเวอร์ชันที่คัดลอกสูตรมาเขียนใหม่ |
| 5 | `return celsius * 9 / 5 + 3` | คืนค่า 48.0 | สูตรผิดเพราะพิมพ์เลข 32 เป็น 3 ตอนคัดลอก |
| 9 | `print(...)` | แสดงผล 48.0 | ผลลัพธ์ผิดพลาด ทั้งที่ตั้งใจให้เหมือนกับฟังก์ชันแรก |

**คำอธิบายภาพรวม:** บรรทัด 1–2 และ 4–5 เป็นฟังก์ชันสองตัวที่ผู้เขียนตั้งใจให้ทำงานเหมือนกัน แต่ `celsius_to_fahrenheit_v2`
เกิดจากการคัดลอกสูตรของฟังก์ชันแรกมาพิมพ์ใหม่แทนที่จะเรียกใช้ฟังก์ชันเดิมซ้ำ บรรทัด 8–9 เรียกทั้งสองฟังก์ชันด้วยอินพุตเดียวกัน (25)
แต่ได้ผลลัพธ์ต่างกัน (77.0 กับ 48.0) เพราะพิมพ์เลข 32 ผิดเป็น 3 ระหว่างคัดลอก ตัวอย่างนี้แสดงความเสี่ยงของการไม่ทำ DRY:
เมื่อตรรกะเดียวกันถูกเขียนซ้ำหลายแห่ง การพิมพ์ผิดหรือแก้ไขในจุดหนึ่งจะไม่ส่งผลไปยังจุดอื่นโดยอัตโนมัติ ทำให้ตรรกะของโปรแกรมไม่สอดคล้องกัน

## 2: Function Definition กับ Function Call

### แนวคิด

การใช้งานฟังก์ชันแบ่งออกเป็นสองขั้นตอนที่แยกจากกันอย่างชัดเจน ขั้นตอนแรกคือ **การนิยามฟังก์ชัน (function definition)**
คือการบอกให้โปรแกรมรู้จักว่าฟังก์ชันนี้ชื่ออะไร รับข้อมูลอะไรเข้ามา และมีคำสั่งอะไรอยู่ภายในบ้าง ขั้นตอนนี้เป็นเพียงการ "ประกาศ" ไว้เท่านั้น
คำสั่งภายในฟังก์ชันจะยังไม่ถูกรันจนกว่าจะถึงขั้นตอนที่สอง คือ **การเรียกใช้ฟังก์ชัน (function call)**
ซึ่งเป็นจุดที่โปรแกรมกระโดดเข้าไปรันคำสั่งภายในฟังก์ชันนั้นจริง ๆ ฟังก์ชันหนึ่งสามารถถูกนิยามเพียงครั้งเดียว แต่ถูกเรียกใช้ได้หลายครั้งตามต้องการ
ความเข้าใจผิดที่พบบ่อยของผู้เริ่มต้นคือคิดว่าเพียงแค่นิยามฟังก์ชันไว้ โปรแกรมก็จะรันคำสั่งภายในให้โดยอัตโนมัติ ซึ่งไม่เป็นความจริง

ในภาษา Python การนิยามฟังก์ชันใช้คำสั่ง `def ชื่อฟังก์ชัน():` ตามด้วยกลุ่มคำสั่งที่เยื้องเข้าไป (indent)
ส่วนการเรียกใช้ทำได้โดยเขียนชื่อฟังก์ชันตามด้วยวงเล็บ `()` ดังตัวอย่างต่อไปนี้

### ตัวอย่าง Python

#### ตัวอย่างที่ 2.1 — แนวคิดหลัก: นิยามครั้งเดียว เรียกใช้ได้หลายครั้ง

````python
def greet():
    print("สวัสดี")

print("ก่อนเรียกฟังก์ชัน")
greet()
greet()
print("หลังเรียกฟังก์ชัน")
````

**Output:**
```
ก่อนเรียกฟังก์ชัน
สวัสดี
สวัสดี
หลังเรียกฟังก์ชัน
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | หมายเหตุ |
|---|---|---|---|
| 1–2 | นิยามฟังก์ชัน `greet` | (ยังไม่ทำงาน) | เพียงบันทึกนิยามไว้ ยังไม่มีการรันโค้ดข้างในบรรทัด 2 |
| 4 | `print("ก่อนเรียกฟังก์ชัน")` | แสดงผล | ทำงานตามลำดับปกติของโปรแกรม |
| 5 | เรียก `greet()` ครั้งที่ 1 | เข้าสู่ scope ของ greet | เริ่มรันโค้ดในฟังก์ชัน |
| 2 | `print("สวัสดี")` | แสดงผล | พิมพ์จากภายในฟังก์ชัน |
| 6 | เรียก `greet()` ครั้งที่ 2 | เข้าสู่ scope ของ greet อีกครั้ง | เรียกซ้ำด้วยนิยามเดิม ได้ผลเหมือนเดิม |
| 2 | `print("สวัสดี")` | แสดงผล | พิมพ์ซ้ำ |
| 7 | `print("หลังเรียกฟังก์ชัน")` | แสดงผล | พิมพ์ข้อความหลังเรียกฟังก์ชันครบ 2 ครั้ง |

**คำอธิบายภาพรวม:** บรรทัด 1–2 เป็นเพียงการนิยามฟังก์ชัน (definition) ณ จุดนี้ Python ยังไม่รันคำสั่ง `print` ที่อยู่ภายใน
มีเพียงการสร้างฟังก์ชันไว้ใช้งานเท่านั้น บรรทัด 5 และ 6 คือการเรียกใช้ (call) ฟังก์ชันเดิมสองครั้ง ทำให้โค้ดในบรรทัด 2 ถูกรันจริงสองครั้ง
แสดงให้เห็นว่า "การนิยาม" กับ "การเรียกใช้" เป็นคนละขั้นตอนกัน นิยามเพียงครั้งเดียวแต่เรียกใช้ได้หลายครั้งตามต้องการ

#### ตัวอย่างที่ 2.2 — บริบทอื่นที่ต่างโดเมน: เรียกฟังก์ชันเดิมซ้ำผ่านลูปในงานประมวลผลตัวเลข

````python
def square(number):
    return number * number

numbers = [2, 4, 6]
for n in numbers:
    print(square(n))
````

## 3: Parameter กับ Argument

### แนวคิด

เมื่อออกแบบฟังก์ชันให้รับข้อมูลเข้ามาประมวลผล จำเป็นต้องเข้าใจคำศัพท์สองคำที่มักถูกใช้ปะปนกัน คือ **พารามิเตอร์ (parameter)**
และ **อาร์กิวเมนต์ (argument)** พารามิเตอร์คือชื่อตัวแปรที่ถูกระบุไว้ในวงเล็บตอน**นิยาม**ฟังก์ชัน ทำหน้าที่เป็นเพียง "ช่องว่าง"
ที่รอรับค่าเข้ามา ยังไม่มีค่าจริงอยู่ในตัว ส่วนอาร์กิวเมนต์คือค่าจริงที่ถูกส่งเข้าไปตอน**เรียกใช้**ฟังก์ชัน ซึ่งจะถูกนำไปผูก (bind)
เข้ากับพารามิเตอร์ตามลำดับตำแหน่งที่ปรากฏ วิธีการส่งอาร์กิวเมนต์ตามลำดับตำแหน่งเช่นนี้เรียกว่า
**อาร์กิวเมนต์ตามตำแหน่ง (positional argument)** ซึ่งอาร์กิวเมนต์ตัวแรกจะถูกผูกกับพารามิเตอร์ตัวแรกเสมอ
ตัวที่สองผูกกับตัวที่สอง เรียงตามลำดับไปเรื่อย ๆ โดยภาษาโปรแกรมส่วนใหญ่จะไม่ตรวจสอบว่าค่าที่ส่งเข้ามาตรงกับความหมายที่ต้องการหรือไม่
ทำให้การส่งอาร์กิวเมนต์ผิดลำดับเป็นข้อผิดพลาดที่พบได้บ่อยและตรวจจับได้ยาก

ในภาษา Python พารามิเตอร์ถูกระบุไว้ในวงเล็บของ `def` เช่น `def introduce(name, age):` ส่วนอาร์กิวเมนต์คือค่าที่ส่งเข้าไปตอนเรียก
เช่น `introduce("Aom", 20)` ซึ่ง `"Aom"` จะถูกผูกกับ `name` และ `20` จะถูกผูกกับ `age` ตามลำดับตำแหน่ง

### ตัวอย่าง Python

#### ตัวอย่างที่ 3.1 — แนวคิดหลัก: พารามิเตอร์รับค่าจากอาร์กิวเมนต์ตามลำดับตำแหน่ง

````python
def introduce(name, age):
    print(f"ฉันชื่อ {name} อายุ {age} ปี")

introduce("Aom", 20)
introduce("Beam", 22)
````

**Output:**
```
ฉันชื่อ Aom อายุ 20 ปี
ฉันชื่อ Beam อายุ 22 ปี
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | หมายเหตุ |
|---|---|---|---|
| 1 | นิยามฟังก์ชัน `introduce(name, age)` | name, age เป็นพารามิเตอร์ | กำหนดชื่อตัวแปรที่จะรับค่า ยังไม่มีค่าจริง |
| 4 | เรียก `introduce("Aom", 20)` | name="Aom", age=20 | ส่งอาร์กิวเมนต์ "Aom" และ 20 เข้าไปผูกกับพารามิเตอร์ตามลำดับ |
| 2 | `print(f"...")` | แสดงผล ฉันชื่อ Aom อายุ 20 ปี | ใช้ค่าของพารามิเตอร์ในการพิมพ์ |
| 5 | เรียก `introduce("Beam", 22)` | name="Beam", age=22 | เรียกซ้ำด้วยอาร์กิวเมนต์ชุดใหม่ |
| 2 | `print(f"...")` | แสดงผล ฉันชื่อ Beam อายุ 22 ปี | พารามิเตอร์ชุดเดิม รับค่าใหม่ทุกครั้งที่ถูกเรียก |

**คำอธิบายภาพรวม:** บรรทัด 1 นิยามฟังก์ชันโดยระบุพารามิเตอร์ `name` และ `age` ซึ่งเป็นเพียงชื่อตัวแปรที่รอรับค่า ยังไม่ใช่ค่าจริง
บรรทัด 4 และ 5 คือการเรียกใช้ฟังก์ชันพร้อมส่งอาร์กิวเมนต์ ได้แก่ค่าจริงเช่น `"Aom"` และ `20` เข้าไปผูกกับพารามิเตอร์ตามลำดับตำแหน่ง
ทำให้ฟังก์ชันเดียวกันสามารถประมวลผลข้อมูลที่แตกต่างกันได้ในแต่ละครั้งที่ถูกเรียก

#### ตัวอย่างที่ 3.2 — บริบทอื่นที่ต่างโดเมน: คำนวณราคารวมด้วยอาร์กิวเมนต์ตามตำแหน่งและแบบระบุชื่อ

````python
def calculate_total(price, quantity):
    return price * quantity

print(calculate_total(50, 3))
print(calculate_total(quantity=3, price=50))
````

**Output:**
```
150
150
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | หมายเหตุ |
|---|---|---|---|
| 1 | นิยามฟังก์ชัน `calculate_total(price, quantity)` | price, quantity เป็นพารามิเตอร์ | กำหนดชื่อตัวแปรที่จะรับค่า ยังไม่มีค่าจริง |
| 4 | เรียก `calculate_total(50, 3)` | price=50, quantity=3 | ส่งอาร์กิวเมนต์ตามตำแหน่ง (positional) 50 ผูกกับ price, 3 ผูกกับ quantity |
| 2 | `return price * quantity` | คืนค่า 150 | คำนวณราคารวมสินค้า |
| 4 | `print(...)` | แสดงผล 150 | ผลลัพธ์จากการเรียกแบบ positional |
| 5 | เรียก `calculate_total(quantity=3, price=50)` | price=50, quantity=3 | ส่งอาร์กิวเมนต์แบบระบุชื่อ (keyword argument) ลำดับที่เขียนสลับกันได้เพราะระบุชื่อพารามิเตอร์กำกับไว้ |
| 2 | `return price * quantity` | คืนค่า 150 | ได้ผลลัพธ์เดียวกับการเรียกแบบ positional |
| 5 | `print(...)` | แสดงผล 150 | ผลลัพธ์จากการเรียกแบบ keyword argument |

**คำอธิบายภาพรวม:** บรรทัด 1–2 นิยามฟังก์ชัน `calculate_total` ในบริบทการคำนวณราคารวมสินค้า ต่างจากตัวอย่างที่ 3.1 ที่เป็นการแนะนำตัวบุคคล
บรรทัด 4 เรียกฟังก์ชันด้วยอาร์กิวเมนต์ตามตำแหน่ง (positional argument) เช่นเดียวกับตัวอย่างที่ 3.1 ส่วนบรรทัด 5 สาธิตอาร์กิวเมนต์แบบระบุชื่อ
(keyword argument) ซึ่งระบุชื่อพารามิเตอร์กำกับไว้ตอนเรียกใช้ ทำให้สลับลำดับการเขียนอาร์กิวเมนต์ได้โดยไม่เปลี่ยนความหมาย เพราะ Python
ผูกค่าตามชื่อพารามิเตอร์แทนที่จะผูกตามตำแหน่งเมื่อระบุชื่อไว้ชัดเจน ต่างจากตัวอย่างที่ 3.3 (ถัดไป) ที่แสดงว่าการส่งอาร์กิวเมนต์ตามตำแหน่งผิดลำดับ
จะทำให้ค่าผูกผิดพารามิเตอร์โดยไม่มี error เตือน

#### ตัวอย่างที่ 3.3 — ข้อผิดพลาดที่พบบ่อย: ส่งอาร์กิวเมนต์ผิดลำดับ

````python
def introduce(name, age):
    print(f"ฉันชื่อ {name} อายุ {age} ปี")

introduce("Aom", 20)   # ลำดับถูกต้อง: name, age ตรงกับพารามิเตอร์
introduce(20, "Aom")   # ลำดับผิด: สลับตำแหน่งอาร์กิวเมนต์
````

**Output:**
```
ฉันชื่อ Aom อายุ 20 ปี
ฉันชื่อ 20 อายุ Aom ปี
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | หมายเหตุ |
|---|---|---|---|
| 4 | เรียก `introduce("Aom", 20)` | name="Aom", age=20 | อาร์กิวเมนต์ตามตำแหน่งตรงกับความหมายของพารามิเตอร์ |
| 2 | `print(f"...")` | แสดงผล ฉันชื่อ Aom อายุ 20 ปี | ผลลัพธ์ถูกต้องตามที่ตั้งใจ |
| 5 | เรียก `introduce(20, "Aom")` | name=20, age="Aom" | สลับตำแหน่งอาร์กิวเมนต์ Python ผูกค่าตามลำดับโดยไม่ตรวจชนิดข้อมูล |
| 2 | `print(f"...")` | แสดงผล ฉันชื่อ 20 อายุ Aom ปี | ไม่มี error แต่ความหมายผิด เพราะพารามิเตอร์รับค่าผิดตัว |

**คำอธิบายภาพรวม:** บรรทัด 4 เรียกฟังก์ชันด้วยอาร์กิวเมนต์ตามตำแหน่งที่เรียงลำดับถูกต้องตรงกับพารามิเตอร์ `name` และ `age`
บรรทัด 5 สาธิตข้อผิดพลาดที่พบบ่อย คือสลับลำดับอาร์กิวเมนต์ ซึ่ง Python จะไม่แจ้งข้อผิดพลาดใด ๆ เพราะภาษาไม่ได้ตรวจสอบชนิดข้อมูลของพารามิเตอร์
แต่จะผูกค่าตามตำแหน่งเสมอ ผลลัพธ์จึงผิดเพี้ยนไปจากที่ตั้งใจโดยที่โปรแกรมยังทำงานได้ตามปกติ (ไม่มี error หยุดโปรแกรม)

## 4: Return Value กับ Procedure (Void Function)

### แนวคิด

ฟังก์ชันแบ่งออกได้เป็นสองลักษณะตามพฤติกรรมการส่งผลลัพธ์กลับ ลักษณะแรกคือฟังก์ชันที่มี**ค่าที่ส่งกลับ (return value)**
คือฟังก์ชันที่ประมวลผลบางอย่างแล้วส่งผลลัพธ์กลับไปให้จุดที่เรียกใช้ นำไปเก็บในตัวแปรหรือใช้ในนิพจน์ต่อได้ทันที
ทำได้ด้วยคำสั่ง `return` ซึ่งเมื่อถูกรันจะหยุดการทำงานของฟังก์ชันทันทีและส่งค่าที่ระบุกลับไป ลักษณะที่สองคือฟังก์ชันแบบ
**procedure** (บางครั้งเรียกว่า **void function**) คือฟังก์ชันที่ทำงานบางอย่าง เช่น พิมพ์ข้อความออกหน้าจอ แต่ไม่มีคำสั่ง `return`
ที่ส่งค่ากลับมาให้ใช้งานต่อ ฟังก์ชันประเภทนี้ถูกเรียกใช้เพื่อ "ผลข้างเคียง" (side effect) ของมัน เช่นการแสดงผล ไม่ใช่เพื่อค่าที่ได้กลับมา

ในภาษา Python หากฟังก์ชันใดไม่มีคำสั่ง `return` หรือมี `return` โดยไม่ระบุค่า ฟังก์ชันนั้นจะคืนค่าพิเศษที่ชื่อ `None` กลับไปโดยอัตโนมัติเสมอ
ซึ่งเป็นพฤติกรรมสำคัญที่ต้องเข้าใจ เพราะการนำผลลัพธ์จากฟังก์ชันแบบ procedure ไปใช้งานต่อราวกับว่ามันมีค่าที่ส่งกลับ จะได้ `None` แทนค่าที่ต้องการ

### ตัวอย่าง Python

#### ตัวอย่างที่ 4.1 — แนวคิดหลัก: เปรียบเทียบฟังก์ชันที่มี return กับฟังก์ชันแบบ procedure

````python
def add(x, y):
    return x + y

def show_greeting(name):
    print(f"สวัสดี {name}")

total = add(3, 5)
result = show_greeting("Aom")
print(total)
print(result)
````

**Output:**
```
สวัสดี Aom
8
None
```

ส่วนบรรทัด 4–5 เป็นฟังก์ชันแบบ procedure คือ `show_greeting` ที่ทำหน้าที่พิมพ์ข้อความเป็นผลข้างเคียงเท่านั้น ไม่มีคำสั่ง `return`
บรรทัด 7–10 แสดงผลต่างที่ชัดเจน: การเก็บผลจาก `add` ได้ค่า 8 มาใช้งานต่อได้จริง ในขณะที่การเก็บผลจาก `show_greeting`
ได้ค่า `None` เพราะฟังก์ชันที่ไม่มี `return` จะคืนค่า `None` โดยอัตโนมัติเสมอ

#### ตัวอย่างที่ 4.2 — บริบทอื่นที่ต่างโดเมน: return value กับ procedure ในงานประมวลผลลิสต์

**คำอธิบายภาพรวม:** บรรทัด 1–2 เป็นฟังก์ชันที่มีค่าที่ส่งกลับ คือ `find_max` ซึ่งคำนวณค่าสูงสุดของลิสต์แล้วส่งผลลัพธ์กลับให้นำไปใช้ต่อ
บรรทัด 4–6 เป็นฟังก์ชันแบบ procedure ชื่อ `print_list` ที่วนพิมพ์สมาชิกแต่ละตัวในลิสต์เป็นผลข้างเคียง (side effect) ไม่มีคำสั่ง `return`
บรรทัด 8–12 ทดสอบทั้งสองฟังก์ชันกับลิสต์คะแนนเดียวกัน (`scores`) แสดงหลักการเดียวกับตัวอย่างที่ 4.1 แต่อยู่ในบริบทการประมวลผลข้อมูลแบบลิสต์
แทนตัวเลขเดี่ยวและข้อความทักทาย ยืนยันอีกครั้งว่าฟังก์ชันแบบ procedure จะคืนค่า `None` เสมอเมื่อนำไปเก็บไว้ในตัวแปร

#### ตัวอย่างที่ 4.3 — ข้อผิดพลาดที่พบบ่อย: เข้าใจผิดว่าฟังก์ชันแบบ procedure ส่งค่ากลับมาด้วย

````python
def show_greeting(name):
    print(f"สวัสดี {name}")

message = show_greeting("Beam")

try:
    print("ข้อความคือ: " + message)
except TypeError as error:
    print(f"เกิดข้อผิดพลาด: {error}")
````

ที่ไม่มี `return` บรรทัด 6–9 สาธิตข้อผิดพลาดที่พบบ่อยเมื่อผู้เขียนโปรแกรมเข้าใจผิดว่าฟังก์ชันที่พิมพ์ข้อความออกหน้าจอ (`print`)

## 5: ขอบเขตตัวแปร (Scope) — ตัวแปรเฉพาะที่ (Local) กับตัวแปรส่วนกลาง (Global)

### แนวคิด

**ขอบเขต (scope)** ของตัวแปร หมายถึงบริเวณของโปรแกรมที่ตัวแปรนั้นสามารถถูกอ้างถึงและใช้งานได้ ตัวแปรที่ถูกสร้างขึ้นภายในฟังก์ชัน
เรียกว่า **ตัวแปรเฉพาะที่ (local variable)** จะมีขอบเขตจำกัดอยู่เฉพาะภายในฟังก์ชันนั้นเท่านั้น เมื่อฟังก์ชันทำงานเสร็จสิ้น
ตัวแปรเฉพาะที่จะถูกทำลายไป และไม่สามารถอ้างถึงได้จากภายนอกฟังก์ชัน ในทางกลับกัน ตัวแปรที่ถูกสร้างขึ้นภายนอกฟังก์ชันทั้งหมด (ในระดับบนสุดของโปรแกรม)
เรียกว่า **ตัวแปรส่วนกลาง (global variable)** จะมีขอบเขตครอบคลุมทั้งโปรแกรม สามารถอ่านค่าได้จากทุกจุด รวมถึงจากภายในฟังก์ชันต่าง ๆ ด้วย

อย่างไรก็ตาม การ**อ่าน**ค่าตัวแปรส่วนกลางจากภายในฟังก์ชันทำได้โดยตรงโดยไม่ต้องประกาศอะไรเพิ่มเติม แต่การ**แก้ไข**ค่าตัวแปรส่วนกลาง
จากภายในฟังก์ชันนั้น จำเป็นต้องประกาศชื่อตัวแปรด้วยคำสั่งพิเศษก่อน (ในภาษา Python คือคำสั่ง `global`) มิฉะนั้นภาษาโปรแกรมจะตีความว่ากำลังสร้าง
ตัวแปรเฉพาะที่ตัวใหม่ขึ้นมาแทน ซึ่งเป็นแหล่งความเข้าใจผิดที่พบบ่อยที่สุดข้อหนึ่งของผู้เริ่มต้นเขียนโปรแกรม

### ตัวอย่าง Python

#### ตัวอย่างที่ 5.1 — แนวคิดหลัก: อ่านค่า global ได้อิสระ แต่ต้องประกาศ `global` ก่อนจึงจะแก้ไขได้

````python
balance = 100

def deposit(amount):
    global balance
    balance = balance + amount

def show_balance():
    print("ยอดคงเหลือ:", balance)

show_balance()
deposit(50)
show_balance()
````

**Output:**
```
ยอดคงเหลือ: 100
ยอดคงเหลือ: 150
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | หมายเหตุ |
|---|---|---|---|
| 1 | `balance = 100` | balance=100 (global) | สร้างตัวแปรส่วนกลาง |
| 10 | เรียก `show_balance()` ครั้งที่ 1 | อ่านค่า balance จาก global scope | ฟังก์ชันอ่านตัวแปรส่วนกลางได้โดยไม่ต้องประกาศ global |
| 8 | `print("ยอดคงเหลือ:", balance)` | แสดงผล ยอดคงเหลือ: 100 | ค่าปัจจุบันของ balance |
| 11 | เรียก `deposit(50)` | amount=50 | เข้าฟังก์ชัน deposit |
| 4 | `global balance` | ประกาศว่าจะใช้ตัวแปร balance จาก global scope | หากไม่มีบรรทัดนี้ บรรทัดถัดไปจะสร้างตัวแปร local ใหม่แทน (ดูตัวอย่างที่ 5.3) |
| 5 | `balance = balance + amount` | balance=150 (global ถูกแก้ไข) | แก้ไขค่าตัวแปรส่วนกลางจริง เพราะมี global กำกับไว้ |
| 12 | เรียก `show_balance()` ครั้งที่ 2 | อ่านค่า balance ล่าสุด | ยืนยันว่าค่า global ถูกเปลี่ยนแปลงจริง |
| 8 | `print(...)` | แสดงผล ยอดคงเหลือ: 150 | ค่าที่เปลี่ยนแปลงคงอยู่ข้ามการเรียกฟังก์ชัน |

**คำอธิบายภาพรวม:** บรรทัด 1 สร้างตัวแปรส่วนกลาง `balance` บรรทัด 7–8 แสดงว่าฟังก์ชัน `show_balance` สามารถอ่านค่าตัวแปรส่วนกลางได้โดยตรง
โดยไม่ต้องประกาศคำสั่งใดเพิ่มเติม ส่วนบรรทัด 3–5 แสดงกรณีที่ต้องการแก้ไขค่าตัวแปรส่วนกลางจากภายในฟังก์ชัน ซึ่งจำเป็นต้องประกาศ `global balance`
ก่อนเสมอ มิฉะนั้น Python จะตีความว่า `balance` ในบรรทัด 5 เป็นตัวแปรเฉพาะที่ตัวใหม่ บรรทัด 10–12 ยืนยันว่าการแก้ไขผ่าน `deposit`
ส่งผลต่อค่า `balance` ในขอบเขตส่วนกลางจริง เพราะการเรียก `show_balance` ครั้งที่สองพิมพ์ค่าที่เปลี่ยนแปลงแล้ว

#### ตัวอย่างที่ 5.2 — บริบทอื่นที่ต่างโดเมน: scope ของตัวแปรข้อความในระบบสถานะเกม

````python
def update_status(new_status):
    global game_status
    game_status = new_status

def show_status():
    print("สถานะเกม:", game_status)

game_status = "กำลังเล่น"
show_status()
update_status("จบเกม")
show_status()
````

**Output:**
```
สถานะเกม: กำลังเล่น
สถานะเกม: จบเกม
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | หมายเหตุ |
|---|---|---|---|
| 8 | `game_status = "กำลังเล่น"` | game_status="กำลังเล่น" (global) | สร้างตัวแปรส่วนกลาง |
| 9 | เรียก `show_status()` ครั้งที่ 1 | อ่านค่า game_status จาก global scope | ฟังก์ชันอ่านตัวแปรส่วนกลางได้โดยไม่ต้องประกาศ global |
| 6 | `print("สถานะเกม:", game_status)` | แสดงผล สถานะเกม: กำลังเล่น | ค่าปัจจุบันของ game_status |
| 10 | เรียก `update_status("จบเกม")` | new_status="จบเกม" | เข้าฟังก์ชัน |
| 2 | `global game_status` | ประกาศว่าจะใช้ตัวแปร game_status จาก global scope | เช่นเดียวกับหลักการในตัวอย่างที่ 5.1 |
| 3 | `game_status = new_status` | game_status="จบเกม" (global ถูกแก้ไข) | แก้ไขค่าตัวแปรส่วนกลางจริง เพราะมี global กำกับไว้ |
| 11 | เรียก `show_status()` ครั้งที่ 2 | อ่านค่า game_status ล่าสุด | ยืนยันว่าค่า global ถูกเปลี่ยนแปลงจริง |
| 6 | `print(...)` | แสดงผล สถานะเกม: จบเกม | ค่าที่เปลี่ยนแปลงคงอยู่ข้ามการเรียกฟังก์ชัน |

**คำอธิบายภาพรวม:** บรรทัด 8 สร้างตัวแปรส่วนกลาง `game_status` เป็นข้อความ (string) แทนที่จะเป็นตัวเลขอย่างตัวอย่างที่ 5.1 (`balance`)
บรรทัด 5–6 แสดงว่าฟังก์ชัน `show_status` อ่านค่าตัวแปรส่วนกลางได้โดยตรง ส่วนบรรทัด 1–3 แสดงการแก้ไขค่าตัวแปรส่วนกลางจากภายในฟังก์ชัน
โดยต้องประกาศ `global game_status` ก่อนเสมอ เช่นเดียวกับหลักการในตัวอย่างที่ 5.1 บรรทัด 9–11 ยืนยันว่าการเรียก `update_status`
ส่งผลต่อค่า `game_status` ในขอบเขตส่วนกลางจริง แสดงให้เห็นว่าแนวคิดเรื่อง scope ใช้ได้กับตัวแปรชนิดข้อความเช่นเดียวกับตัวเลข
ไม่ได้จำกัดอยู่เฉพาะโดเมนการเงินเท่านั้น

#### ตัวอย่างที่ 5.3 — ข้อผิดพลาดที่พบบ่อย: ลืมประกาศ `global` ก่อนแก้ไขตัวแปรส่วนกลาง

````python
balance = 100

def deposit_wrong(amount):
    balance = balance + amount  # ไม่ได้ประกาศ global ไว้ก่อน
    return balance

try:
    deposit_wrong(50)
except UnboundLocalError as error:
    print(f"เกิดข้อผิดพลาด: {error}")

print(balance)
````

**Output:**
```
เกิดข้อผิดพลาด: cannot access local variable 'balance' where it is not associated with a value
100
```

**ตารางการเปลี่ยนแปลงค่า:**

| บรรทัดที่ | คำสั่งที่ทำงาน | ตัวแปร/ค่า | หมายเหตุ |
|---|---|---|---|
| 1 | `balance = 100` | balance=100 (global) | ตัวแปรส่วนกลาง |
| 8 | เรียก `deposit_wrong(50)` | amount=50 | เข้าฟังก์ชัน |
| 4 | `balance = balance + amount` | เกิด UnboundLocalError | มีการกำหนดค่า (assign) ให้ `balance` ภายในฟังก์ชันบรรทัด 4 ทำให้ Python ถือว่า `balance` เป็นตัวแปรเฉพาะที่ตลอดทั้งฟังก์ชัน แม้แต่ตอนอ่านค่าฝั่งขวาของเครื่องหมาย `=` ก็ยังไม่มีค่าให้อ่าน |
| 9–10 | `except UnboundLocalError as error` | แสดงผล เกิดข้อผิดพลาด: cannot access local variable 'balance' where it is not associated with a value | ดักข้อผิดพลาดไว้แทนที่จะปล่อยให้โปรแกรมหยุดทำงานกะทันหัน |
| 12 | `print(balance)` | แสดงผล 100 | ตัวแปร global balance ไม่ถูกแก้ไข เพราะฟังก์ชันไม่เคยไปถึงบรรทัด return |

**คำอธิบายภาพรวม:** บรรทัด 3–5 นิยามฟังก์ชัน `deposit_wrong` ที่ตั้งใจจะนำค่าตัวแปรส่วนกลาง `balance` มาบวกกับ `amount`
แล้วเก็บกลับที่ `balance` เดิม แต่เนื่องจากไม่ได้ประกาศ `global balance` ก่อน (ต่างจากตัวอย่างที่ 5.1) Python จะตีความว่าบรรทัด 4
กำลังสร้างตัวแปรเฉพาะที่ชื่อ `balance` ขึ้นใหม่ทั้งฟังก์ชัน ทำให้เมื่ออ่านค่า `balance` ทางฝั่งขวาของนิพจน์ (ก่อนที่จะมีการกำหนดค่าจริง)
เกิด `UnboundLocalError` ทันที บรรทัด 7–10 ดักข้อผิดพลาดนี้ไว้เพื่อแสดงข้อความอธิบาย และบรรทัด 12 ยืนยันว่าตัวแปร `balance` ระดับ global
ยังคงมีค่า 100 เท่าเดิม เพราะฟังก์ชันไม่สามารถทำงานจนจบได้ ตัวอย่างนี้เป็นข้อผิดพลาดด้าน scope ที่พบบ่อยที่สุดข้อหนึ่งของผู้เริ่มต้นเขียนโปรแกรม

> **หมายเหตุเรื่องเวอร์ชัน Python:** ข้อความ error ข้างต้น (`cannot access local variable ... where it is not associated with a value`)
> เป็นรูปแบบข้อความของ Python 3.11 ขึ้นไป (ยืนยันด้วยการรันจริงบน Python 3.11.15) หากรันด้วย Python เวอร์ชันเก่ากว่า 3.11
> ข้อความจะแตกต่างออกไปเล็กน้อยเป็น `local variable 'balance' referenced before assignment` แต่ประเภทข้อผิดพลาด (`UnboundLocalError`)
> และสาเหตุยังคงเหมือนเดิม

### 5: ขอบเขตตัวแปร (Scope) (ข้อ 17–20)

**ข้อ 17 (เขียนอธิบายแนวคิด):** เพราะเหตุใดการ**แก้ไข**ค่าตัวแปรส่วนกลาง (global variable) จากภายในฟังก์ชันจึงต้องประกาศคำสั่ง `global` ก่อนเสมอ
