**เฉลยข้อ 16:**

````python
def count_vowels(text):
    vowels = "aeiou"
    count = 0
    for ch in text:
        if ch in vowels:
            count += 1
    return count

print(count_vowels("education"))
````

**Output (รันจริง):**
```
5
```

**เฉลยข้อ 17:** Python อนุญาตให้ฟังก์ชัน**อ่าน**ค่าตัวแปรส่วนกลางได้โดยตรงโดยไม่ต้องประกาศเพิ่มเติม เพราะการอ่านค่าไม่มีความกำกวมว่าเป็นตัวแปรใด

เพื่อป้องกันไม่ให้ฟังก์ชันแก้ไขตัวแปรส่วนกลางโดยไม่ตั้งใจ ดังนั้นหากต้องการแก้ไขค่าตัวแปรส่วนกลางจริง ๆ จึงต้องประกาศ `global` ก่อน
เพื่อบอก Python อย่างชัดเจนว่าชื่อตัวแปรนี้ให้อ้างอิงไปยังตัวแปรส่วนกลาง ไม่ใช่สร้างตัวแปรเฉพาะที่ใหม่

**เฉลยข้อ 19:** โค้ดผิดที่บรรทัด 4 (`counter = counter + 1`) ซึ่งกำหนดค่าให้ `counter` ภายในฟังก์ชัน `increment` โดยไม่ได้ประกาศ `global counter` ก่อน
ทำให้ Python ตีความว่า `counter` เป็นตัวแปรเฉพาะที่ตลอดทั้งฟังก์ชัน เมื่อพยายามอ่านค่า `counter` ทางฝั่งขวาของ `=` (ก่อนที่จะมีการกำหนดค่าจริง)
จึงเกิดข้อผิดพลาด:
```
UnboundLocalError: cannot access local variable 'counter' where it is not associated with a value
```
วิธีแก้คือเพิ่มบรรทัด `global counter` ไว้ก่อนบรรทัด 4 เมื่อแก้ไขแล้วจะรันได้ผลลัพธ์:
```
1
```

**เฉลยข้อ 20:**

````python
score = 0

def add_point(points):
    global score
    score = score + points

add_point(15)
add_point(20)
print(score)
````

**Output (รันจริง):**
```
35
```

- **Step 9 — ตัวอย่างที่ 3 ต่อหัวข้อ:** ✅ เสร็จสมบูรณ์ ทั้ง 5 หัวข้อมีครบ 3 ตัวอย่างแล้ว (รวมเพิ่มขึ้นจาก 10 เป็น 15 ตัวอย่างทั้งไฟล์)
  ตัวอย่างใหม่ที่เพิ่มเข้าไปแต่ละหัวข้อ (ตัวอย่างที่ X.2 ในลำดับใหม่) อยู่คนละโดเมนปัญหากับตัวอย่างเดิมสองอันเสมอ (เช่น หัวข้อ 1 เดิมเป็นโดเมนตัวเลข
  ล้วน — เรขาคณิตและอุณหภูมิ — ตัวอย่างใหม่จึงเปลี่ยนไปใช้โดเมนข้อความ/การจัดรูปแบบใบเสร็จแทน) ทุกตัวอย่างมีครบ 4 องค์ประกอบ (โค้ดเลขบรรทัด + output
  ที่รันจริง + trace table + คำอธิบายภาพรวม) และรันจริงยืนยัน output แล้วทุกตัวอย่าง (0 ตัวอย่างที่เดา output)
- **Step 11 — แบบฝึกหัดท้ายบท:** ✅ เสร็จสมบูรณ์ เพิ่มหมวด "แบบฝึกหัดท้ายบท" จำนวน 20 ข้อ (≥20 ตามเกณฑ์) ครอบคลุมทั้ง 5 หัวข้ออย่างทั่วถึง
  หัวข้อละ 4 ข้อเท่ากันทุกหัวข้อ ผสม 4 ประเภทคำถาม (เขียนอธิบายแนวคิด 5 ข้อ, ทายผลลัพธ์ 5 ข้อ, หาข้อผิดพลาด 5 ข้อ, เขียนโค้ด 5 ข้อ) มีหมวด
  "เฉลยแบบฝึกหัด" แยกต่างหากท้ายไฟล์ ข้อทายผลลัพธ์และข้อเขียนโค้ดทุกข้อ (รวมทั้งเฉลยของข้อหาข้อผิดพลาดที่มีการรันโค้ดทั้งฉบับผิดและฉบับแก้ไขแล้ว)
  ผ่านการรันจริงด้วย Python 3.11.15 ยืนยัน output ก่อนบันทึกลงเอกสารทุกข้อ ไม่มีข้อใดเดา output
